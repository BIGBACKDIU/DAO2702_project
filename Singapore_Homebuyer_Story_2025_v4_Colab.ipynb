# Upgrading from HDB to a private condo: a data guide for a family in Tampines

### Who this guide is for
A Singaporean family of four, both parents 38, in a **5-room HDB flat in Tampines** that has passed its minimum occupation period. They want a condo of **about 1,050 sqft** (a size proxy; bedroom count and layout are not recorded) in the east, to stay 10+ years. All of the figures below are assumptions for illustration; edit the persona cell to try your own.
- **Income:** S\$16,000 a month combined (gross, excluding employer CPF), about 1.5 times the median employed household (S\$10,591, SingStat 2025).
- **Their flat:** worth about S\$795,000 (Huttons 4Q 2025 average for a 5-room), with S\$150,000 of HDB loan left.
- **Proceeds:** after an assumed S\$20,000 of agent, legal and moving costs, selling frees about **S\$625,000**. Of that, S\$350,000 is an assumed CPF refund and about S\$275,000 is cash.
- **Budget: about S\$2.18M.** It's limited by those funds, which must cover the down payment and stamp duty. The loan limit (TDSR, 55% of income at a 4% test rate) would allow about S\$2.32M.

### The answer in brief
1. **Sell the flat first, then buy.** Buying first would need about S\$1.27M upfront, before the flat's money arrives.
2. **Look at resales with 85–88 years of lease left.** The District 18 (Tampines/Pasir Ris) 2025 median is S\$1.62M, leaving about S\$170k of assumed funds after upfront costs; only 10 of 21 comparable sales are actually in Tampines. The provisional 2026-Q3 index proxy is about S\$1.65M. Units with 89–93 years left in the same area cost about S\$0.3M more. The trade-off: most of the cheaper ones are a 20-minute walk from the MRT, and homes within 5 minutes of an MRT entrance show a 17% higher per-sqft price than those over 15 minutes away after controls in 2025 (about +7–11% in 2020–24, Step 7).
3. **Choose by readiness, not a predicted dip.** In the 2025 illustration, lower rates reduce payments while the assessed borrowing limit stays unchanged. In our own 2020–25 sales the top-up to route B widened from about S\$0.62M to S\$0.85M, even though their flat rose faster in percentage terms; future prices are uncertain.
4. **Five years add context (Steps 7–8):** projects with 85–88 years left showed no clear shortfall against longer-lease ones, and the east was among the fastest-rising areas. The MRT gap appears every year (about +7% to +17%), but no clear extra rise was detected when TEL4 opened.

Each step below shows the analysis behind one of these. Steps 1–6 use 24,151 private home sales from 2025; Steps 7–8 use a 106,449-sale model-ready series spanning August 2020 to December 2025, including 2025; official URA, HDB and MAS data fill in the rest. Project transactions are from 2025. An October 2026 decision is also checked against a provisional 2026-Q3 proxy: 2025 comparable prices × 1.021, the change in the URA non-landed index. This is not an observed 2026 project price. Every chart names its source.

**How to read this guide:** blue and red distinguish comparisons; each chart defines their meaning. “Price per sqft” is the transaction price divided by floor area. A 95% confidence interval shows the model’s statistical uncertainty, not the full range of prices a buyer might face. Options A–G are labels for types of homes, not available listings.


**External sources for this section:** [SingStat: household income table](https://tablebuilder.singstat.gov.sg/api/table/tabledata/17896) · [Huttons: 2025 Q4 HDB price report](https://www.huttonsgroup.com/wp-content/uploads/HDB-4Q-2025-Quarterly-Report-Generic.pdf) · [MAS: housing loan limits and TDSR](https://www.mas.gov.sg/publications/macroprudential-policies-in-singapore)

```python
# Google Colab setup: fetch the project datasets from the public GitHub repository.
# This setup cell precedes the current family v4 analysis; running downloads and checks six analysis inputs.
# Open this existing repository in Colab and run this cell first. Preprocessing inputs and the cleaner are also published for a separate audit.
import hashlib
import os
import sys
from pathlib import Path
from urllib.request import urlopen
import shutil

GITHUB_REPOSITORY = "BIGBACKDIU/DAO2702_project"
GITHUB_BRANCH = "main"
GITHUB_RAW_BASE = f"https://raw.githubusercontent.com/{GITHUB_REPOSITORY}/{GITHUB_BRANCH}/"

DATA_SHA256 = {'properties2025.csv': 'e12f1469f4f71f2bd93f6ab5f31715daadb9de3524e2725a186930577788c6f8', 'properties_2020_2025_clean.csv': 'edede72a9ab8c628d30ce9605dc6a6c7ddb8864f22a734c5a92c78117aacf8b6', 'sg_map.json': '2375b6865eaec2c0a0ce30f2016c667bd97c984cecfe95a1d89d91c6cac46d29', 'sg_mrt.json': 'ffb34541f6b5f9071b04911eb5ac9428d5cf4e758519a30288bc1729e61ce647', 'project_coords.csv': '2999c492c8091d2baef9b37595d283ad0a41b539eb6e9edb2a523ef9b8b14bf4', 'mrt_station_openings.csv': 'd54b66427766ad640fae264a96b9d1b853e2f5e02d0d40e77d77da522e7fc5a0'}

# Colab's working directory is /content. Existing relative file paths will work here.
if "google.colab" in sys.modules:
    os.chdir("/content")

for filename, expected_sha in DATA_SHA256.items():
    destination = Path(filename)
    if destination.exists() and hashlib.sha256(destination.read_bytes()).hexdigest() == expected_sha:
        print(f"Already verified: {filename}")
        continue

    url = GITHUB_RAW_BASE + filename
    temp_file = destination.with_name(destination.name + ".download")
    try:
        if filename == "properties_2020_2025_clean.csv":
            # The 19 MB source CSV is stored on GitHub in byte-exact parts.
            # Rebuild it locally so all original analysis cells remain unchanged.
            with temp_file.open("wb") as dest:
                for part_num in range(1, 12):
                    part_url = GITHUB_RAW_BASE + f"data/{filename}.part{part_num:02d}"
                    with urlopen(part_url, timeout=90) as response:
                        shutil.copyfileobj(response, dest)
                    print(f"Fetched historical CSV part {part_num}/11")
        else:
            with urlopen(url, timeout=90) as response, temp_file.open("wb") as dest:
                shutil.copyfileobj(response, dest)
        digest = hashlib.sha256(temp_file.read_bytes()).hexdigest()
        if digest != expected_sha:
            raise ValueError(f"SHA-256 mismatch for {filename}: {digest}; expected {expected_sha}")
        temp_file.replace(destination)
        print(f"Downloaded and verified: {filename}")
    except Exception as exc:
        temp_file.unlink(missing_ok=True)
        raise RuntimeError(
            f"Failed to download {filename} from {url}. "
            "Make sure the public GitHub repository exists, all six datasets "
            "are uploaded to the main branch, and their original bytes are unchanged."
        ) from exc

print("All six analysis inputs are ready. Run the remaining notebook cells below.")
```

```python
# Requirements: pandas, numpy, scipy, matplotlib. Put properties2025.csv, properties_2020_2025_clean.csv (made by clean_2020_2025.py from
# properties23.csv, properties2024.csv and properties2025.csv), sg_map.json, sg_mrt.json, project_coords.csv and mrt_station_openings.csv
# in the same folder as this notebook.
import warnings, hashlib, json, os
import numpy as np, pandas as pd, scipy.stats as st
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon, PathPatch
from matplotlib.path import Path
from matplotlib.colors import LinearSegmentedColormap, TwoSlopeNorm
warnings.filterwarnings('ignore')
import logging; logging.getLogger('matplotlib.font_manager').setLevel(logging.ERROR)
pd.set_option('display.width', 170); pd.set_option('display.max_columns', 30)

def ols(y, X, names=None, cluster=None, const=True):
    """OLS with one-way cluster-robust SEs. Returns (table with % effects and 95% CIs, extras)."""
    X = np.asarray(X, float); y = np.asarray(y, float)
    names = list(names) if names is not None else [f'x{i}' for i in range(X.shape[1])]
    if const: X = np.column_stack([np.ones(len(X)), X]); names = ['const'] + names
    n, k = X.shape; XtXi = np.linalg.pinv(X.T @ X); b = XtXi @ X.T @ y; e = y - X @ b
    cl = pd.factorize(np.asarray(cluster))[0]; G = cl.max() + 1
    S = np.zeros((G, k)); np.add.at(S, cl, X * e[:, None])
    V = XtXi @ (S.T @ S) @ XtXi * G / (G - 1) * (n - 1) / (n - k)
    se = np.sqrt(np.clip(np.diag(V), 0, None)); tc = st.t.ppf(0.975, G - 1)
    tab = pd.DataFrame({'coef': b, 'se': se, 'p': 2 * st.t.sf(np.abs(b / se), G - 1)}, index=names)
    tab['pct'] = np.expm1(b) * 100; tab['pct_lo'] = np.expm1(b - tc * se) * 100; tab['pct_hi'] = np.expm1(b + tc * se) * 100
    r2 = 1 - (e @ e) / ((y - y.mean()) @ (y - y.mean()))
    return tab, dict(b=b, V=V, names=names, e=e, r2=r2, G=G)

def dummies(s, prefix, base):
    return pd.get_dummies(s, prefix=prefix).astype(float).drop(columns=f'{prefix}_{base}')
def sgd(x): return f'S${x:,.0f}'
from decimal import Decimal, ROUND_HALF_UP
def sgm(x): return 'S$' + str((Decimal(str(float(x))) / Decimal('1000000')).quantize(Decimal('0.01'), rounding=ROUND_HALF_UP)) + 'M'

CSV_SHA256 = 'e12f1469f4f71f2bd93f6ab5f31715daadb9de3524e2725a186930577788c6f8'   # the file all results below were computed from
assert hashlib.sha256(open('properties2025.csv', 'rb').read()).hexdigest() == CSV_SHA256, \
    'properties2025.csv differs from the file used for this notebook; numbers in the text may not match'
df = pd.read_csv('properties2025.csv')
df['tenure'] = df.tenure.replace({'freenhold': 'Freehold', 'other': '999-yr/other', '99 years': '99-yr'})
df['psf'] = df.price / df.area; df['lpsf'] = np.log(df.psf); df['larea'] = np.log(df.area)
df['month'] = pd.to_datetime(df.date, format='%b-%y').dt.month
df['floor'] = df.level.str.extract(r'(\d+) to')[0].astype(float)
df.loc[df.level.str.startswith('B'), 'floor'] = 0
df['lease_used'] = np.where(df.tenure == '99-yr', 2025 - df.commence, 0.0)    # commence = lease start year
CENT = {1:(1.2805,103.8520),2:(1.2765,103.8430),3:(1.2880,103.8100),4:(1.2650,103.8200),5:(1.3000,103.7700),6:(1.2920,103.8500),
        7:(1.3000,103.8580),8:(1.3110,103.8560),9:(1.3020,103.8350),10:(1.3150,103.8050),11:(1.3220,103.8420),12:(1.3300,103.8560),
        13:(1.3360,103.8800),14:(1.3180,103.8950),15:(1.3050,103.9050),16:(1.3240,103.9400),17:(1.3600,103.9650),18:(1.3550,103.9450),
        19:(1.3650,103.8900),20:(1.3600,103.8450),21:(1.3400,103.7750),22:(1.3400,103.7200),23:(1.3700,103.7600),25:(1.4350,103.7850),
        26:(1.3900,103.8200),27:(1.4300,103.8350),28:(1.3950,103.8700)}
def km_to_cbd(dist):
    la, lo = np.radians(CENT[dist][0]), np.radians(CENT[dist][1]); la0, lo0 = np.radians(1.2840), np.radians(103.8515)
    h = np.sin((la - la0) / 2) ** 2 + np.cos(la) * np.cos(la0) * np.sin((lo - lo0) / 2) ** 2
    return 6371 * 2 * np.arcsin(np.sqrt(h))
df['km'] = df.district.map(km_to_cbd)
AREA_NAME = {1: 'Raffles Place', 2: 'Tanjong Pagar', 3: 'Queenstown', 4: 'Sentosa/Telok Blangah', 5: 'Clementi/West Coast', 6: 'City Hall',
             7: 'Bugis', 8: 'Little India', 9: 'Orchard/River Valley', 10: 'Bukit Timah/Holland', 11: 'Newton/Novena', 12: 'Balestier/Toa Payoh',
             13: 'Macpherson/Potong Pasir', 14: 'Geylang/Eunos', 15: 'Katong/Marine Parade', 16: 'Bedok/Upper East Coast', 17: 'Changi/Loyang',
             18: 'Tampines/Pasir Ris', 19: 'Serangoon/Hougang/Sengkang/Punggol', 20: 'Bishan/Ang Mo Kio', 21: 'Upper Bukit Timah', 22: 'Jurong/Boon Lay',
             23: 'Bukit Batok/Bukit Panjang', 25: 'Woodlands/Kranji', 26: 'Upper Thomson', 27: 'Yishun/Sembawang', 28: 'Seletar/Yio Chu Kang'}

# ---- one chart style for every figure: white card, Inter (or Helvetica/Arial), takeaway title, source line ----
C = dict(blue='#3568AE', blue_d='#214875', blue_l='#DFE9F6', orange='#BD4D61', orange_l='#F5E2E6', aqua='#496B8B', yellow='#BD4D61',
         violet='#6F7F91', red='#BD4D61', grey='#8a939c', grey_l='#d5d9dd', ink='#141a20', ink2='#4f5862', grid='#eceef0', bg='#ffffff')
plt.rcParams.update({'font.family': ['Inter', 'Helvetica Neue', 'Helvetica', 'Arial', 'DejaVu Sans'], 'font.size': 10,
                     'figure.facecolor': C['bg'], 'axes.facecolor': C['bg'], 'savefig.facecolor': C['bg'], 'figure.dpi': 130,
                     'axes.edgecolor': C['grid'], 'axes.labelcolor': C['ink2'], 'axes.labelsize': 9.5, 'axes.grid': True,
                     'grid.color': C['grid'], 'grid.linewidth': 0.8, 'axes.axisbelow': True, 'axes.spines.top': False,
                     'axes.spines.right': False, 'axes.spines.left': False, 'xtick.color': C['ink2'], 'ytick.color': C['ink2'],
                     'xtick.labelsize': 9, 'ytick.labelsize': 9, 'xtick.major.size': 0, 'ytick.major.size': 0,
                     'legend.frameon': False, 'legend.fontsize': 9, 'text.parse_math': False})   # so 'S$' isn't read as maths
def finish(fig, title, subtitle=None, source='2025 private home sales (properties2025.csv)'):
    """Takeaway title + subtitle at the top, a source line at the bottom, layout fitted between them."""
    import textwrap
    h, w = fig.get_figheight(), fig.get_figwidth()
    title = textwrap.fill(title, int(w * 9.6)); subtitle = textwrap.fill(subtitle, int(w * 13.8)) if subtitle else None
    nt, ns_ = title.count('\n') + 1, (subtitle.count('\n') + 1 if subtitle else 0)
    top = 1 - (0.20 + 0.25 * nt + 0.19 * ns_ + (0.06 if subtitle else 0)) / h; bottom = (0.48 / h) if source else 0
    fig.tight_layout(rect=[0, bottom, 1, top])
    fig.text(0.012, 1 - 0.12 / h, title, fontsize=13.5, fontweight='bold', color=C['ink'], va='top', ha='left')
    if subtitle: fig.text(0.012, 1 - (0.17 + 0.25 * nt) / h, subtitle, fontsize=10, color=C['ink2'], va='top', ha='left', linespacing=1.3)
    if source: fig.text(0.012, 0.07 / h, textwrap.fill(f'Source: {source}', int(w * 17)), fontsize=7.8, color=C['grey'], va='bottom', ha='left')
    plt.show()

SGMAP = json.load(open('sg_map.json')) if os.path.exists('sg_map.json') else None
print(f'{len(df):,} transactions loaded; map file {"found" if SGMAP else "missing (maps will be skipped)"}')

# ---- picture-style chart helpers (a condo block, MRT walking rings, lease batteries, towers) ----
from matplotlib.patches import Rectangle, FancyBboxPatch, Circle, Wedge
def sg_(x, d=0):
    v = round(float(x), d); v = 0.0 if v == 0 else v
    return f'{v:+.{d}f}'
SEQ_BLUE = LinearSegmentedColormap.from_list('seq_blue', ['#F1F5FA', '#B2C9E4', C['blue'], C['blue_d']])
DIV_LEASE = LinearSegmentedColormap.from_list('div_lease', [C['orange'], '#E8B9C3', '#eef0f2', '#B2C9E4', C['blue_d']])

def draw_building(ax, bands, est, lo, hi, base_price=None, badges=None, title=None):
    """Side view of a condo block: one storey per floor band (lowest band = reference), shaded by its price premium."""
    n, W, H = len(bands), 4.0, 1.0
    vmax = max(max(est), 1e-9)
    ax.add_patch(Rectangle((-0.9, -0.32), W + 1.8, 0.32, color=C['grey_l'], lw=0))                  # ground
    ax.add_patch(Rectangle((W / 2 - 0.45, -0.02), 0.9, 0.62, color=C['ink2'], lw=0, zorder=4))       # lobby door
    for i, (b, e, l, h) in enumerate(zip(bands, est, lo, hi)):
        y = i * H; share = max(e, 0) / vmax
        ax.add_patch(Rectangle((0, y), W, H, facecolor=SEQ_BLUE(0.06 + 0.9 * share), edgecolor='white', lw=2.2, zorder=2))
        for wx in np.linspace(0.3, W - 0.72, 5):                                                     # windows
            if i == 0 and abs(wx + 0.21 - W / 2) < 0.6: continue
            ax.add_patch(Rectangle((wx, y + 0.3), 0.42, 0.42, facecolor='white', alpha=0.75 if share < 0.6 else 0.45, lw=0, zorder=3))
        ax.text(-0.18, y + H / 2, 'Floors ' + b.replace(' to ', '–'), ha='right', va='center', fontsize=9, color=C['ink2'])
        if badges and badges[i]:
            num, colr = badges[i]
            ax.add_patch(Circle((-2.95, y + H / 2), 0.27, color=colr, zorder=5))
            ax.text(-2.95, y + H / 2, num, ha='center', va='center', fontsize=9, color='white', fontweight='bold', zorder=6)
        if i == 0:
            ax.text(W + 0.3, y + H / 2, 'reference', ha='left', va='center', fontsize=10, color=C['ink2'])
            continue
        ax.text(W + 0.3, y + H * 0.62, f'{e:+.1f}%', ha='left', va='center', fontsize=11.5, fontweight='bold', color=C['ink'])
        ax.text(W + 0.3, y + H * 0.24, f'95% CI {sg_(l, 1)} to {sg_(h, 1)}', ha='left', va='center', fontsize=7.6, color=C['grey'])
        if base_price:
            ax.text(W + 2.45, y + H * 0.62, f'≈ +S${base_price * e / 100 / 1e3:,.0f}k', ha='left', va='center', fontsize=10, color=C['blue_d'], fontweight='bold')
    ax.add_patch(Rectangle((-0.12, n * H), W + 0.24, 0.16, color=C['ink2'], lw=0))                    # roof slab
    ax.add_patch(Rectangle((W * 0.62, n * H + 0.16), 0.9, 0.38, color=C['grey'], lw=0))              # rooftop plant room
    if title: ax.set_title(title, loc='left', fontsize=10, color=C['ink'], fontweight='bold')
    ax.set_xlim(-3.35 if badges else -2.3, W + (4.6 if base_price else 2.9)); ax.set_ylim(-0.45, n * H + 0.75)
    ax.set_aspect('equal'); ax.axis('off')

def draw_rings(ax, rows, label_main, label_alt=None):
    """MRT walking rings: a station in the middle, rings at 5 / 10 / 15 minutes, each labelled with its price gap."""
    radii = [1.55, 2.6, 3.6]; vmax = max(r['pct'] for r in rows)
    ax.add_patch(Circle((0, 0), 4.4, color='#f3f4f6', lw=0, zorder=0))
    for j in range(2, -1, -1):
        r = rows[j]
        ax.add_patch(Circle((0, 0), radii[j], facecolor=SEQ_BLUE(0.1 + 0.85 * max(r['pct'], 0) / vmax), edgecolor='white', lw=2.5, zorder=1 + (2 - j)))
    ax.add_patch(FancyBboxPatch((-0.36, -0.26), 0.72, 0.52, boxstyle='round,pad=0.05,rounding_size=0.12', color=C['ink'], zorder=6))
    ax.text(0, 0, 'MRT', ha='center', va='center', color='white', fontsize=9, fontweight='bold', zorder=7)
    for j, r in enumerate(rows):                                                 # callouts to the right of the rings
        rm = (0 if j == 0 else radii[j - 1]) * 0.5 + radii[j] * 0.5 if j else 0.95
        ang = np.radians([25, 0, -25][j]); px, py = rm * np.cos(ang), rm * np.sin(ang); ty = [2.3, 0.1, -2.1][j]
        ax.plot([px, 4.7], [py, ty], color=C['ink2'], lw=0.9, zorder=9); ax.add_patch(Circle((px, py), 0.08, color=C['ink'], zorder=10))
        ax.text(4.85, ty + 0.28, ['Walk 5 min or less', 'Walk 5-10 min', 'Walk 10-15 min'][j], ha='left', va='center', fontsize=8.6, color=C['ink2'])
        ax.text(4.85, ty - 0.2, f"{r['pct']:+.1f}%", ha='left', va='center', fontsize=14, fontweight='bold', color=C['ink'])
        ax.text(4.85, ty - 0.68, f"95% CI {sg_(r['lo'])} to {sg_(r['hi'])}", ha='left', va='center', fontsize=8, color=C['grey'])
    for j, lab in enumerate(['5 min', '10 min', '15 min']):
        ax.text(0, radii[j] - 0.13, lab, ha='center', va='top', fontsize=7.8, color='white' if max(rows[j]['pct'], 0) / vmax > 0.55 else C['ink2'], zorder=8)
    ax.text(0, 4.02, 'over 15 min = baseline', ha='center', va='center', fontsize=8, color=C['grey'], zorder=8)
    if label_alt:
        ax.text(-4.4, -4.6, label_alt, ha='left', va='top', fontsize=8, color=C['ink2'])
    ax.set_title(label_main, loc='left', fontsize=10, color=C['ink'])
    ax.set_xlim(-4.5, 9.2); ax.set_ylim(-5.4, 4.5); ax.set_aspect('equal'); ax.axis('off')

def draw_tower(ax, x, h, color, label, width=0.62, floors=None):
    """A simple tower whose height is a price: windows every storey, value on top."""
    ax.add_patch(Rectangle((x - width / 2, 0), width, h, facecolor=color, edgecolor='white', lw=1.5, zorder=2))
    step = floors or max(h / 9, 0.08)
    for yy in np.arange(step * 0.5, h - step * 0.4, step):
        for dx in (-0.16, 0.0, 0.16):
            ax.add_patch(Rectangle((x + dx - 0.045, yy), 0.09, step * 0.42, facecolor='white', alpha=0.55, lw=0, zorder=3))
    ax.text(x, h + 0.03, label, ha='center', va='bottom', fontsize=9, fontweight='bold', color=C['ink'])

def draw_battery(ax, x, frac, color, top, top_sub=None, bottom=None, bottom_sub=None, pin=None, BH=2.4, BW=0.56):
    """A battery whose charge is a share of a 99-year lease; labels above (the finding) and below (the band)."""
    ax.add_patch(FancyBboxPatch((x - BW / 2, 0), BW, BH, boxstyle='round,pad=0,rounding_size=0.06', facecolor='white', edgecolor=C['ink2'], lw=1.6, zorder=2))
    ax.add_patch(Rectangle((x - 0.09, BH), 0.18, 0.12, color=C['ink2'], lw=0, zorder=2))
    ax.add_patch(Rectangle((x - BW / 2 + 0.06, 0.06), BW - 0.12, (BH - 0.12) * min(frac, 1), color=color, lw=0, zorder=3))
    ax.text(x, BH + 0.62, top, ha='center', va='bottom', fontsize=12 if top != 'reference' else 9.5, fontweight='bold' if top != 'reference' else 'normal',
            color=C['ink'] if top != 'reference' else C['ink2'])
    if top_sub: ax.text(x, BH + 0.4, top_sub, ha='center', va='bottom', fontsize=7.4, color=C['grey'])
    if bottom: ax.text(x, -0.22, bottom, ha='center', va='top', fontsize=9.5, color=C['ink'], fontweight='bold')
    if bottom_sub: ax.text(x, -0.6, bottom_sub, ha='center', va='top', fontsize=8, color=C['ink2'])
    if pin: ax.text(x, BH + 1.25, pin, ha='center', va='bottom', fontsize=9, color='white', fontweight='bold', bbox=dict(boxstyle='round,pad=0.3', fc=C['orange'], ec='none'))

def draw_heat(ax, rows, cols, cell, col_notes=None, vmax=None):
    """Tiles: one row per walking band, one column per year; cell(row, col) -> (pct, lo, hi) or None."""
    vals = [cell(r, c) for r in rows for c in cols]; vmax = vmax or max(v[0] for v in vals if v)
    for r_i, (rk, rlab) in enumerate(rows):
        yb = len(rows) - 1 - r_i
        ax.text(-0.62, yb, rlab, ha='right', va='center', fontsize=10, fontweight='bold', color=C['ink'])
        for c_i, ck in enumerate(cols):
            v = cell((rk, rlab), ck)
            if v is None: continue
            pct, lo, hi = v; sh = min(max(pct, 0) / vmax, 1)
            ax.add_patch(FancyBboxPatch((c_i - 0.45, yb - 0.42), 0.9, 0.84, boxstyle='round,pad=0,rounding_size=0.08', color=SEQ_BLUE(0.08 + 0.88 * sh) if pct >= 0 else '#F5E2E6', lw=0))
            tc = 'white' if sh > 0.55 else C['ink']
            ax.text(c_i, yb + 0.1, f'{sg_(pct)}%', ha='center', va='center', fontsize=14, fontweight='bold', color=tc)
            ax.text(c_i, yb - 0.22, f'CI {sg_(lo)} to {sg_(hi)}', ha='center', va='center', fontsize=7.6, color=tc)
    top = len(rows) - 1 + 0.86
    for c_i, ck in enumerate(cols):
        ax.text(c_i, top, f'{ck}', ha='center', va='center', fontsize=10.5, fontweight='bold', color=C['ink'])
        if col_notes and col_notes.get(ck): ax.text(c_i, top - 0.24, col_notes[ck], ha='center', va='center', fontsize=7.5, color=C['grey'])
    ax.set_xlim(-1.9, len(cols) - 0.5); ax.set_ylim(-0.55, top + 0.2); ax.axis('off')

def draw_event(ax, ks, pct, lo, hi, labels, open_after, base_k, open_label='Stations open'):
    """Before/after picture: shaded 'before' and 'after' zones, the 95% range as a soft band, estimates as dots, zero = no difference."""
    ks = np.asarray(ks, float); pct, lo, hi = map(np.asarray, (pct, lo, hi))
    x0, x1 = ks.min() - 0.6, ks.max() + 0.6
    ax.axvspan(x0, open_after, color='#f3f4f6', lw=0, zorder=0); ax.axvspan(open_after, x1, color='#EDF3F9', lw=0, zorder=0)
    ax.fill_between(ks, lo, hi, color=C['blue'], alpha=0.15, lw=0)
    ax.plot(ks, pct, color=C['blue'], lw=2, zorder=3); ax.scatter(ks, pct, color=C['blue'], s=46, zorder=4, edgecolor='white', linewidth=1.5)
    ax.axhline(0, color=C['ink'], lw=1.1); ax.axvline(open_after, color=C['aqua'], lw=1.6)
    lim = max(np.abs(lo).max(), np.abs(hi).max()) * 1.35
    ax.set_ylim(-lim, lim); ax.set_xlim(x0, x1)
    ax.text(x0 + 0.1, lim * 0.88, 'BEFORE', ha='left', va='top', fontsize=10, fontweight='bold', color=C['ink2'])
    ax.text(open_after + 0.1, lim * 0.88, f'AFTER · {open_label}', ha='left', va='top', fontsize=10, fontweight='bold', color=C['aqua'])
    ax.text(x1 - 0.05, 0, 'no difference ', ha='right', va='bottom', fontsize=8, color=C['ink'])
    ax.annotate('reference', (base_k, 0), xytext=(0, -16), textcoords='offset points', ha='center', fontsize=7.8, color=C['ink2'])
    ax.set_xticks(ks); ax.set_xticklabels(labels, fontsize=8.5); ax.grid(False, axis='x')
    ax.text(x0 + 0.1, -lim * 0.9, 'shaded band = 95% range', ha='left', va='bottom', fontsize=7.8, color=C['ink2'])
# Human-readable district names; these are broad postal-district areas, not exact town boundaries.
MAP_NAME = {1:'Raffles Place',2:'Tanjong Pagar',3:'Queenstown',4:'Sentosa',5:'Clementi',6:'City Hall',7:'Bugis',8:'Little India',9:'Orchard',10:'Bukit Timah',11:'Novena',12:'Toa Payoh',13:'Potong Pasir',14:'Geylang / Eunos',15:'Katong / Marine Parade',16:'Bedok / Upper East Coast',17:'Changi',18:'Tampines / Pasir Ris',19:'North-east towns',20:'Bishan / Ang Mo Kio',21:'Upper Bukit Timah',22:'Jurong',23:'Bukit Batok',25:'Woodlands',26:'Upper Thomson',27:'Yishun / Sembawang',28:'Seletar'}
```

```python
# External data, embedded so the notebook runs offline. Every series was pulled from the official source listed,
# checked for gaps, and spot-checked against published figures (e.g. URA all-residential index 4Q2024 = 209.4).
PPI_NONLANDED = {'1975-Q1': 10.5, '1975-Q2': 11.2, '1975-Q3': 11.5, '1975-Q4': 11.5, '1976-Q1': 11.5, '1976-Q2': 11.7, '1976-Q3': 11.7, '1976-Q4': 11.5, '1977-Q1': 11.5, '1977-Q2': 11.5, '1977-Q3': 11.5, '1977-Q4': 11.5, '1978-Q1': 11.5, '1978-Q2': 11.7, '1978-Q3': 11.7, '1978-Q4': 12.1, '1979-Q1': 12.7, '1979-Q2': 13.3, '1979-Q3': 14.6, '1979-Q4': 15.8, '1980-Q1': 18.9, '1980-Q2': 22.0, '1980-Q3': 25.8, '1980-Q4': 31.0, '1981-Q1': 39.6, '1981-Q2': 44.3, '1981-Q3': 43.1, '1981-Q4': 40.0, '1982-Q1': 38.7, '1982-Q2': 39.0, '1982-Q3': 38.7, '1982-Q4': 39.6, '1983-Q1': 40.6, '1983-Q2': 41.2, '1983-Q3': 40.6, '1983-Q4': 39.6, '1984-Q1': 38.1, '1984-Q2': 35.4, '1984-Q3': 33.7, '1984-Q4': 33.1, '1985-Q1': 30.9, '1985-Q2': 31.3, '1985-Q3': 29.9, '1985-Q4': 28.5, '1986-Q1': 26.5, '1986-Q2': 23.5, '1986-Q3': 24.8, '1986-Q4': 26.5, '1987-Q1': 26.8, '1987-Q2': 28.3, '1987-Q3': 31.0, '1987-Q4': 31.4, '1988-Q1': 30.7, '1988-Q2': 31.6, '1988-Q3': 32.6, '1988-Q4': 34.4, '1989-Q1': 35.2, '1989-Q2': 36.4, '1989-Q3': 38.7, '1989-Q4': 39.4, '1990-Q1': 41.6, '1990-Q2': 43.1, '1990-Q3': 42.7, '1990-Q4': 41.8, '1991-Q1': 42.8, '1991-Q2': 46.7, '1991-Q3': 47.5, '1991-Q4': 47.7, '1992-Q1': 49.9, '1992-Q2': 52.7, '1992-Q3': 55.2, '1992-Q4': 56.3, '1993-Q1': 57.2, '1993-Q2': 61.6, '1993-Q3': 65.4, '1993-Q4': 68.5, '1994-Q1': 72.7, '1994-Q2': 83.4, '1994-Q3': 88.7, '1994-Q4': 92.3, '1995-Q1': 95.2, '1995-Q2': 96.8, '1995-Q3': 97.7, '1995-Q4': 105.5, '1996-Q1': 110.8, '1996-Q2': 122.2, '1996-Q3': 119.9, '1996-Q4': 118.8, '1997-Q1': 111.9, '1997-Q2': 110.9, '1997-Q3': 106.2, '1997-Q4': 102.9, '1998-Q1': 93.0, '1998-Q2': 85.4, '1998-Q3': 76.8, '1998-Q4': 71.2, '1999-Q1': 75.2, '1999-Q2': 83.5, '1999-Q3': 90.4, '1999-Q4': 95.1, '2000-Q1': 98.5, '2000-Q2': 100.4, '2000-Q3': 97.7, '2000-Q4': 94.2, '2001-Q1': 90.0, '2001-Q2': 89.8, '2001-Q3': 87.3, '2001-Q4': 82.7, '2002-Q1': 82.0, '2002-Q2': 81.4, '2002-Q3': 81.6, '2002-Q4': 81.3, '2003-Q1': 80.6, '2003-Q2': 80.2, '2003-Q3': 79.9, '2003-Q4': 79.9, '2004-Q1': 79.6, '2004-Q2': 79.6, '2004-Q3': 79.9, '2004-Q4': 80.8, '2005-Q1': 81.4, '2005-Q2': 81.9, '2005-Q3': 83.1, '2005-Q4': 84.4, '2006-Q1': 85.7, '2006-Q2': 87.2, '2006-Q3': 90.0, '2006-Q4': 93.8, '2007-Q1': 98.8, '2007-Q2': 107.1, '2007-Q3': 116.0, '2007-Q4': 124.4, '2008-Q1': 129.0, '2008-Q2': 129.0, '2008-Q3': 125.8, '2008-Q4': 117.8, '2009-Q1': 100.0, '2009-Q2': 95.3, '2009-Q3': 110.5, '2009-Q4': 118.4, '2010-Q1': 124.3, '2010-Q2': 130.5, '2010-Q3': 132.6, '2010-Q4': 135.0, '2011-Q1': 137.3, '2011-Q2': 139.2, '2011-Q3': 140.8, '2011-Q4': 141.2, '2012-Q1': 141.0, '2012-Q2': 141.6, '2012-Q3': 142.3, '2012-Q4': 144.8, '2013-Q1': 145.9, '2013-Q2': 148.0, '2013-Q3': 148.9, '2013-Q4': 147.6, '2014-Q1': 145.7, '2014-Q2': 144.5, '2014-Q3': 143.9, '2014-Q4': 142.5, '2015-Q1': 140.9, '2015-Q2': 139.8, '2015-Q3': 137.7, '2015-Q4': 137.4, '2016-Q1': 136.6, '2016-Q2': 136.5, '2016-Q3': 134.9, '2016-Q4': 133.8, '2017-Q1': 133.8, '2017-Q2': 133.7, '2017-Q3': 134.5, '2017-Q4': 135.6, '2018-Q1': 141.6, '2018-Q2': 146.1, '2018-Q3': 146.1, '2018-Q4': 146.8, '2019-Q1': 145.2, '2019-Q2': 148.1, '2019-Q3': 150.0, '2019-Q4': 149.6, '2020-Q1': 148.1, '2020-Q2': 148.7, '2020-Q3': 148.8, '2020-Q4': 153.3, '2021-Q1': 157.1, '2021-Q2': 158.8, '2021-Q3': 159.9, '2021-Q4': 168.4, '2022-Q1': 167.9, '2022-Q2': 173.9, '2022-Q3': 181.6, '2022-Q4': 182.1, '2023-Q1': 186.8, '2023-Q2': 185.7, '2023-Q3': 189.8, '2023-Q4': 194.2, '2024-Q1': 196.1, '2024-Q2': 197.3, '2024-Q3': 197.5, '2024-Q4': 203.4, '2025-Q1': 205.4, '2025-Q2': 206.8, '2025-Q3': 208.5, '2025-Q4': 208.1, '2026-Q1': 210.8, '2026-Q2': 210.6}
PPI_ALLRES = {'2000-Q1': 98.9, '2000-Q2': 100.4, '2000-Q3': 97.6, '2000-Q4': 94.9, '2001-Q1': 91.1, '2001-Q2': 90.3, '2001-Q3': 87.6, '2001-Q4': 83.8, '2002-Q1': 82.8, '2002-Q2': 82.2, '2002-Q3': 82.6, '2002-Q4': 82.3, '2003-Q1': 81.6, '2003-Q2': 81.1, '2003-Q3': 80.7, '2003-Q4': 80.6, '2004-Q1': 80.3, '2004-Q2': 80.4, '2004-Q3': 80.7, '2004-Q4': 81.3, '2005-Q1': 81.9, '2005-Q2': 82.3, '2005-Q3': 83.3, '2005-Q4': 84.5, '2006-Q1': 85.8, '2006-Q2': 87.3, '2006-Q3': 89.6, '2006-Q4': 93.1, '2007-Q1': 97.6, '2007-Q2': 105.6, '2007-Q3': 114.4, '2007-Q4': 122.1, '2008-Q1': 126.7, '2008-Q2': 126.9, '2008-Q3': 123.9, '2008-Q4': 116.4, '2009-Q1': 100.0, '2009-Q2': 95.3, '2009-Q3': 110.3, '2009-Q4': 118.4, '2010-Q1': 125.1, '2010-Q2': 131.7, '2010-Q3': 135.5, '2010-Q4': 139.2, '2011-Q1': 142.3, '2011-Q2': 145.1, '2011-Q3': 147.0, '2011-Q4': 147.4, '2012-Q1': 147.2, '2012-Q2': 147.9, '2012-Q3': 148.8, '2012-Q4': 151.5, '2013-Q1': 152.4, '2013-Q2': 154.0, '2013-Q3': 154.6, '2013-Q4': 153.2, '2014-Q1': 151.3, '2014-Q2': 149.7, '2014-Q3': 148.6, '2014-Q4': 147.0, '2015-Q1': 145.5, '2015-Q2': 144.2, '2015-Q3': 142.3, '2015-Q4': 141.6, '2016-Q1': 140.6, '2016-Q2': 140.0, '2016-Q3': 137.9, '2016-Q4': 137.2, '2017-Q1': 136.7, '2017-Q2': 136.6, '2017-Q3': 137.6, '2017-Q4': 138.7, '2018-Q1': 144.1, '2018-Q2': 149.0, '2018-Q3': 149.7, '2018-Q4': 149.6, '2019-Q1': 148.6, '2019-Q2': 150.8, '2019-Q3': 152.8, '2019-Q4': 153.6, '2020-Q1': 152.1, '2020-Q2': 152.6, '2020-Q3': 153.8, '2020-Q4': 157.0, '2021-Q1': 162.2, '2021-Q2': 163.5, '2021-Q3': 165.3, '2021-Q4': 173.6, '2022-Q1': 174.8, '2022-Q2': 180.9, '2022-Q3': 187.8, '2022-Q4': 188.6, '2023-Q1': 194.8, '2023-Q2': 194.4, '2023-Q3': 196.0, '2023-Q4': 201.5, '2024-Q1': 204.3, '2024-Q2': 206.1, '2024-Q3': 204.7, '2024-Q4': 209.4, '2025-Q1': 211.1, '2025-Q2': 213.2, '2025-Q3': 215.1, '2025-Q4': 216.4, '2026-Q1': 218.3, '2026-Q2': 219.4}
HDB_RPI = {'1990-Q1': 24.3, '1990-Q2': 24.4, '1990-Q3': 25.0, '1990-Q4': 24.7, '1991-Q1': 24.9, '1991-Q2': 25.5, '1991-Q3': 25.2, '1991-Q4': 25.1, '1992-Q1': 25.8, '1992-Q2': 27.4, '1992-Q3': 28.0, '1992-Q4': 28.6, '1993-Q1': 30.2, '1993-Q2': 39.6, '1993-Q3': 47.7, '1993-Q4': 49.0, '1994-Q1': 50.4, '1994-Q2': 51.7, '1994-Q3': 54.6, '1994-Q4': 54.8, '1995-Q1': 57.3, '1995-Q2': 63.9, '1995-Q3': 67.7, '1995-Q4': 73.7, '1996-Q1': 80.5, '1996-Q2': 90.9, '1996-Q3': 96.4, '1996-Q4': 99.0, '1997-Q1': 98.6, '1997-Q2': 97.5, '1997-Q3': 93.6, '1997-Q4': 88.0, '1998-Q1': 81.7, '1998-Q2': 78.3, '1998-Q3': 74.9, '1998-Q4': 72.3, '1999-Q1': 71.2, '1999-Q2': 72.2, '1999-Q3': 78.1, '1999-Q4': 79.8, '2000-Q1': 80.3, '2000-Q2': 79.3, '2000-Q3': 77.6, '2000-Q4': 75.8, '2001-Q1': 73.2, '2001-Q2': 72.1, '2001-Q3': 70.6, '2001-Q4': 69.6, '2002-Q1': 69.1, '2002-Q2': 69.2, '2002-Q3': 69.9, '2002-Q4': 69.9, '2003-Q1': 71.0, '2003-Q2': 72.5, '2003-Q3': 74.3, '2003-Q4': 75.1, '2004-Q1': 75.3, '2004-Q2': 76.2, '2004-Q3': 76.3, '2004-Q4': 77.1, '2005-Q1': 77.2, '2005-Q2': 73.5, '2005-Q3': 73.2, '2005-Q4': 73.5, '2006-Q1': 73.6, '2006-Q2': 74.3, '2006-Q3': 74.2, '2006-Q4': 74.9, '2007-Q1': 75.8, '2007-Q2': 78.1, '2007-Q3': 83.2, '2007-Q4': 88.0, '2008-Q1': 91.3, '2008-Q2': 95.4, '2008-Q3': 99.4, '2008-Q4': 100.8, '2009-Q1': 100.0, '2009-Q2': 101.4, '2009-Q3': 105.0, '2009-Q4': 109.0, '2010-Q1': 112.1, '2010-Q2': 116.6, '2010-Q3': 121.3, '2010-Q4': 124.4, '2011-Q1': 126.4, '2011-Q2': 130.4, '2011-Q3': 135.4, '2011-Q4': 137.7, '2012-Q1': 138.5, '2012-Q2': 140.3, '2012-Q3': 143.1, '2012-Q4': 146.7, '2013-Q1': 148.6, '2013-Q2': 149.4, '2013-Q3': 148.1, '2013-Q4': 145.8, '2014-Q1': 143.5, '2014-Q2': 141.5, '2014-Q3': 139.1, '2014-Q4': 137.0, '2015-Q1': 135.6, '2015-Q2': 135.0, '2015-Q3': 134.6, '2015-Q4': 134.8, '2016-Q1': 134.7, '2016-Q2': 134.7, '2016-Q3': 134.7, '2016-Q4': 134.6, '2017-Q1': 133.9, '2017-Q2': 133.7, '2017-Q3': 132.8, '2017-Q4': 132.6, '2018-Q1': 131.6, '2018-Q2': 131.7, '2018-Q3': 131.6, '2018-Q4': 131.4, '2019-Q1': 131.0, '2019-Q2': 130.8, '2019-Q3': 130.9, '2019-Q4': 131.5, '2020-Q1': 131.5, '2020-Q2': 131.9, '2020-Q3': 133.9, '2020-Q4': 138.1, '2021-Q1': 142.2, '2021-Q2': 146.4, '2021-Q3': 150.6, '2021-Q4': 155.7, '2022-Q1': 159.5, '2022-Q2': 163.9, '2022-Q3': 168.1, '2022-Q4': 171.9, '2023-Q1': 173.6, '2023-Q2': 176.2, '2023-Q3': 178.5, '2023-Q4': 180.4, '2024-Q1': 183.7, '2024-Q2': 187.9, '2024-Q3': 192.9, '2024-Q4': 197.9, '2025-Q1': 201.0, '2025-Q2': 202.9, '2025-Q3': 203.7, '2025-Q4': 203.6, '2026-Q1': 203.4, '2026-Q2': 202.8}
RENT_NONLANDED = {'2004-Q1': 63.4, '2004-Q2': 63.5, '2004-Q3': 63.6, '2004-Q4': 64.0, '2005-Q1': 64.7, '2005-Q2': 65.1, '2005-Q3': 65.6, '2005-Q4': 66.0, '2006-Q1': 66.7, '2006-Q2': 68.3, '2006-Q3': 71.8, '2006-Q4': 75.9, '2007-Q1': 82.1, '2007-Q2': 91.1, '2007-Q3': 101.9, '2007-Q4': 108.7, '2008-Q1': 115.1, '2008-Q2': 117.5, '2008-Q3': 116.2, '2008-Q4': 109.6, '2009-Q1': 100.0, '2009-Q2': 94.4, '2009-Q3': 92.4, '2009-Q4': 92.9, '2010-Q1': 97.4, '2010-Q2': 103.2, '2010-Q3': 106.9, '2010-Q4': 110.0, '2011-Q1': 111.2, '2011-Q2': 112.4, '2011-Q3': 113.2, '2011-Q4': 113.7, '2012-Q1': 114.3, '2012-Q2': 114.6, '2012-Q3': 115.6, '2012-Q4': 116.5, '2013-Q1': 117.3, '2013-Q2': 117.6, '2013-Q3': 117.8, '2013-Q4': 117.3, '2014-Q1': 116.8, '2014-Q2': 116.5, '2014-Q3': 115.2, '2014-Q4': 114.3, '2015-Q1': 112.2, '2015-Q2': 111.0, '2015-Q3': 110.2, '2015-Q4': 109.0, '2016-Q1': 107.7, '2016-Q2': 107.3, '2016-Q3': 105.8, '2016-Q4': 105.1, '2017-Q1': 104.4, '2017-Q2': 104.2, '2017-Q3': 104.1, '2017-Q4': 103.3, '2018-Q1': 103.6, '2018-Q2': 104.2, '2018-Q3': 104.5, '2018-Q4': 103.7, '2019-Q1': 104.8, '2019-Q2': 106.3, '2019-Q3': 106.7, '2019-Q4': 105.7, '2020-Q1': 107.1, '2020-Q2': 105.9, '2020-Q3': 105.3, '2020-Q4': 105.2, '2021-Q1': 107.7, '2021-Q2': 111.0, '2021-Q3': 112.6, '2021-Q4': 115.6, '2022-Q1': 120.3, '2022-Q2': 128.8, '2022-Q3': 139.5, '2022-Q4': 150.0, '2023-Q1': 159.3, '2023-Q2': 163.0, '2023-Q3': 163.3, '2023-Q4': 160.4, '2024-Q1': 157.8, '2024-Q2': 156.5, '2024-Q3': 157.3, '2024-Q4': 157.6, '2025-Q1': 158.4, '2025-Q2': 159.7, '2025-Q3': 161.5, '2025-Q4': 161.3, '2026-Q1': 161.9, '2026-Q2': 162.5}
SORA_3M = {'2005-10': 2.0404, '2005-11': 2.3302, '2005-12': 2.6352, '2006-01': 2.9672, '2006-02': 3.094, '2006-03': 3.1717, '2006-04': 3.1216, '2006-05': 3.1211, '2006-06': 3.1431, '2006-07': 3.1847, '2006-08': 3.2396, '2006-09': 3.2573, '2006-10': 3.3125, '2006-11': 3.3216, '2006-12': 3.2918, '2007-01': 3.205, '2007-02': 3.1456, '2007-03': 2.9207, '2007-04': 2.6611, '2007-05': 2.3771, '2007-06': 2.2379, '2007-07': 2.2052, '2007-08': 2.2695, '2007-09': 2.2837, '2007-10': 2.2575, '2007-11': 2.0178, '2007-12': 1.5836, '2008-01': 1.2604, '2008-02': 1.1332, '2008-03': 1.0839, '2008-04': 0.8531, '2008-05': 0.6457, '2008-06': 0.5125, '2008-07': 0.4498, '2008-08': 0.4307, '2008-09': 0.7137, '2008-10': 0.7723, '2008-11': 0.7117, '2008-12': 0.4392, '2009-01': 0.3089, '2009-02': 0.2298, '2009-03': 0.2047, '2009-04': 0.1974, '2009-05': 0.1683, '2009-06': 0.1232, '2009-07': 0.135, '2009-08': 0.1919, '2009-09': 0.2064, '2009-10': 0.2574, '2009-11': 0.2541, '2009-12': 0.2149, '2010-01': 0.1703, '2010-02': 0.1352, '2010-03': 0.1326, '2010-04': 0.123, '2010-05': 0.1054, '2010-06': 0.117, '2010-07': 0.1218, '2010-08': 0.1224, '2010-09': 0.1139, '2010-10': 0.0997, '2010-11': 0.0934, '2010-12': 0.0792, '2011-01': 0.0757, '2011-02': 0.0732, '2011-03': 0.0609, '2011-04': 0.0583, '2011-05': 0.0554, '2011-06': 0.0619, '2011-07': 0.0511, '2011-08': 0.0449, '2011-09': 0.0472, '2011-10': 0.0511, '2011-11': 0.0555, '2011-12': 0.0548, '2012-01': 0.0611, '2012-02': 0.0522, '2012-03': 0.0494, '2012-04': 0.0362, '2012-05': 0.0454, '2012-06': 0.0484, '2012-07': 0.0577, '2012-08': 0.0625, '2012-09': 0.0769, '2012-10': 0.0794, '2012-11': 0.0751, '2012-12': 0.0583, '2013-01': 0.0505, '2013-02': 0.0505, '2013-03': 0.0483, '2013-04': 0.0494, '2013-05': 0.0523, '2013-06': 0.0496, '2013-07': 0.0512, '2013-08': 0.0428, '2013-09': 0.0395, '2013-10': 0.051, '2013-11': 0.057, '2013-12': 0.0642, '2014-01': 0.0524, '2014-02': 0.048, '2014-03': 0.0575, '2014-04': 0.0652, '2014-05': 0.0785, '2014-06': 0.0718, '2014-07': 0.071, '2014-08': 0.073, '2014-09': 0.0811, '2014-10': 0.0868, '2014-11': 0.0979, '2014-12': 0.1183, '2015-01': 0.1433, '2015-02': 0.1981, '2015-03': 0.2331, '2015-04': 0.2619, '2015-05': 0.2362, '2015-06': 0.229, '2015-07': 0.2248, '2015-08': 0.2331, '2015-09': 0.214, '2015-10': 0.2521, '2015-11': 0.3049, '2015-12': 0.3483, '2016-01': 0.3526, '2016-02': 0.3027, '2016-03': 0.3075, '2016-04': 0.309, '2016-05': 0.2836, '2016-06': 0.2257, '2016-07': 0.1903, '2016-08': 0.1473, '2016-09': 0.1331, '2016-10': 0.0807, '2016-11': 0.0854, '2016-12': 0.1458, '2017-01': 0.1688, '2017-02': 0.2413, '2017-03': 0.2604, '2017-04': 0.3518, '2017-05': 0.4018, '2017-06': 0.4935, '2017-07': 0.5993, '2017-08': 0.6832, '2017-09': 0.7045, '2017-10': 0.7386, '2017-11': 0.7284, '2017-12': 0.8446, '2018-01': 0.6769, '2018-02': 0.7711, '2018-03': 0.6786, '2018-04': 0.8108, '2018-05': 0.8277, '2018-06': 0.9068, '2018-07': 1.0639, '2018-08': 1.1723, '2018-09': 1.2178, '2018-10': 1.2119, '2018-11': 1.2724, '2018-12': 1.4639, '2019-01': 1.5455, '2019-02': 1.5944, '2019-03': 1.5423, '2019-04': 1.6249, '2019-05': 1.6846, '2019-06': 1.715, '2019-07': 1.6703, '2019-08': 1.5875, '2019-09': 1.5278, '2019-10': 1.4567, '2019-11': 1.3102, '2019-12': 1.2636, '2020-01': 1.2023, '2020-02': 1.1404, '2020-03': 0.8809, '2020-04': 0.534, '2020-05': 0.2707, '2020-06': 0.1411, '2020-07': 0.0811, '2020-08': 0.0985, '2020-09': 0.0988, '2020-10': 0.1167, '2020-11': 0.116, '2020-12': 0.1661, '2021-01': 0.1694, '2021-02': 0.1996, '2021-03': 0.2216, '2021-04': 0.1862, '2021-05': 0.1679, '2021-06': 0.1326, '2021-07': 0.1321, '2021-08': 0.13, '2021-09': 0.1306, '2021-10': 0.1555, '2021-11': 0.1579, '2021-12': 0.1943, '2022-01': 0.2065, '2022-02': 0.2504, '2022-03': 0.276, '2022-04': 0.3227, '2022-05': 0.5263, '2022-06': 0.772, '2022-07': 1.2712, '2022-08': 1.5967, '2022-09': 2.0655, '2022-10': 2.4613, '2022-11': 2.9243, '2022-12': 3.0294, '2023-01': 3.1419, '2023-02': 3.2206, '2023-03': 3.5775, '2023-04': 3.6057, '2023-05': 3.6241, '2023-06': 3.6549, '2023-07': 3.6748, '2023-08': 3.6964, '2023-09': 3.7017, '2023-10': 3.7527, '2023-11': 3.7464, '2023-12': 3.7018, '2024-01': 3.6548, '2024-02': 3.6488, '2024-03': 3.6829, '2024-04': 3.6491, '2024-05': 3.6597, '2024-06': 3.6336, '2024-07': 3.6381, '2024-08': 3.5694, '2024-09': 3.4932, '2024-10': 3.3769, '2024-11': 3.2147, '2024-12': 3.0346, '2025-01': 2.8993, '2025-02': 2.7291, '2025-03': 2.5507, '2025-04': 2.3587, '2025-05': 2.2692, '2025-06': 2.0533, '2025-07': 1.845, '2025-08': 1.5638, '2025-09': 1.4532, '2025-10': 1.3203, '2025-11': 1.253, '2025-12': 1.1857, '2026-01': 1.1502, '2026-02': 1.1219, '2026-03': 1.0708, '2026-04': 1.0298, '2026-05': 1.0564, '2026-06': 1.0825, '2026-07': 1.1354}

q = pd.DataFrame({'ppi': pd.Series(PPI_NONLANDED), 'hdb': pd.Series(HDB_RPI), 'rent': pd.Series(RENT_NONLANDED)})
sora_m = pd.Series(SORA_3M); sora_m.index = pd.PeriodIndex(sora_m.index, freq='M')
sq = sora_m.groupby(sora_m.index.asfreq('Q')).mean(); sq.index = [f'{p.year}-Q{p.quarter}' for p in sq.index]
q['sora'] = sq
q['g'] = np.log(q.ppi).diff() * 100                      # quarterly % growth of private non-landed prices
q['g_hdb'] = np.log(q.hdb).diff() * 100; q['g_rent'] = np.log(q.rent).diff() * 100; q['dsora'] = q.sora.diff()
TIGHTENING = ['2009-Q3', '2010-Q1', '2010-Q3', '2011-Q1', '2011-Q4', '2012-Q4', '2013-Q1', '2013-Q2', '2018-Q3', '2021-Q4', '2022-Q3', '2023-Q2', '2025-Q3']
q['tight'] = q.index.isin(TIGHTENING).astype(float)
ACTUAL_2026_Q3_FLASH = round(q.loc['2026-Q2', 'ppi'] * 1.009, 1)   # URA flash 1 Oct 2026: non-landed +0.9% q-o-q

# Rules in force for 2025-26 purchases (IRAS, MOF, MAS)
BSD_TIERS = [(180e3, .01), (180e3, .02), (640e3, .03), (500e3, .04), (1.5e6, .05), (1e12, .06)]
TDSR, STRESS_RATE = 0.55, 0.04
LTV = {'first loan': 0.75, 'second loan': 0.45}
# TDSR counts gross income excluding the employer's CPF. Matching benchmark: SingStat table 17896, median monthly employment income of
# resident employed households excluding employer CPF, S$10,591 in 2025 (preliminary).
MEDIAN_INCOME_TDSR = 10_591
def bsd(p):
    tax, rem = 0.0, p
    for amt, r in BSD_TIERS:
        x = min(rem, amt); tax += x * r; rem -= x
        if rem <= 0: break
    return tax
def annuity(rate, years): r = rate / 12; n = years * 12; return (1 - (1 + r) ** -n) / r
def payment(loan, rate, years): return loan / annuity(rate, years)
def max_price(income, cash, cpf, tenure, ltv=0.75, costs=0.0, min_cash_share=0.05):
    """Largest price they can pay. At each price the loan is the SMALLER of the LTV limit and the TDSR limit (55% of income at 4%);
    cash + CPF must cover the rest of the price, buyer's stamp duty and other costs; at least min_cash_share of the price must be cash."""
    tdsr_loan = TDSR * income * annuity(STRESS_RATE, tenure); g = np.arange(3e5, 5e6, 1000)
    loan = np.minimum(ltv * g, tdsr_loan); bsd_ = np.array([bsd(p) for p in g])
    ok = ((g - loan + bsd_ + costs) <= cash + cpf) & (cash - costs >= min_cash_share * g)
    p_best = g[ok].max()
    funds_cap = g[((g - ltv * g + bsd_ + costs) <= cash + cpf) & (cash - costs >= min_cash_share * g)].max()   # if the loan were unlimited
    return p_best, funds_cap, tdsr_loan / ltv, ('savings / cash' if p_best >= funds_cap - 1000 else 'loan (TDSR)')
print('Quarterly series:', q.index[0], 'to', q.index[-1], '| SORA:', sora_m.index[0], 'to', sora_m.index[-1])
```

```python
# Can anyone time this market? Test forecasting rules out of sample (2014-2025), each using only data available at the time.
q['g1'] = q.g.shift(1); q['g2'] = q.g.shift(2); q['trend4'] = q.g.rolling(4).mean().shift(1)
q['ds4'] = q.dsora.rolling(4).sum().shift(1); q['ghdb1'] = q.g_hdb.shift(1); q['grent1'] = q.g_rent.shift(1); q['tight1'] = q.tight.shift(1)
SPECS = {'Past growth (AR model)': ['g1', 'g2'], '+ interest rates': ['g1', 'g2', 'ds4'],
         '+ rates, HDB prices, rents, policy': ['g1', 'g2', 'ds4', 'ghdb1', 'grent1', 'tight1']}  # only lagged policy status is known one quarter ahead
idx = list(q.index)
def fit(cols, end):
    m = q.loc['2006-Q2':end].dropna(subset=['g'] + cols)
    Xf = np.column_stack([np.ones(len(m))] + [m[c] for c in cols]); return np.linalg.lstsq(Xf, m.g, rcond=None)[0]
errs = {k: [] for k in list(SPECS) + ["Last year's trend continues", 'No change']}
for t in idx[idx.index('2014-Q1'): idx.index('2025-Q4') + 1]:
    prev = idx[idx.index(t) - 1]
    for k, cols in SPECS.items(): errs[k].append(q.loc[t, 'g'] - np.r_[1, [q.loc[t, c] for c in cols]] @ fit(cols, prev))
    errs["Last year's trend continues"].append(q.loc[t, 'g'] - q.loc[t, 'trend4']); errs['No change'].append(q.loc[t, 'g'])
rmse = pd.Series({k: np.sqrt(np.mean(np.square(v))) for k, v in errs.items()}).sort_values()
trend = q.loc['2025-Q1':'2025-Q4', 'g'].mean()
horizon = ['2026-Q1', '2026-Q2', '2026-Q3', '2026-Q4', '2027-Q1', '2027-Q2', '2027-Q3', '2027-Q4']
hist_err, HIST_E = {}, {}
for h in range(1, 9):
    e_h = []
    for t in idx[idx.index('2010-Q1'): idx.index('2025-Q4') - h + 2]:
        start = idx[idx.index(t) - 1]; tr = q.loc[:start, 'g'].tail(4).mean()
        e_h.append((np.log(q.ppi[idx[idx.index(start) + h]]) - np.log(q.ppi[start])) * 100 - tr * h)
    hist_err[h] = np.percentile(e_h, [10, 90]); HIST_E[h] = np.array(e_h)
lvl0 = q.loc['2025-Q4', 'ppi']
fc = pd.DataFrame({'forecast': [lvl0 * np.exp(trend * h / 100) for h in range(1, 9)],
                   'low': [lvl0 * np.exp((trend * h + hist_err[h][0]) / 100) for h in range(1, 9)],
                   'high': [lvl0 * np.exp((trend * h + hist_err[h][1]) / 100) for h in range(1, 9)]}, index=horizon)
fc['actual'] = [q.ppi.get('2026-Q1'), q.ppi.get('2026-Q2'), ACTUAL_2026_Q3_FLASH] + [np.nan] * 5
fc['error %'] = (fc.forecast / fc.actual - 1) * 100
def trend_fan(col, level):
    """Trend-rule forecast (2025 average quarterly growth) for any quarterly index, with an 80% range from its own past errors."""
    g_ = np.log(q[level]).diff() * 100; tr_ = g_.loc['2025-Q1':'2025-Q4'].mean(); err_ = {}
    for h in range(1, 9):
        e_h = []
        for t in idx[idx.index('2010-Q1'): idx.index('2025-Q4') - h + 2]:
            start = idx[idx.index(t) - 1]; e_h.append((np.log(q[level][idx[idx.index(start) + h]]) - np.log(q[level][start])) * 100 - g_.loc[:start].tail(4).mean() * h)
        err_[h] = np.percentile(e_h, [10, 90])
    return tr_, err_
H_RMSE = {}
for h in (4, 8):                                             # longer-horizon test: total growth over h quarters, trend rule vs no change
    tr_err, nc_err = [], []
    for t in idx[idx.index('2014-Q1'): idx.index('2025-Q4') - h + 2]:
        start = idx[idx.index(t) - 1]; act = (np.log(q.ppi[idx[idx.index(start) + h]]) - np.log(q.ppi[start])) * 100
        tr_err.append(act - q.loc[:start, 'g'].tail(4).mean() * h); nc_err.append(act)
    H_RMSE[h] = (np.sqrt(np.mean(np.square(tr_err))), np.sqrt(np.mean(np.square(nc_err))))
print(f"Trend rule: private prices +{trend * 4:.1f}% a year; lowest out-of-sample error of {len(rmse)} rules ({rmse.iloc[0]:.2f} vs {rmse.iloc[-1]:.2f} points a quarter)")
```

```python
# The buyer: an HDB upgrader family. An illustrative profile: edit these numbers to try your own.
PERSON = dict(income=16_000,          # combined gross monthly income (S$), excl. employer CPF; about 1.5x the 2025 median of S$10,591
              hdb_value=795_000,      # 5-room HDB in Tampines, average resale price 4Q 2025 (Huttons)
              hdb_loan_left=150_000,  # outstanding HDB loan
              cpf_refund=350_000,     # assumed: CPF used for the flat (plus accrued interest), refunded to their CPF accounts on sale
              sell_costs=20_000,      # assumed: agent, legal and moving costs
              age=38, want_sqft=1050) # about 1,050 sqft (typical 3-bedroom size); both Singapore citizens, flat past its minimum occupation period
PAYMENT_RATE = 0.016  # Fixed route-comparison assumption, separate from monthly SORA illustrations.
TENURE = min(30, 65 - PERSON['age'])
def budget_for(income, hdb_value, loan_left=PERSON['hdb_loan_left']):
    net = hdb_value - loan_left - PERSON['sell_costs']; cpf = min(PERSON['cpf_refund'], net); cash = net - cpf
    return max_price(income, cash, cpf, TENURE)
BUDGET, CASH_CAP, LOAN_CAP, BINDING = budget_for(PERSON['income'], PERSON['hdb_value'])
NET_PROCEEDS = PERSON['hdb_value'] - PERSON['hdb_loan_left'] - PERSON['sell_costs']
OPTIONS = [('A', 'New launch, Tampines/Pasir Ris', 18, 'New Sale', '99-yr', 97, 1050),
           ('B', 'Resale, 85-88 yrs lease left, Tampines/Pasir Ris', 18, 'Resale', '99-yr', 87, 1050),
           ('C', 'Resale, 85-88 yrs lease left, Serangoon/Hougang/Sengkang/Punggol', 19, 'Resale', '99-yr', 87, 1050),
           ('D', 'Older resale, Bedok (too few sales)', 16, 'Resale', '99-yr', 74, 1050),
           ('E', 'Freehold resale, Katong/Marine Parade', 15, 'Resale', 'Freehold', None, 1050),
           ('F', 'Resale, 89-93 yrs lease left, Tampines/Pasir Ris', 18, 'Resale', '99-yr', 91, 1050),
           ('G', 'Resale, 89-93 yrs lease left, Geylang/Eunos fringe', 14, 'Resale', '99-yr', 91, 1050)]
OPT_FLOOR, OPT_DESC = 11, '1,050 sqft unit on floor 11'
REF = dict(district=18, sale='Resale', tenure='99-yr', area=1050, floor=11, lease_left=91)
REF_DESC = 'a 1,050 sqft longer-lease resale (91 years left) in Tampines, floor 11'
MENU_SQFT, MENU_DISTRICTS = 200, [14, 10]
MENU_TITLE = 'Location and newness move the price most; 10 floors add about S$75k'
RATES_TITLE = 'Rates fell {rate_drop:.0%} in 2025: payments fell {drop:.0%}, but the loan limit didn\'t move'
MAP_TITLE = 'The model puts a family-size resale within budget in {n_ok} of {n} districts'
MAP_SUB = 'Model price of a 1,050 sqft resale with 91 years left (floor 11) in each district, compared with their budget.'
OPT_TITLE = 'Resales with 85-88 years of lease left fit comfortably; ones with 89-93 years left cost about S$0.3M more'
SEG = (df.sale == 'Resale') & df.segment.isin(['OCR', 'RCR']) & df.area.between(950, 1300)
SEG_LABEL = 'Family-size resales outside prime areas'
SEG_TITLE = 'Is a higher floor or a bigger unit worth it? Floors yes; no clear within-project size effect'
LEASE_EDGES = [0, 70, 75, 80, 85, 89, 93, 100]
LEASE_TITLE = 'For a family home, each year of lease used is linked to about {:.1f}% lower psf'
VF_MASK = ((df.sale == 'Resale') & df.district.isin([14, 15, 16, 17, 18, 19]) & df.area.between(850, 1300) &
             (df.tenure == '99-yr') & ((99 - df.lease_used) >= 80))      # freehold excluded: its building age isn't recorded
VF_TITLE = 'Eight eastern projects picked on Jan-Jun discounts, then checked on Jul-Dec sales'
VF_SUB = 'Family-size 99-year resales in the east and north-east with 80+ years left. MRT walk enters the model. Inspect, do not assume a bargain.'
FIT_MASK = df.area.between(950, 1150)                     # a sale "like theirs": family-size unit (any sale type, any lease)
ZOOM_EXT, ZOOM_LABEL = (103.865, 103.99, 1.300, 1.392), 'Zoom: the east (Geylang to Pasir Ris)'
MRTMAP_TITLE = 'In the east, {n_fit} of {n_all} projects had family-size sales under budget; {n_fit_near} of those had an estimated MRT walk of at most 5 minutes'
MRTMAP_SUB = 'Each dot is a project with 950-1,150 sqft sales in 2025 (any sale type), coloured by their median price against the budget; bigger dots = more sales. Labelled: the busiest under-budget projects within 5 minutes of a station.'
MRTPREM_TITLE = 'Family-size resales within 5 minutes of MRT sell for {near:+.0f}% more per sqft than those over 15 minutes away'
print(f"Net proceeds {sgm(NET_PROCEEDS)} (cash {sgm(NET_PROCEEDS - PERSON['cpf_refund'])}, CPF {sgm(PERSON['cpf_refund'])}). Budget {sgm(BUDGET)}, limited by {BINDING}: "
      f"their funds alone would allow {sgm(CASH_CAP)}, the loan alone {sgm(LOAN_CAP)}. "
      f"ABSD if they buy before selling: {sgd(0.2 * BUDGET)} (refund subject to IRAS conditions).")
print('Fixed option payment assumption (%):',100*PAYMENT_RATE)
print('December 2025 embedded SORA + assumed 0.30-point spread (%):',SORA_3M['2025-12']+0.30)
```

```python
# The hedonic regression: log price per sqft on district, sale type, tenure, property type, size, floor, lease used and month,
# with standard errors clustered by project.
d = df.copy()
X = pd.concat([dummies(d.district, 'D', 9), dummies(d.sale, 'sale', 'Resale'), dummies(d.tenure, 'ten', '99-yr'),
               dummies(d.type, 'type', 'Condominium'), d[['larea', 'floor', 'lease_used']], dummies(d.month, 'm', 1)], axis=1)
hed, H = ols(d.lpsf, X, list(X.columns), cluster=d.project)
df['resid'] = H['e']
def model_price(district, sale='Resale', tenure='99-yr', area=1050, floor=11, lease_left=91, typ='Condominium', month=12):
    """Model price (S$) for a unit with these features, at December 2025 prices."""
    x = pd.Series(0.0, index=H['names']); x['const'] = 1
    if district != 9: x[f'D_{district}'] = 1
    if sale != 'Resale': x[f'sale_{sale}'] = 1
    if tenure != '99-yr': x[f'ten_{tenure}'] = 1
    if typ == 'Apartment': x['type_Apartment'] = 1
    x['larea'] = np.log(area); x['floor'] = floor; x['lease_used'] = (99 - lease_left) if tenure == '99-yr' else 0
    if month != 1: x[f'm_{month}'] = 1
    return float(np.exp(x.values @ H['b'])) * area
# Comparable real sales for a route: same district, sale type and tenure, size within 100 sqft, floors 6-20, similar years of lease left.
def lease_band(left):
    """Comparable lease window for a route: longer-lease 89-93 years left, 10-14 years old 85-88, otherwise +/-4 years."""
    return (89, 93) if left >= 89 else ((85, 88) if left >= 85 else (left - 4, left + 4))
def similar(dist, sale, ten, left, sqft):
    c_ = df[(df.district == dist) & (df.sale == sale) & (df.tenure == ten) & df.area.between(sqft - 100, sqft + 100) & df.floor.between(6, 20)]
    if ten != '99-yr': return c_
    if sale == 'New Sale': return c_[(99 - c_.lease_used) >= 92]
    lo_, hi_ = lease_band(left); return c_[(99 - c_.lease_used).between(lo_, hi_)]
print(f"R2 = {H['r2']:.3f}; residual SD = {H['e'].std():.3f} log points (about ±{np.expm1(H['e'].std()) * 100:.0f}% for one SD)")
```

```python
# Diagnostic quantities used later (leverage, studentised residuals, Cook's distance, VIF, residual spread by sale type).
Xa = np.column_stack([np.ones(len(X)), X.values]); n_, p_ = Xa.shape
XtXi = np.linalg.pinv(Xa.T @ Xa); e_ = H['e']; s2 = e_ @ e_ / (n_ - p_); fitted = Xa @ H['b']
lev = np.einsum('ij,jk,ik->i', Xa, XtXi, Xa); rstud = e_ / np.sqrt(s2 * (1 - lev)); cook = rstud ** 2 / p_ * lev / (1 - lev)
SD_BY_SALE = pd.Series(e_).groupby(df.sale.values).std()
VIF = {}
for v in ['larea', 'floor', 'lease_used']:
    A = np.column_stack([np.ones(n_), X.drop(columns=v).values]); yy_ = X[v].values; bb = np.linalg.lstsq(A, yy_, rcond=None)[0]
    VIF[v] = 1 / (((yy_ - A @ bb) ** 2).sum() / ((yy_ - yy_.mean()) ** 2).sum())
VIF_ALL = {}
for j_, v in enumerate(X.columns):
    A = np.column_stack([np.ones(n_), np.delete(X.values, j_, axis=1)]); yy_ = X[v].values; bb = np.linalg.lstsq(A, yy_, rcond=None)[0]
    VIF_ALL[v] = 1 / (((yy_ - A @ bb) ** 2).sum() / ((yy_ - yy_.mean()) ** 2).sum())
VIF_MAX = max(VIF_ALL.values())
cook_proj = pd.Series(cook, index=df.project.values).groupby(level=0).sum().sort_values(ascending=False)
# Held-out check: 5 folds of whole projects; each fold is predicted by a model fitted without it.
fold_ = pd.factorize(df.project)[0] % 5; e_cv = np.zeros(n_); yl_all = df.lpsf.values
for k_ in range(5):
    tr_ = fold_ != k_; b_ = np.linalg.lstsq(Xa[tr_], yl_all[tr_], rcond=None)[0]; e_cv[~tr_] = yl_all[~tr_] - Xa[~tr_] @ b_
R2_CV = 1 - (e_cv ** 2).sum() / ((yl_all - yl_all.mean()) ** 2).sum()
COVER = pd.Series(np.abs(e_cv) <= 1.2816 * df.sale.map(SD_BY_SALE).values).groupby(df.sale.values).mean()   # share of held-out sales inside the 80% band
# Influence check: refit without the 10 projects with the largest total Cook's distance.
keep_ = ~df.project.isin(pd.Series(cook, index=df.project.values).groupby(level=0).sum().nlargest(10).index).values
b_k = np.linalg.lstsq(Xa[keep_], yl_all[keep_], rcond=None)[0]
INFL = pd.DataFrame({'all sales': np.expm1(H['b']) * 100, 'without top-10 influential projects': np.expm1(b_k) * 100}, index=H['names']).loc[['lease_used', 'floor', 'sale_New Sale', 'larea']]
print(f"Residual SD by sale type: {SD_BY_SALE.round(3).to_dict()}; largest Cook's distance {cook.max():.3f}; highest VIF {VIF_MAX:.1f}; "
      f"held-out-project R2 {R2_CV:.3f}; 80% band covers {COVER.round(3).to_dict()} of held-out sales")
```

```python
# Walking time to the MRT. Project locations: OneMap (Singapore Land Authority's official geocoder), searched by project name and
# matched by name and street where possible; other name/street fallbacks are retained only when their postal district agrees.
# Station entrances: LTA data as compiled in sgraildata (Jul 2026), filtered to the MRT network open by end-2025.
# MRT/LRT interchange entrances count as MRT; LRT-only stops do not. CCL6 opened in July 2026 and is excluded.
# Walking time = straight-line distance to the nearest MRT entrance x 1.3 (streets aren't straight) / 80 m a minute.
POSTAL_D = {**{s: 1 for s in range(1, 7)}, 7: 2, 8: 2, 14: 3, 15: 3, 16: 3, 9: 4, 10: 4, 11: 5, 12: 5, 13: 5, 17: 6, 18: 7, 19: 7, 20: 8, 21: 8,
            22: 9, 23: 9, **{s: 10 for s in range(24, 28)}, 28: 11, 29: 11, 30: 11, 31: 12, 32: 12, 33: 12, **{s: 13 for s in range(34, 38)},
            **{s: 14 for s in range(38, 42)}, **{s: 15 for s in range(42, 46)}, 46: 16, 47: 16, 48: 16, 49: 17, 50: 17, 81: 17, 51: 18, 52: 18,
            53: 19, 54: 19, 55: 19, 82: 19, 56: 20, 57: 20, 58: 21, 59: 21, **{s: 22 for s in range(60, 65)}, **{s: 23 for s in range(65, 69)},
            69: 24, 70: 24, 71: 24, 72: 25, 73: 25, 77: 26, 78: 26, 75: 27, 76: 27, 79: 28, 80: 28}
HAS_MRT = os.path.exists('project_coords.csv') and os.path.exists('sg_mrt.json')
WALK_BANDS = ['5 min or less', '5-10 min', '10-15 min', 'over 15 min']
if HAS_MRT:
    MRT = json.load(open('sg_mrt.json'))
    pc = pd.read_csv('project_coords.csv', dtype={'postal': str}).set_index('project')
    pc['pd'] = pd.to_numeric(pc.postal.str.zfill(6).str[:2], errors='coerce').map(POSTAL_D)
    pc['ok'] = pc.lat.notna() & ((pc.method == 'name+street') | (pc.pd == pc.district))
    CCL6 = {'Keppel', 'Cantonment', 'Prince Edward Road'}
    ex_ = np.array([[e[2], e[3]] for e in MRT['exits'] if 'singapore-mrt' in e[1] and e[0] not in CCL6])
    def walk_min(la, lo):
        dy = (ex_[:, 0] - la) * 110.57; dx = (ex_[:, 1] - lo) * 111.32 * np.cos(np.radians(1.35))
        return np.hypot(dx, dy).min() * 1000 * 1.3 / 80
    pc['walk'] = [walk_min(a, b) if k else np.nan for a, b, k in zip(pc.lat, pc.lon, pc.ok)]
    if SGMAP:
        from matplotlib.path import Path as MplPath
        town_paths = [(town, MplPath(ring)) for town, rings in SGMAP['planning_areas'].items() for ring in rings]
        def planning_area(lo, la):
            if not np.isfinite(lo) or not np.isfinite(la): return np.nan
            return next((town.title() for town, polygon in town_paths if polygon.contains_point((lo, la))), 'Outside planning areas')
        pc['town'] = [planning_area(lo, la) if ok else np.nan for lo, la, ok in zip(pc.lon, pc.lat, pc.ok)]
        df['town'] = df.project.map(pc.town)
    df['lat'] = df.project.map(pc.lat.where(pc.ok)); df['lon'] = df.project.map(pc.lon.where(pc.ok)); df['walk'] = df.project.map(pc.walk)
    df['walk_band'] = pd.cut(df.walk, [0, 5, 10, 15, 1e9], labels=WALK_BANDS)
    in25_ = pc.index.isin(df.project)                       # report location coverage for the 2025 projects only
    GEO_OK = (pc.ok[in25_].mean(), df.walk.notna().mean(), (pc.pd == pc.district)[in25_ & (pc.method == 'name+street')].mean())
    # The hedonic model plus walking-time bands (base: over 15 min), on the sales with a location. District dummies stay in, so the
    # MRT effect compares homes in the same district.
    mk = df.walk.notna().values
    def mrt_fit(mask):
        Xm = pd.concat([X[mask], dummies(df.walk_band[mask].astype(str), 'W', 'over 15 min')], axis=1)
        Xm = Xm.loc[:, Xm.std() > 0]
        return ols(df.lpsf[mask], Xm, list(Xm.columns), cluster=df.project[mask])
    hm, Hm = mrt_fit(mk)
    hm_lin, _ = ols(df.lpsf[mk], pd.concat([X[mk], df.walk[mk].clip(upper=20).rename('walk_min')], axis=1),
                    list(X.columns) + ['walk_min'], cluster=df.project[mk])
    PER_MIN = hm_lin.loc['walk_min', ['pct', 'pct_lo', 'pct_hi']]
    hk_, _ = ols(df.lpsf[mk], X[mk], list(X.columns), cluster=df.project[mk])     # same sales, no MRT term: do the other effects move?
    MRT_SHIFT = pd.DataFrame({'without MRT term': hk_.loc[['lease_used', 'floor', 'sale_New Sale', 'larea'], 'pct'],
                              'with MRT bands': hm.loc[['lease_used', 'floor', 'sale_New Sale', 'larea'], 'pct']}).round(2)
    print(f"Located {GEO_OK[0]:.0%} of projects ({GEO_OK[1]:.1%} of sales); postal code's district agrees for {GEO_OK[2]:.1%} of name+street matches. "
          f"Median walk {df.walk.median():.0f} min. Each extra minute of walk (up to 20): {PER_MIN.pct:+.2f}% "
          f"(95% CI {PER_MIN.pct_lo:+.2f} to {PER_MIN.pct_hi:+.2f}), same district and features.")
else:
    print('project_coords.csv or sg_mrt.json not found: MRT analysis skipped.')
```

## Step 1 · Sell first, then buy
An upgrader's first decision is the order. Buying before selling looks safer, because you have somewhere to live, but the rules (MAS, IRAS) make it very expensive:
- **Smaller loan:** with the HDB loan still outstanding, banks can lend at most 45% instead of 75%, and at least 25% of the price must be paid in cash.
- **ABSD upfront:** 20% additional buyer's stamp duty is due at purchase. A married couple buying jointly can have it refunded only if they sell the flat within 6 months of buying a completed home (or within 6 months of TOP/CSC for an uncompleted one) and meet IRAS's other conditions.
- **No flat money yet:** none of the S\$625k is available until the flat is sold.

Selling first has its own cost (finding somewhere to live in between), but the funding gap is far smaller. The bars show down payment plus stamp duties, not an all-cash requirement; eligible CPF may fund part of the amount. They assume the maximum permitted loan is approved. Existing debt repayments can reduce it further.

```python
# Sell first or buy first? Cash needed upfront for their two most likely routes under each order.
ROUTES_SEQ = ['B', 'C']; SEQ_PRICE = {}
for code_, name, dist, sale, ten, left, sqft in OPTIONS:          # typical real 2025 price of each route (same filter as the shortlist in Step 5)
    if code_ in ROUTES_SEQ:
        SEQ_PRICE[code_] = similar(dist, sale, ten, left, sqft).price.median()
rows_ = []
for code_ in ROUTES_SEQ:
    p = SEQ_PRICE[code_]
    rows_.append({'route': code_, 'order': 'Sell the flat first', 'Down payment': (1 - LTV['first loan']) * p, "Buyer's stamp duty": bsd(p), 'ABSD (20%, refundable later)': 0.0})
    rows_.append({'route': code_, 'order': 'Buy before selling', 'Down payment': (1 - LTV['second loan']) * p, "Buyer's stamp duty": bsd(p), 'ABSD (20%, refundable later)': 0.20 * p})
seq = pd.DataFrame(rows_); seq['total'] = seq[['Down payment', "Buyer's stamp duty", 'ABSD (20%, refundable later)']].sum(axis=1)
# Compare both purchase orders as cash requirements, with the available proceeds visible.
fig, ax = plt.subplots(figsize=(11.7, 5.8))
bars_ = [('B', 'Sell the flat first'), ('B', 'Buy before selling'),
         ('C', 'Sell the flat first'), ('C', 'Buy before selling')]
ys = [3, 2, 1, 0]
for y, (route, order) in zip(ys, bars_):
    rw = seq[(seq.route == route) & (seq.order == order)].iloc[0]
    if y in [3, 1]: ax.axhspan(y - .45, y + .45, color='#f7f9fc', zorder=0)
    offset = 0
    for comp, color in [('Down payment', C['blue']), ("Buyer's stamp duty", C['blue_l']),
                        ('ABSD (20%, refundable later)', C['orange'])]:
        value = rw[comp] / 1e6
        if value > 0: ax.barh(y, value, left=offset, height=.55, color=color, edgecolor='white', lw=1, zorder=3)
        offset += value
    ax.text(offset + .025, y, sgm(rw.total), va='center', fontsize=12, fontweight='bold', color=C['ink'])
ax.axvline(NET_PROCEEDS / 1e6, color=C['aqua'], lw=2, ls=(0, (5, 3)), zorder=5)
ax.text(NET_PROCEEDS / 1e6 + .02, 3.59, f'Flat proceeds available after sale: {sgm(NET_PROCEEDS)}',
        color=C['aqua'], fontsize=9.5, fontweight='bold', va='center')
ax.set_yticks(ys)
ax.set_yticklabels([f'{"Tampines / Pasir Ris" if r == "B" else "North-east towns"}\n{"Sell first" if o.startswith("Sell") else "Buy first"}' for r, o in bars_], fontsize=10)
ax.set_xlim(0, seq.total.max() / 1e6 * 1.18); ax.set_ylim(-.65, 3.68)
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f'S${v:.1f}M'))
ax.set_xlabel('Down payment and stamp duties (cash + eligible CPF; S$ million)'); ax.grid(axis='x', alpha=.7); ax.grid(False, axis='y')
ax.legend(handles=[Rectangle((0, 0), 1, 1, color=C['blue'], label='Down payment'),
                   Rectangle((0, 0), 1, 1, color=C['blue_l'], label='BSD'),
                   Rectangle((0, 0), 1, 1, color=C['orange'], label='ABSD due upfront if buying first')],
          loc='upper center', bbox_to_anchor=(.5, -.14), ncol=3, fontsize=8.6)
finish(fig, f"Sell first: about {sgm(seq.total.iloc[0])} upfront. Buy first: about {sgm(seq.total.iloc[1])}, before the flat's money arrives",
       'Two resale options with 85–88 years of lease left, priced at their observed 2025 medians. With an outstanding HDB loan, LTV is at most 45%; at least 25% of price must be cash. Totals assume the maximum loan is approved. '
       'The 20% ABSD is paid first and refunded only if IRAS remission conditions are met.',
       source='MAS LTV limits; IRAS BSD and ABSD rules; real 2025 median prices (properties2025.csv)')

display(seq.round(0))
```

## Step 2 · How much can the family spend?
The family has two limits: the loan the bank will approve, and the cash plus eligible CPF available for the down payment and taxes. The lower limit determines the budget.

The chart keeps their income at **S\$16,000 a month** and changes only their flat’s sale price. Their assumed **S\$795,000** sale produces a budget ceiling of about **S\$2.18 million**. This is a ceiling, not a sensible spending target: it leaves no extra reserve for renovation, purchase legal fees, temporary accommodation or emergencies. The cheaper resale options leave room for these costs.

The full income-and-flat-price sensitivity table is printed below the chart. It remains available for checking, without putting every combination in the main visual.


```python
# How sure is the budget? The flat's sale price and the income are assumptions, so vary both.
inc_grid = [12_000, 14_000, 16_000, 18_000, 20_000]; flat_grid = [700_000, 750_000, 795_000, 850_000, 900_000]
sens = pd.DataFrame([[budget_for(i, v)[0] / 1e6 for i in inc_grid] for v in flat_grid], index=flat_grid, columns=inc_grid)
fig, ax = plt.subplots(figsize=(10.8,4.8))
ys=np.arange(len(flat_grid))[::-1]
vals=sens[PERSON['income']].values
for y,flat,ceiling in zip(ys,flat_grid,vals):
    focus=flat==PERSON['hdb_value']
    if focus: ax.axhspan(y-.42,y+.42,color=C['blue_l'],alpha=.55)
    ax.plot([vals.min(),ceiling],[y,y],lw=4,color=C['grey_l'],solid_capstyle='round')
    ax.scatter(ceiling,y,s=120 if focus else 75,color=C['red'] if focus else C['blue'],zorder=3)
    ax.text(ceiling+.035,y,f'S${ceiling:.2f}M'+('  ·  family assumption' if focus else ''),va='center',fontsize=11,fontweight='bold' if focus else 'normal')
ax.set_yticks(ys);ax.set_yticklabels([f'Flat sells for {sgd(v)}' for v in flat_grid])
ax.set_xlim(vals.min()-.08,vals.max()+.5);ax.set_ylim(-.6,4.6)
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda v,_:f'S${v:.1f}M'))
ax.set_xlabel('Maximum home price under the stated financing assumptions')
ax.grid(False,axis='y')
finish(fig,'A higher flat-sale price gives this family more room to buy',
       'Keep their monthly income at S$16,000. Change only what their HDB flat sells for. The highlighted row is the working budget; this ceiling leaves no extra reserve for the new home.',
       source='MAS loan limits; IRAS buyer’s stamp duty; illustrative family profile; full income sensitivity printed below')
display(sens.rename_axis('HDB sale price').round(2))
```

### Lower rates reduce payments, but do not always increase the borrowing limit
The chart now shows **2023–2025**. It prices the same hypothetical loan at each month’s SORA plus an assumed **0.30 percentage-point** bank spread. It does not track an actual mortgage or a particular bank offer.

In 2025, SORA fell about **59%** and the illustrated monthly payment fell about **19%**. The assessed payment stayed at the **4% floor** because the assumed loan rate stayed below it. Banks use the higher of 4% and the thereafter interest rate (the highest applicable rate excluding introductory or promotional rates). Here we use SORA plus the assumed spread as a simple thereafter-rate proxy. This distinction also matters in 2023, when the illustrated rate briefly exceeded 4%.

**Why the options use 1.6%.** The payment tables and recommendation cards use a separate, fixed 1.6% planning scenario so routes can be compared at one rate. December 2025’s embedded SORA plus the assumed spread is about 1.49%, as printed below. The scenario is not a quoted bank offer or a rounded value of that December rate.


```python
# 2025: interest rates fell a lot. What did that do to this buyer's monthly payment, and to how much they can borrow?
months = pd.period_range('2025-01', '2025-12', freq='M')
LOAN = round(LTV['first loan'] * BUDGET, -4); SPREAD = 0.003                  # 75% loan on a home at their budget; SORA + 0.30% (typical 2025)
pay = pd.DataFrame({'sora': [sora_m[m] for m in months]}, index=months.strftime('%b'))
pay['payment'] = [payment(LOAN, (r + SPREAD * 100) / 100, TENURE) for r in pay.sora]
stress_pay = payment(LOAN, STRESS_RATE, TENURE); need = stress_pay / TDSR
# Fixed loan and tenure across months isolate the payment effect of the interest-rate proxy.
drop = 1 - pay.payment.iloc[-1] / pay.payment.iloc[0]
rates_history = pd.DataFrame({'SORA %':sora_m.loc['2023-01':'2025-12']})
rates_history['Illustrative loan rate %']=rates_history['SORA %']+SPREAD*100
rates_history['Monthly payment']= [payment(LOAN,r/100,TENURE) for r in rates_history['Illustrative loan rate %']]
rates_history['Assessment rate %']=np.maximum(STRESS_RATE*100,rates_history['Illustrative loan rate %'])
rates_history['Assessment payment']=[payment(LOAN,r/100,TENURE) for r in rates_history['Assessment rate %']]
fig, ax=plt.subplots(figsize=(11.7,5.5))
x=np.arange(len(rates_history)); vals=rates_history['Monthly payment'].values
ax.fill_between(x,0,vals,color=C['blue_l'],alpha=.45)
ax.plot(x,vals,color=C['blue'],lw=2.8,label='Illustrative payment: SORA + 0.30 percentage points')
ax.plot(x,rates_history['Assessment payment'],color=C['red'],lw=2,ls='--',label='Payment used in borrowing assessment: higher of 4% or assumed thereafter rate')
for rate_i,dy in [(0,-30),(24,-28),(35,-30)]:
    ax.scatter(rate_i,vals[rate_i],color=C['blue'],s=50,zorder=5)
    ax.annotate(f'{rates_history.index[rate_i].strftime("%b %Y")}\n{sgd(vals[rate_i])}',(rate_i,vals[rate_i]),xytext=(0,dy),textcoords='offset points',ha='left' if rate_i==0 else 'right',fontsize=10,color=C['blue_d'])
ax.set_ylim(0,max(rates_history['Assessment payment'].max(),vals.max())*1.17)
ax.set_xlim(-.5,35.5);ax.set_xticks([0,6,12,18,24,30,35]);ax.set_xticklabels(['Jan 2023','Jul','Jan 2024','Jul','Jan 2025','Jul','Dec'])
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v,_:f'S${v:,.0f}'))
ax.set_ylabel('Monthly payment on the same hypothetical loan');ax.grid(False,axis='x')
ax.legend(loc='lower left',bbox_to_anchor=(0,1.01),fontsize=8.6)
finish(fig,f'Lower rates cut the illustrated payment by {sgd(pay.payment.iloc[0]-pay.payment.iloc[-1])} a month during 2025',
       f'2023–2025 history, with a zero baseline. Keep a {sgm(LOAN)} loan and {TENURE}-year term fixed; this is not one borrower’s repayment history. The 0.30-point spread is assumed, not an observed bank offer.',
       source='MAS/SingStat end-of-month 3-month compounded SORA via data.gov.sg; MAS TDSR assessment rules')
display(rates_history.round(2))
print(f'2025 SORA fall: {1-sora_m["2025-12"]/sora_m["2025-01"]:.1%}; payment fall: {drop:.1%}; 4% assessment payment: {sgd(stress_pay)}; income needed: {sgd(need)}.')
```

## Step 3b · How far from the MRT?
We matched project locations to **OneMap** and estimated a walk to the nearest MRT entrance: straight-line distance × 1.3 ÷ 80 m/min. This is a routing proxy, not measured walking time. It covers 99% of projects. MRT/LRT interchange exits count; LRT-only stops do not. The three Circle Line 6 stations that opened in July 2026 are excluded from this 2025 analysis.
- **Broad east-side scan:** 242 of 285 projects with 950–1,150 sqft sales had a 2025 median under the S\$2.18M budget; 49 of those had an estimated MRT walk of at most 5 minutes. This scan mixes sale types and lease lengths, so it is not the family's exact resale shortlist. Across the island, 70% of located projects are within an estimated 10 minutes of MRT.
- **Conditional price association:** among family-size resales outside prime areas, a ≤5-minute MRT walk goes with **+17.1% per sqft** versus >15 minutes, adjusting for district, size, floor, lease, tenure, type, and month (95% CI +12.7% to +21.6%, p < 0.001; 909 near sales, 649 baseline sales). The 5–10-minute gap is +9.6%; the 10–15-minute estimate is +3.4% (CI crosses zero, p = 0.064). The whole-market ≤5-minute association is +8.7% (95% CI +5.3% to +12.3%). Appendix C gives the full table and an LRT sensitivity check.
- **What it does not say:** the estimate compares different projects. Mall links, layouts, facing, and actual walking routes can differ. It is not a station's causal effect or a buyer-specific S\$ price adjustment.

```python
# Map 1: every 2025 project, coloured by its median price per sqft, with the MRT network on top.
# Map 2: this buyer's search area. Each dot is a project with sales like theirs, coloured by the median price of those sales
# against their budget, so you can see which affordable projects sit next to a station.
# MRT colours identify lines; they deliberately retain their familiar network colours.
LINE_COL = {'North South Line':'#d42e12', 'East West Line':'#009645',
            'North East Line':'#9900aa', 'Circle Line':'#fa9e0d',
            'Downtown Line':'#005ec4', 'Thomson-East Coast Line':'#9d5b25'}
if HAS_MRT and SGMAP:
    import matplotlib.patheffects as pe
    from matplotlib.colors import LogNorm
    STX = [s for s in MRT['stations'] if 'singapore-mrt' in s[2] and s[0] not in CCL6]
    def draw_base(ax, ext, lw=2.0):
        ax.grid(False); ax.set_facecolor('#e9eef3')
        for ring in SGMAP['land']: ax.add_patch(Polygon(ring, closed=True, facecolor='#fbfbfa', edgecolor='#c9ced3', lw=0.6, zorder=1))
        for name, rings in SGMAP['planning_areas'].items():
            for ring in rings: ax.add_patch(Polygon(ring, closed=True, fill=False, edgecolor='#e2e5e8', linewidth=0.5, zorder=2))
        for ln in MRT['lines']:
            if ln['name'] not in LINE_COL: continue
            for path_i, path in enumerate(ln['paths']):
                # The first Circle Line path contains the 2026 CCL6 link at its two ends.
                a_ = np.array(path[16:503] if ln['name'] == 'Circle Line' and path_i == 0 else path)
                ax.plot(a_[:, 0], a_[:, 1], color=LINE_COL[ln['name']], lw=lw, alpha=1.0, zorder=3, solid_capstyle='round')
        ax.set_xlim(ext[0], ext[1]); ax.set_ylim(ext[2], ext[3]); ax.set_aspect('equal'); ax.set_xticks([]); ax.set_yticks([])
        for s_ in ax.spines.values(): s_.set_visible(False)
    # ---- map 1: the whole island ----
    pj = df[df.lat.notna()].groupby('project').agg(lat=('lat', 'first'), lon=('lon', 'first'), psf=('psf', 'median'), n=('price', 'size'), walk=('walk', 'first'))
    pcmap = LinearSegmentedColormap.from_list('psf', ['#1765AA', '#5FA0D7', '#C6D9EC', '#F2EFF0', '#DF8B9B', '#BE3F57', '#81213A']); pnorm = LogNorm(1000, 4500)
    EXT1 = (103.62, 104.03, 1.235, 1.465)
    fig, ax = plt.subplots(figsize=(10.5, 7.0)); draw_base(ax, EXT1)
    ax.scatter(pj.lon, pj.lat, s=np.clip(pj.n, 1, 150) ** 0.5 * 4, c=pj.psf, cmap=pcmap, norm=pnorm, alpha=1.0, edgecolor='white', linewidth=0.35, zorder=4)
    stx = np.array([[s[4], s[3]] for s in STX]); ax.scatter(stx[:, 0], stx[:, 1], s=11, color='white', edgecolor=C['ink'], linewidth=0.8, zorder=6)
    ax.add_patch(plt.Rectangle((ZOOM_EXT[0], ZOOM_EXT[2]), ZOOM_EXT[1] - ZOOM_EXT[0], ZOOM_EXT[3] - ZOOM_EXT[2], fill=False, ec=C['ink'], lw=1.2, ls='--', zorder=8))
    ax.text(ZOOM_EXT[0], ZOOM_EXT[3] + 0.003, 'Zoomed in on the next map', fontsize=8, color=C['ink'], va='bottom', zorder=8)
    hs=[plt.Line2D([],[],color=col,lw=2.5,label=name.replace(' Line','')) for name,col in LINE_COL.items()]
    hs.append(plt.Line2D([],[],marker='o',ls='',mfc='white',mec=C['ink'],ms=5,label='MRT station'))
    ax.legend(handles=hs,loc='lower right',ncol=2,fontsize=7.5,frameon=True,framealpha=.97,
              facecolor='white',edgecolor='none',title='MRT lines (LRT not shown)',title_fontsize=8)
    cb = fig.colorbar(plt.cm.ScalarMappable(norm=pnorm, cmap=pcmap), ax=ax, orientation='horizontal', fraction=0.04, pad=0.02, aspect=50, ticks=[1000, 1500, 2000, 3000, 4000])
    cb.ax.set_xticklabels(['S$1,000', 'S$1,500', 'S$2,000', 'S$3,000', 'S$4,000']); cb.outline.set_visible(False); cb.ax.tick_params(labelsize=8.5, colors=C['ink2'])
    cb.set_label('Median 2025 price per sqft of each project (dot size = number of sales)', fontsize=8.8, color=C['ink2'])
    near_share = (pj.walk <= 10).mean()
    finish(fig, f'Prices fall away from the city, and most private homes cluster along the MRT: {near_share:.0%} of projects are within a 10-minute walk',
           f'Each dot is one of the {len(pj):,} projects sold in 2025. Walk = straight-line distance to the nearest MRT entrance x 1.3, at 80 m a minute.',
           source='project locations: OneMap (SLA); MRT network open by end-2025: LTA via sgraildata (Jul 2026), excluding CCL6; prices: properties2025.csv; URA planning areas')
    # ---- map 2: their search area, priced against their budget ----
    zs = df[FIT_MASK & df.lat.notna() & df.lon.between(ZOOM_EXT[0], ZOOM_EXT[1]) & df.lat.between(ZOOM_EXT[2], ZOOM_EXT[3])]
    zp = zs.groupby('project').agg(lat=('lat', 'first'), lon=('lon', 'first'), price=('price', 'median'), n=('price', 'size'), walk=('walk', 'first'))
    zp['gap'] = (zp.price / BUDGET - 1) * 100
    dcmap = LinearSegmentedColormap.from_list('div', ['#153D70', '#2873BE', '#83B0DE', '#F4F1F1', '#E6A1AF', '#C74B63', '#872B40'])
    asp_ = (ZOOM_EXT[3] - ZOOM_EXT[2]) / ((ZOOM_EXT[1] - ZOOM_EXT[0]) * np.cos(np.radians(1.35)))
    fig, ax = plt.subplots(figsize=(10.5, 10.5 * asp_ * 0.92 + 1.6)); draw_base(ax, ZOOM_EXT, lw=3)
    for s in STX:
        if ZOOM_EXT[0] < s[4] < ZOOM_EXT[1] and ZOOM_EXT[2] < s[3] < ZOOM_EXT[3]:
            ax.scatter(s[4], s[3], s=34, color='white', edgecolor=C['ink'], linewidth=1.2, zorder=6)
            ax.annotate(s[0] if s[0] in {'Tampines','Pasir Ris','Bedok','Simei','Upper Changi','Tanah Merah','Expo','Serangoon','Kovan','Hougang','Punggol','Sengkang','Bartley','Kaki Bukit','Bedok Reservoir','Bayshore','Marine Parade','Eunos','Paya Lebar','Ubi'} else '', (s[4], s[3]), xytext=(4, -9), textcoords='offset points', fontsize=7.5, color=C['ink2'], zorder=7, clip_on=True,
                        path_effects=[pe.withStroke(linewidth=2.5, foreground='white')])
    sc = ax.scatter(zp.lon, zp.lat, s=np.clip(zp.n, 1, 40) * 9 + 18, c=zp.gap, cmap=dcmap, norm=TwoSlopeNorm(0, -45, 45),
                    edgecolor=C['ink'], linewidth=0.6, alpha=1.0, zorder=5)
    pick_ = zp[(zp.gap <= 0) & (zp.walk <= 5)].sort_values('n', ascending=False).head(6)
    placed_ = []
    for pj_, r_ in pick_.iterrows():
        if any(abs(r_.lon - a_) < 0.012 and abs(r_.lat - b_) < 0.006 for a_, b_ in placed_): continue
        placed_.append((r_.lon, r_.lat))
        ax.annotate(f'{pj_.title()}\n{sgm(r_.price)} · {r_.walk:.0f} min', (r_.lon, r_.lat),
                    xytext=(8, 16 if r_.lat < ZOOM_EXT[2] + 0.014 else 6), textcoords='offset points', fontsize=7.6,
                    va='bottom' if r_.lat < ZOOM_EXT[2] + 0.014 else 'baseline',
                    color=C['ink'], fontweight='bold', zorder=9, path_effects=[pe.withStroke(linewidth=3, foreground='white')])
    cb = fig.colorbar(sc, ax=ax, orientation='horizontal', fraction=0.045, pad=0.02, aspect=45)
    cb.set_label(f'Median price of their kind of unit in the project vs their {sgm(BUDGET)} budget (%)   ·   blue = under budget, red = over', fontsize=8.8, color=C['ink2'])
    cb.outline.set_visible(False); cb.ax.tick_params(labelsize=8.5, colors=C['ink2'])
    n_fit = int((zp.gap <= 0).sum()); n_fit_near = int(((zp.gap <= 0) & (zp.walk <= 5)).sum())
    finish(fig, MRTMAP_TITLE.format(n_fit=n_fit, n_fit_near=n_fit_near, n_all=len(zp)), MRTMAP_SUB,
           source='project locations: OneMap (SLA); MRT network open by end-2025: LTA via sgraildata (Jul 2026), excluding CCL6; prices: properties2025.csv')
```

## Step 4 · What are you paying for?
We priced homes with a **hedonic regression**: log price per sqft on district, new vs resale, tenure, property type, size, floor, years of lease used and month, with errors clustered by project. It explains **82%** of variation in log price per sqft in the fitting sample, and **77%** when predicting projects excluded from fitting. Clustering allows sales in the same project to share unobserved influences; it does not make the comparison causal. Appendix A shows the checks. Here is the price menu for the family's target home.

```python
# What moves the price of their home? Start from their target home and change one thing at a time.
base = model_price(**REF)
menu = {'New launch, 97 yrs left vs resale, 91 yrs left': model_price(**{**REF, 'sale': 'New Sale', 'lease_left': 97}) - base,
        '10 floors higher': model_price(**{**REF, 'floor': REF['floor'] + 10}) - base,
        '10 fewer years of lease left': model_price(**{**REF, 'lease_left': REF['lease_left'] - 10}) - base,
        f'{MENU_SQFT} sqft bigger': model_price(**{**REF, 'area': REF['area'] + MENU_SQFT}) - base,
}
for k_ in MENU_DISTRICTS: menu[f'Same home in {AREA_NAME[k_]}'] = model_price(**{**REF, 'district': k_}) - base
menu = pd.Series(menu).sort_values()
# Picture it: a price-check receipt. Start from their target home's model price; each line changes one thing.
from matplotlib.patches import Polygon as Poly_
fig, ax = plt.subplots(figsize=(10, 6.0))
items_ = menu.reindex(menu.abs().sort_values(ascending=False).index)
n_ = len(items_); top_ = 9.4; bot_ = top_ - 2.55 - 1.0 - 0.62 * n_ - 1.55
zig_ = [(x_, bot_ - (0.18 if k_ % 2 else 0)) for k_, x_ in enumerate(np.linspace(1.0, 9.0, 41))]
paper_ = [(1.0, top_)] + [(9.0, top_)] + zig_[::-1] + []
ax.add_patch(Poly_([(x_ + 0.12, y_ - 0.12) for x_, y_ in paper_], closed=True, color='#dfe3e8', lw=0, zorder=1))
ax.add_patch(Poly_(paper_, closed=True, facecolor='white', edgecolor='#cfd5dc', lw=1, zorder=2))
MONO = dict(family='DejaVu Sans Mono')
ax.text(5, top_ - 0.55, 'PRICE CHECK', ha='center', va='center', fontsize=15, fontweight='bold', color=C['ink'], zorder=3, **MONO)
ax.text(5, top_ - 1.05, 'one change at a time', ha='center', va='center', fontsize=8.5, color=C['ink2'], zorder=3, **MONO)
import textwrap as tw_
ax.text(1.35, top_ - 1.6, tw_.fill('Start: ' + REF_DESC, 47), ha='left', va='top', fontsize=8.2, color=C['ink2'], zorder=3, linespacing=1.3, **MONO)
y_ = top_ - 2.55
ax.text(1.35, y_, 'MODEL PRICE', ha='left', va='center', fontsize=10.5, fontweight='bold', color=C['ink'], zorder=3, **MONO)
ax.text(8.65, y_, sgm(base), ha='right', va='center', fontsize=12, fontweight='bold', color=C['ink'], zorder=3, **MONO)
ax.plot([1.35, 8.65], [y_ - 0.4] * 2, color=C['grey'], lw=1, ls=(0, (2, 2)), zorder=3)
for k_, (lab_, v_) in enumerate(items_.items()):
    yy_ = y_ - 1.0 - 0.62 * k_
    ax.text(1.35, yy_, (lab_.replace(' vs resale, 91 yrs left', ' (vs resale)') + ' ').ljust(38, '.'), ha='left', va='center', fontsize=8.3, color=C['ink'], zorder=3, **MONO)
    ax.text(8.65, yy_, f'{"+" if v_ > 0 else "−"}S${abs(v_) / 1e3:,.0f}k', ha='right', va='center', fontsize=10, fontweight='bold',
            color=C['blue_d'] if v_ > 0 else C['orange'], zorder=3, **MONO)
yf_ = y_ - 1.0 - 0.62 * n_ + 0.1
ax.plot([1.35, 8.65], [yf_] * 2, color=C['grey'], lw=1, ls=(0, (2, 2)), zorder=3)
ax.text(5, yf_ - 0.42, f'hedonic regression · explains {H["r2"]:.0%} of price variation', ha='center', va='center', fontsize=7.8, color=C['ink2'], zorder=3, **MONO)
bars_ = np.random.default_rng(7).integers(1, 4, 46)
xb_ = 3.3
for w_ in bars_:
    ax.add_patch(Rectangle((xb_, yf_ - 1.15), 0.03 * w_, 0.42, color=C['ink'], lw=0, zorder=3)); xb_ += 0.03 * w_ + 0.045
ax.set_xlim(0.6, 9.4); ax.set_ylim(bot_ - 0.5, top_ + 0.2); ax.set_aspect('equal'); ax.axis('off')
finish(fig, MENU_TITLE, f'Starting point: {REF_DESC}, model price {sgm(base)}. Each line changes one thing and keeps the rest; blue = higher price, red = lower price. '
       'Freehold is left out: freehold buildings have no recorded age, so their premium can\'t be priced fairly.')
```

**How much is lease worth in a family home?** The data records years of **lease left**, not building age. Refitted on family-size resales outside prime areas, each year of lease used goes with about **1.8%** lower prices (1.25% across the whole market). This is an association across projects in 2025, not a measured yearly loss for one home.
- **Check the shape, not just one yearly number:** the bands show differences that a straight line may miss. The classical curvature test does not account for project clustering, so it is not the main evidence for the shape.
- **85–88 years left:** about 4% above 80–84, *below* the straight line.
- **89–92 years left:** about 21% above.
- **93+ years left:** about 40% above, from only 23 resales.

**What it means:** the lower-cost resales in Tampines/Pasir Ris and the north-east (options B and C; 85–88 years left) lie in a lower-priced lease band; project and neighbourhood differences still matter. The 89–92-year band sits about 21% above the 80–84-year band, compared with about 4% for the 85–88-year band (conditional associations).

```python
# How much is each year of lease worth for THIS buyer's kind of home? Lease left in bands (not one straight line),
# fitted on their segment's 99-year resales with district, month, type, size and floor adjusted for. Their routes are marked.
r = df[SEG & (df.tenure == '99-yr') & (df.sale == 'Resale')].copy(); r['left'] = 99 - r.lease_used
LABS = [f'{a}-{b - 1}' if b < 100 else f'{a}+' for a, b in zip(LEASE_EDGES[:-1], LEASE_EDGES[1:])]; LABS[0] = f'under {LEASE_EDGES[1]}'
r['band'] = pd.cut(r.left, LEASE_EDGES, right=False, labels=LABS).astype(str)
BASE_BAND = [l for l in LABS if l.startswith('80')][0]
base_ = [pd.get_dummies(r[c], prefix=c, drop_first=True, dtype=float) for c in ['district', 'month', 'type'] if r[c].nunique() > 1] + [r[['larea', 'floor']]]
Xb = pd.concat(base_ + [dummies(r.band, 'b', BASE_BAND)], axis=1); tband, eb_ = ols(r.lpsf, Xb, list(Xb.columns), cluster=r.project)
Xl = pd.concat(base_ + [r[['lease_used']]], axis=1); tline, el_ = ols(r.lpsf, Xl, list(Xl.columns), cluster=r.project)
Xfull = pd.concat([Xl, dummies(r.band, 'b', BASE_BAND)], axis=1); _, ef_ = ols(r.lpsf, Xfull, list(Xfull.columns), cluster=r.project)   # straight line + band deviations (nested)
rss_b, rss_l = (ef_['e'] ** 2).sum(), (el_['e'] ** 2).sum(); kb, kl = np.linalg.matrix_rank(np.column_stack([np.ones(len(Xfull)), Xfull.values])), np.linalg.matrix_rank(np.column_stack([np.ones(len(Xl)), Xl.values]))
F_LEASE = ((rss_l - rss_b) / (kb - kl)) / (rss_b / (len(r) - kb)); P_LEASE = st.f.sf(F_LEASE, kb - kl, len(r) - kb)   # bands vs straight line (classical F)
MID = {l: (max(a, 60) + min(b, 98)) / 2 for l, a, b in zip(LABS, LEASE_EDGES[:-1], LEASE_EDGES[1:])}; mid0 = MID[BASE_BAND]
curve = pd.DataFrame({'resales': r.band.value_counts().reindex(LABS),
                      'est': [0.0 if b == BASE_BAND else tband.loc['b_' + b, 'pct'] for b in LABS],
                      'low': [0.0 if b == BASE_BAND else tband.loc['b_' + b, 'pct_lo'] for b in LABS],
                      'high': [0.0 if b == BASE_BAND else tband.loc['b_' + b, 'pct_hi'] for b in LABS],
                      'line': [np.expm1(-tline.loc['lease_used', 'coef'] * (MID[b] - mid0)) * 100 for b in LABS]}, index=LABS)
# A lease ruler with confidence intervals shows price associations without implying measured decay.
by_left = {}
for code_, name, dist, sale, ten, left, sqft in OPTIONS:
    if ten == '99-yr' and left: by_left.setdefault(left, []).append(code_)
band_of = lambda yrs: [b for b, a, z in zip(LABS, LEASE_EDGES[:-1], LEASE_EDGES[1:]) if a <= yrs < z][0]
pins_ = {}
for left, codes in by_left.items(): pins_.setdefault(band_of(left), []).extend(codes)
fig, ax = plt.subplots(figsize=(12, 6.5))
ys = np.arange(len(curve))[::-1]
xmin = min(-30, float(curve.low.min()) - 7); xmax = max(50, float(curve.high.max()) + 18)
ax.axvline(0, color=C['ink2'], lw=1.2, zorder=1)
for y, (band, row) in zip(ys, curve.iterrows()):
    if band in pins_: ax.axhspan(y - 0.43, y + 0.43, color='#EFF4FA', zorder=0)
    if band == BASE_BAND:
        ax.scatter(0, y, s=90, color=C['grey'], zorder=4)
        ax.text(2, y, 'reference', va='center', fontsize=10.5, color=C['ink2'])
    else:
        col = C['orange'] if row.est < 0 else C['blue']
        ax.plot([row.low, row.high], [y, y], color=col, lw=3, solid_capstyle='round', zorder=3)
        ax.plot([row.low, row.low], [y - .10, y + .10], color=col, lw=2, zorder=3)
        ax.plot([row.high, row.high], [y - .10, y + .10], color=col, lw=2, zorder=3)
        ax.scatter(row.est, y, s=110, color=col, edgecolor='white', lw=1.4, zorder=4)
        ax.text(row.high + 1.5, y, f'{row.est:+.0f}%  (CI {row.low:+.0f} to {row.high:+.0f})',
                va='center', fontsize=9.7, fontweight='bold', color=C['ink'])
ax.set_yticks(ys)
ax.set_yticklabels([f'{band} years left  ·  {int(row.resales):,} resales' +
                    ({'under 70':'','70-74':'\nOlder Bedok resale (D)','85-88':'\nLower-cost east / north-east resales (B, C)','89-92':'\nLonger-lease resales (F, G)','93+':'\nLease band used by the new-launch option (A)'}.get(band,''))
                    for band, row in curve.iterrows()], fontsize=9.7)
ax.set_xlim(xmin, xmax); ax.set_ylim(-0.65, len(curve) - 0.35)
ax.set_xlabel(f'Adjusted price per sqft vs {BASE_BAND} years left (%)')
ax.grid(axis='x', alpha=0.75); ax.grid(False, axis='y')
for spine in ['top', 'right', 'left']: ax.spines[spine].set_visible(False)
finish(fig,'Shorter and longer leases sit in different price bands',
       f'Compare each band with homes that have {BASE_BAND} years left, shown at zero. Dots show adjusted price differences; lines show 95% confidence intervals. Family-size 99-year resales only; these are differences between projects, not annual losses on one home.')
display(curve.round(2));display(tline.loc[['lease_used']].round(5))
print(f'Classical curvature F={F_LEASE:.2f}; p={P_LEASE:.3g}. This F-test does not account for within-project clustering.')
```

**Is a higher floor worth it?**
- **Floors pay:** in family homes, 10 floors add about 5.5% to the price per sqft, or about 4% comparing units in the same project.
- **No clear size effect within a project:** a 10% larger unit is estimated at +0.28% per sqft (95% CI −0.21% to +0.77%); a modest discount or premium remains possible.

```python
# Does the market-wide regression hold for THIS buyer's kind of home? Refit it on their segment only.
def hedonic(sub):
    parts = [sub[['lease_used', 'floor', 'larea']]] + [pd.get_dummies(sub[c], prefix=c, drop_first=True, dtype=float)
                                                       for c in ['district', 'month', 'tenure', 'sale', 'type'] if sub[c].nunique() > 1]
    Xx = pd.concat(parts, axis=1); t, _ = ols(sub.lpsf, Xx, list(Xx.columns), cluster=sub.project); return t
def within(sub):          # project fixed effects: compares units inside the same project only
    Xx = pd.concat([sub[['floor', 'larea']], pd.get_dummies(sub.month, prefix='m', drop_first=True, dtype=float)], axis=1)
    Xd = Xx - Xx.groupby(sub.project.values).transform('mean'); yd = sub.lpsf - sub.groupby('project').lpsf.transform('mean')
    t, _ = ols(yd, Xd, list(Xx.columns), cluster=sub.project, const=False); return t
def effects(t):
    out = {}
    for var, lab, scale in [('lease_used', 'Each year of lease used', 1), ('floor', 'Every 10 floors higher', 10), ('larea', 'A 10% bigger unit (per sqft)', np.log(1.1))]:
        if var in t.index:
            b_, s_ = t.loc[var, 'coef'] * scale, t.loc[var, 'se'] * scale
            out[lab] = (np.expm1(b_) * 100, np.expm1(b_ - 1.96 * s_) * 100, np.expm1(b_ + 1.96 * s_) * 100)
    return out
SEG_TABS = {'Whole market': hedonic(df), SEG_LABEL: hedonic(df[SEG]), 'Same project only': within(df[SEG])}
seg_res = {label: effects(tab) for label, tab in SEG_TABS.items()}
SEG_P = pd.DataFrame({label: {effect: tab.loc[var, 'p'] if var in tab.index else np.nan
                              for var, effect in [('lease_used', 'Each year of lease used'),
                                                  ('floor', 'Every 10 floors higher'),
                                                  ('larea', 'A 10% bigger unit (per sqft)')]}
                      for label, tab in SEG_TABS.items()})
# Picture it: a condo block, one storey per floor band, using units compared inside the SAME building (whole market, floors 1-40).
fl_ = df[(df.floor >= 1) & (df.floor <= 36)].copy()
fl_ = fl_[fl_.groupby(['project', 'sale', 'month']).project.transform('size') >= 2]
FB_ = pd.get_dummies(fl_.level).astype(float).drop(columns='01 to 05')
FX_ = pd.concat([FB_, fl_[['larea']]], axis=1); fgrp_ = [fl_.project.values, fl_.sale.values, fl_.month.values]
ftb_, _ = ols(fl_.lpsf - fl_.groupby(['project', 'sale', 'month']).lpsf.transform('mean'), FX_ - FX_.groupby(fgrp_).transform('mean'),
              list(FX_.columns), cluster=fl_.project, const=False)
FLOOR_BANDS = pd.DataFrame({k_: [0.0] + [ftb_.loc[b_, c_] for b_ in FB_.columns] for k_, c_ in [('est', 'pct'), ('lo', 'pct_lo'), ('hi', 'pct_hi')]},
                           index=['01 to 05'] + list(FB_.columns))
LOW_PRICE = model_price(**{**REF, 'floor': 3})          # their target home, priced on floors 1-5
fig, axd_ = plt.subplot_mosaic([['b', 'f'], ['b', 's']], figsize=(11.5, 6.4), width_ratios=[1.45, 1])
axb = axd_['b']
draw_building(axb, list(FLOOR_BANDS.index), FLOOR_BANDS.est.values, FLOOR_BANDS.lo.values, FLOOR_BANDS.hi.values, base_price=LOW_PRICE,
              title='Same project, higher floor (vs floors 1–5)')
labs = list(seg_res); cols_ = [C['grey'], C['orange'], C['blue']]
for k_, m in enumerate(['Every 10 floors higher', 'A 10% bigger unit (per sqft)']):
    ax = axd_['fs'[k_]]
    for i, k in enumerate(labs):
        if m in seg_res[k]:
            e, lo, hi = seg_res[k][m]
            ax.plot([lo, hi], [i, i], color=cols_[i], lw=4, solid_capstyle='round', alpha=0.45); ax.scatter(e, i, color=cols_[i], s=70, zorder=3, edgecolor=C['bg'], linewidth=1.5)
            ax.text(hi, i, f'  {e:+.1f}%', ha='left', va='center', fontsize=9, color=C['ink'], fontweight='bold' if i == 1 else 'normal')
    ax.axvline(0, color=C['ink2'], lw=0.8); ax.set_title(m, fontsize=10, color=C['ink'], loc='left', fontweight='bold'); ax.grid(False, axis='y')
    ax.set_ylim(-0.6, len(labs) - 0.4); ax.set_xlabel('% change in price per sqft (95% CI)', fontsize=8.5)
    ax.set_yticks(range(len(labs))); ax.set_yticklabels([l.replace(', ', ',\n') for l in labs], fontsize=8.5, color=C['ink'])
    ax.yaxis.tick_right(); x0_, x1_ = ax.get_xlim(); ax.set_xlim(x0_, x1_ + (x1_ - x0_) * 0.25)
finish(fig, SEG_TITLE, f'Left: units on different floors of the same project, sold in the same month and sale type ({len(fl_):,} sales); each storey is a 5-floor band. '
       f'Right: the regression refitted on {SEG_LABEL.lower()} only ({SEG.sum():,} sales, {df[SEG].project.nunique()} projects) vs the whole market and within buildings. '
       'S$ = the band premium applied to the model price of their target home on floors 1-5.')
```

**Is a brand-new launch worth its premium?** In the 2025 99-year sample, new launches have a raw geometric-mean price-per-sqft gap of 55% above resales. Matched to resales in the same district, size band, floor band and quarter (6,235 of 9,730 new units matched), the gap is still **+42%**. That's a conditional association, not the effect of newness: matched homes can still differ in project quality, and lease and newness can't be separated. New launches have 92–99 years left and no resale has more than 94. In the only 3 projects with both kinds of sale, the gap is +6%, with too few units to decide.

**How to read the four rows:** they are different comparisons, not a sequence that isolates the causes of the premium. The raw gap has no uncertainty interval shown. Matching still leaves unrecorded differences; adding lease requires extrapolation; the same-project comparison has little overlap.

**One headline, several questions.** Use the **42% matched gap** as the main 2025 comparison. The 55% raw gap describes the whole 99-year new/resale sample before matching; the 15% estimate adds lease controls and requires extrapolation; the 6% same-project estimate has very little overlap. The 26% coefficient in Appendix A uses the broader hedonic model, while Step 8’s annual matched gaps omit floor bands. None measures a pure effect of buying new. Appendix C puts the estimates, intervals and comparison types in one reference table.


```python
# Cause or coincidence? Is a brand-new launch worth its premium over a longer-lease resale (route A vs B)?
# Coarsened exact matching (CEM): compare new and resale 99-year units only within the same district, size band,
# floor band and quarter. Then hold lease left fixed (regression), then compare inside the same project.
def cem(sub, treat, strata, B=200, seed=0):
    s_ = sub.assign(T=treat.astype(int).values, key=strata.values)
    g = s_.groupby(['key', 'T']).lpsf.agg(['mean', 'size']).unstack('T').dropna()
    est_ = np.average(g[('mean', 1)] - g[('mean', 0)], weights=g[('size', 1)])
    rng = np.random.default_rng(seed); projs = s_.project.unique(); boots = []
    for _ in range(B):                                            # bootstrap whole projects for the CI
        w_ = pd.Series(rng.choice(projs, len(projs))).value_counts()
        sb = s_[s_.project.isin(w_.index)]; wt = sb.project.map(w_)
        gb = sb.assign(w=wt, wy=wt * sb.lpsf).groupby(['key', 'T'])[['w', 'wy']].sum().unstack('T').dropna()
        boots.append(np.average(gb[('wy', 1)] / gb[('w', 1)] - gb[('wy', 0)] / gb[('w', 0)], weights=gb[('w', 1)]))
    lo, hi = np.percentile(boots, [2.5, 97.5])
    return {'gap %': np.expm1(est_) * 100, 'low': np.expm1(lo) * 100, 'high': np.expm1(hi) * 100,
            'matched': int(g[('size', 1)].sum()), 'of': int(s_['T'].sum())}
nr = df[df.sale.isin(['New Sale', 'Resale']) & (df.tenure == '99-yr')].copy(); nr['new'] = (nr.sale == 'New Sale')
qtr = 'Q' + ((nr.month - 1) // 3 + 1).astype(str)
steps = {'Raw gap': {'gap %': np.expm1(nr[nr.new].lpsf.mean() - nr[~nr.new].lpsf.mean()) * 100}}
steps['Matched on district, size, floor, quarter'] = cem(nr, nr.new, nr.district.astype(str) + '|' + (nr.area // 200).astype(str) + '|' + nr.level + '|' + qtr)
Xn = pd.concat([nr[['new', 'larea', 'floor', 'lease_used']].astype(float), pd.get_dummies(nr.district, prefix='D', drop_first=True, dtype=float),
                pd.get_dummies(nr.month, prefix='m', drop_first=True, dtype=float)], axis=1)
tn, _ = ols(nr.lpsf, Xn, list(Xn.columns), cluster=nr.project)
steps['Regression adding lease left (extrapolated: no resale has 95+ years)'] = {'gap %': tn.loc['new', 'pct'], 'low': tn.loc['new', 'pct_lo'], 'high': tn.loc['new', 'pct_hi']}
both = nr.groupby('project').new.transform('nunique') == 2; sb = nr[both]
Xs_ = pd.concat([sb[['new', 'larea', 'floor']].astype(float), pd.get_dummies(sb.month, prefix='m', drop_first=True, dtype=float)], axis=1)
Xs_ = Xs_ - Xs_.groupby(sb.project.values).transform('mean'); ys_ = sb.lpsf - sb.groupby('project').lpsf.transform('mean')
tbb, _ = ols(ys_, Xs_, list(Xs_.columns), cluster=sb.project, const=False)
steps[f'Same project only ({int(sb.new.sum())} new units)'] = {'gap %': tbb.loc['new', 'pct'], 'low': tbb.loc['new', 'pct_lo'], 'high': tbb.loc['new', 'pct_hi']}
causal = pd.DataFrame(steps).T
fig,(labels_ax,ax)=plt.subplots(1,2,figsize=(12,5.9),sharey=True,gridspec_kw={'width_ratios':[1.2,1.8],'wspace':.08})
labels_ax.axis('off');labels_ax.set_xlim(0,1)
method_labels=[('All 99-year new sales vs resales','Raw geometric-mean gap; no adjustment'),('Homes matched on recorded features','Same district, size band, floor band and quarter'),('Regression including lease left','Extrapolates beyond resale lease coverage'),('Sales in the same project','Only 13 new sales; too little overlap')]
for y,((name,row),(heading,note)) in zip([3,2,1,0],zip(causal.iterrows(),method_labels)):
    labels_ax.text(0,y+.13,heading,fontsize=10.5,fontweight='bold',va='center')
    labels_ax.text(0,y-.17,note,fontsize=8.6,color=C['ink2'],va='center')
    color=C['red'] if y==3 else C['blue']
    if pd.notna(row['low']):
        ax.errorbar(row['gap %'],y,xerr=[[row['gap %']-row.low],[row.high-row['gap %']]],fmt='o',color=color,ms=9,capsize=5,lw=2.5)
        label=f"{row['gap %']:+.0f}%  (95% CI {row.low:+.0f} to {row.high:+.0f})"
        xx=row.high+2
    else:
        ax.scatter(row['gap %'],y,s=95,color=color);label=f"{row['gap %']:+.0f}%  descriptive";xx=row['gap %']+2
    ax.text(xx,y,label,fontsize=10,va='center',fontweight='bold')
ax.axvline(0,color=C['ink2'],lw=1);ax.set_xlim(min(-20,causal.low.min()-5),95);ax.set_ylim(-.6,3.6)
ax.set_yticks([]);ax.set_xlabel('New-sale price per sqft above resale (%)');ax.grid(False,axis='y')
finish(fig,f"Matching narrows the new-sale gap from {causal['gap %'].iloc[0]:.0f}% to {causal['gap %'].iloc[1]:.0f}%, but does not explain its cause",
       'Four comparisons answer different questions and use different samples. The intervals describe statistical uncertainty; they do not remove differences in quality, location or lease. These are not steps in a causal decomposition.')
display(causal.round(2))
```

## Step 5 · Your shortlist
Seven realistic routes for a 1,050 sqft home. Each row shows real 2025 sales of similar units: same district, sale type and lease band, size within 100 sqft, floors 6–20. The diamond is the district-level model price (MRT walk is not included in this price model). Cash and monthly figures in Appendix C use each route's observed median; the thin bar is an 80% band for a single unit; for projects the model hadn't seen, it covered 82% of resales but only 65% of new sales, so it's too narrow for new launches.
- **Lower upfront spend:** B (85–88 years left, Tampines/Pasir Ris, median about S\$1.62M, about S\$170k spare at 2025 prices; Tampines alone has 10 sales at about S\$1.67M) and C (the north-east towns, about S\$1.69M, about S\$150k spare).
- **More lease, more money:** F (Tampines, 89–93 years left) is about S\$1.91M and still fits. G (Geylang/Eunos, roughly 6 km from the CBD at the district centre) is about S\$2.40M, over budget.
- **Over budget:** A (new launch, about S\$2.52M) and E (Katong/Marine Parade freehold, mostly Marine Parade, about S\$2.36M).
- **Not enough evidence:** D (Bedok, older lease) has only 2 similar sales, so it's shown for completeness only.
- **The catch with B and C:** their similar sales are a median 22 and 11 minutes' estimated walk from the MRT. F is about 12 minutes, and E and G about 3. B's 21 comparables split between Pasir Ris (11) and Tampines (10); town mix and walking distance both matter. The MRT association in Step 3b is not a causal price adjustment.

**Why the map and shortlist differ.** Route G’s S\$2.40M observed median is above its S\$1.94M model prediction; F is S\$1.91M versus S\$1.80M, and E is S\$2.36M versus S\$2.18M. These are different quantities, not interchangeable prices. The observed medians determine the route’s upfront costs and budget status. Their discrepancies are a reason to inspect actual projects rather than assume a model-priced home is available. The comparison table below prints both.

**The district average hides different choices.** Within B, 11 Pasir Ris sales had a median S\$1.54M; 10 Tampines sales had a median S\$1.67M. Within C, the 9 Serangoon sales had a median S\$2.21M (just over budget), versus about S\$1.62–1.73M in Hougang, Sengkang and Punggol. For E, Marine Parade alone had a S\$2.70M median. Appendix C separates the towns; the route medians should not be read as prices available everywhere in the district.

```python
# Their shortlist: each route priced by the regression, and every real 2025 sale of a similar unit (same district, sale type
# and tenure, lease within 6 years, size within 100 sqft, floors 6-20), against their budget.
rows, comps = [], {}
for code_, name, dist, sale, ten, left, sqft in OPTIONS:
    p = model_price(dist, sale, ten, sqft, OPT_FLOOR, left if left else 99); c_ = similar(dist, sale, ten, left, sqft); comps[code_] = c_
    comp_price = c_.price.median() if len(c_) else p
    rows.append({'route': code_, 'option': name, 'district': dist, 'sqft': sqft, 'lease left': left if left else 'freehold', 'district-centre km to CBD': round(km_to_cbd(dist), 1),
                 'model price': p, 'real median': c_.price.median(), 'real sales': len(c_), 'cash + CPF needed': 0.25 * comp_price + bsd(comp_price),
                 'monthly @1.6%': payment(LTV['first loan'] * comp_price, PAYMENT_RATE, TENURE), 'fits budget': (c_.price.median() if len(c_) else p) <= BUDGET,
                 'walk to MRT (min)': round(c_.walk.median(), 1) if HAS_MRT and len(c_) else np.nan})
opt = pd.DataFrame(rows).set_index('route')
# A present-day sensitivity, not 2026 project transactions: shift 2025 comparables by the
# market-wide URA non-landed index through the provisional 2026-Q3 flash estimate.
INDEX_PROXY_FACTOR = ACTUAL_2026_Q3_FLASH / q.ppi['2025-Q4']
opt['2026-Q3 index proxy'] = opt['real median'] * INDEX_PROXY_FACTOR
opt['cash + CPF at proxy'] = 0.25 * opt['2026-Q3 index proxy'] + opt['2026-Q3 index proxy'].map(bsd)
opt['fits at proxy'] = opt['2026-Q3 index proxy'] <= BUDGET
# One row per route: real comparable sales, their median, model price and model uncertainty.
fig, (lab_ax, ax) = plt.subplots(1, 2, figsize=(12.6, 7.0), sharey=True,
                                 gridspec_kw={'width_ratios': [1.25, 1.85], 'wspace': .03})
rng_ = np.random.default_rng(1); yy = np.arange(len(opt))[::-1]
lab_ax.set_xlim(0, 1); lab_ax.set_ylim(-0.7, len(opt) - 0.3); lab_ax.axis('off')
ax.axvspan(0, BUDGET / 1e6, color=C['blue_l'], alpha=0.27, lw=0)
for y_, (code_, rw) in zip(yy, opt.iterrows()):
    c_ = comps[code_]
    for panel_ in (lab_ax, ax):
        if int(y_) % 2 == 0: panel_.axhspan(y_ - .46, y_ + .46, color='#f7f9fc', zorder=0)
    label_ = {'A':'New launch · Tampines / Pasir Ris','B':'Resale · Tampines / Pasir Ris','C':'Resale · North-east towns','D':'Older resale · Bedok','E':'Freehold · Katong / Marine Parade','F':'Longer lease · Tampines / Pasir Ris','G':'Longer lease · Geylang / Eunos'}[code_]
    lab_ax.text(.02, y_ + .14, f'{code_}. {label_}', ha='left', va='center', fontsize=10.5,
                fontweight='bold', color=C['ink'])
    lease_txt = rw['lease left'] if rw['lease left'] == 'freehold' else f'{rw["lease left"]} yrs lease left'
    walk_txt = f' · {rw["walk to MRT (min)"]:.0f} min MRT' if pd.notna(rw['walk to MRT (min)']) else ''
    lab_ax.text(.02, y_ - .16, f'{rw.sqft:,} sqft · {lease_txt}{walk_txt} · {rw["real sales"]} sales',
                ha='left', va='center', fontsize=8.8, color=C['ink2'])
    ax.scatter(c_.price / 1e6, y_ + rng_.uniform(-.15, .15, len(c_)), s=17,
               color=C['blue'] if rw['fits budget'] else C['orange'], alpha=.3, lw=0, zorder=2)
    median_ = rw['real median'] / 1e6
    if len(c_): ax.scatter(median_, y_, marker='|', s=440, color=C['ink'], linewidth=2.5, zorder=5)
    s_pi = SD_BY_SALE[OPTIONS[list(opt.index).index(code_)][3]]
    lo_pi, hi_pi = rw['model price'] * np.exp(-1.2816 * s_pi), rw['model price'] * np.exp(1.2816 * s_pi)
    opt.loc[code_, 'PI low'], opt.loc[code_, 'PI high'] = lo_pi, hi_pi
    ax.plot([lo_pi / 1e6, hi_pi / 1e6], [y_ - .30, y_ - .30], color=C['grey'], lw=1.8, zorder=3)
    ax.scatter(rw['model price'] / 1e6, y_ - .30, marker='D', s=40, color='white', edgecolor=C['ink'], lw=1.4, zorder=5)
    ax.text(median_, y_ + .27, sgm(rw['real median']), ha='center', va='bottom', fontsize=9, fontweight='bold', color=C['ink'])
ax.axvline(BUDGET / 1e6, color=C['red'], lw=2, zorder=6)
ax.text(BUDGET / 1e6, yy[0] + .54, f' BUDGET {sgm(BUDGET)}', color=C['red'], fontsize=9.5,
        fontweight='bold', va='center')
lo_x = min(opt['model price'].min(), min(c_.price.min() for c_ in comps.values() if len(c_))) / 1e6
hi_x = max(opt['model price'].max(), max(c_.price.max() for c_ in comps.values() if len(c_))) / 1e6
ax.set_xlim(lo_x * .91, hi_x * 1.05); ax.set_ylim(-.7, len(opt) - .25)
ax.set_yticks([]); ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f'S${v:.1f}M'))
ax.set_xlabel('2025 sale price (S$ million)'); ax.grid(axis='x', alpha=.7); ax.grid(False, axis='y')
ax.scatter([], [], color=C['blue'], s=24, alpha=.5, label='Real comparable sale')
ax.scatter([], [], marker='|', color=C['ink'], s=160, linewidth=2.5, label='Real median')
ax.scatter([], [], marker='D', color='white', edgecolor=C['ink'], s=35, label='Model price; grey line = 80% unit band')
ax.legend(loc='upper center', bbox_to_anchor=(.48, -.13), ncol=3, fontsize=8.3)
finish(fig, OPT_TITLE,
       f'Each route uses real 2025 sales of similar units. Vertical black mark = observed median; diamond and grey line = model price and 80% single-unit interval for a {OPT_DESC}. '
       'Walk times are estimates. Appendix C prices cash needed and monthly payments from observed medians, not model predictions.')
```

**An inspection list.** These 99-year projects were picked using only January–June residuals and a January–June median within budget. Freehold is left out because its building age isn't recorded.
- **Later-sales check:** we fitted the model (now including walking time to the MRT) on January–June sales only and picked the 8 lowest projects using only H1 eligibility and price. All 8 resold in July–December; relative to the H2 market mean, their prices remained below the model estimate. The model is re-centred using the July–December market average, so this is a retrospective check rather than a forecast available in June. Appendix C shows each result.
- **Mostly far from stations:** 7 of the 8 are 12–25 minutes' estimated walk from the MRT. Walking time over 15 minutes is one band in the model, so a very long walk can still look like a discount.
- **What a gap can mean:** a persistent gap can be value, or drawbacks the model can't see (noise, layout, condition, a very long walk). Visit before believing it.

The horizontal scale is now **percentage below the model estimate**. Zero means equal to the estimate; farther right means a larger shortfall. Blue marks selection-period sales and red outlines mark later sales. The detailed full-year fit remains in Appendix C.


```python
# Value finder: select projects using Jan-Jun residuals; show their Jul-Dec residuals as a holdout check.
# Residual = actual price minus model price. A project that's consistently below the model is either good value or has
# drawbacks the model can't see (noise, layout, facing, condition), so treat it as a shortlist to inspect.
yl_ = df.lpsf.values
# With MRT data, the model also includes walking time (bands), so a project isn't flagged just for being far from a station.
if HAS_MRT:
    Wv = pd.get_dummies(df.walk_band.astype(str)).reindex(columns=['5 min or less', '5-10 min', '10-15 min'], fill_value=0).astype(float).values
    Xa_v = np.column_stack([Xa, Wv])
else: Xa_v = Xa
res_all = pd.Series(yl_ - Xa_v @ np.linalg.lstsq(Xa_v, yl_, rcond=None)[0], index=df.index)
VF_POOL = df[VF_MASK].assign(resid=res_all[VF_MASK])
vf_all = VF_POOL.groupby('project').agg(district=('district', 'first'), deals=('price', 'size'), median_price=('price', 'median'),
                                    lease_left=('lease_used', lambda s_: 99 - s_.median()), tenure=('tenure', 'first'),
                                    vs_model=('resid', lambda r_: np.expm1(r_.mean()) * 100))
vf_whole = vf_all[(vf_all.deals >= 6) & (vf_all.median_price <= BUDGET)].sort_values('vs_model').head(8)
# Out-of-sample check (guards against picking projects on noise): fit the model on Jan-Jun sales only, pick the 8 lowest projects on
# Jan-Jun residuals, then score their Jul-Dec sales with that same Jan-Jun model (no refitting), re-centred for the market's drift.
h1_ = (df.month <= 6).values; nomon = [k_ for k_, nm in enumerate(['const'] + list(X.columns)) if not nm.startswith('m_')] + list(range(Xa.shape[1], Xa_v.shape[1]))
Xh = Xa_v[:, nomon]; b_h1 = np.linalg.lstsq(Xh[h1_], yl_[h1_], rcond=None)[0]; res_h = yl_ - Xh @ b_h1
res_h = np.where(h1_, res_h - res_h[h1_].mean(), res_h - res_h[~h1_].mean())
pool_ = df[VF_MASK].assign(half=np.where(df[VF_MASK].month <= 6, 'H1', 'H2'), rh=res_h[VF_MASK.values])
hh = pool_.groupby(['project', 'half']).rh.mean().unstack(); cnt_ = pool_.groupby(['project', 'half']).size().unstack().fillna(0)
h1_stats = pool_[pool_.half == 'H1'].groupby('project').agg(n=('price', 'size'), median_price=('price', 'median'))
h1_eligible = h1_stats[(h1_stats.n >= 3) & (h1_stats.median_price <= BUDGET)].index
ok_ = cnt_[(cnt_.get('H1', 0) >= 3) & (cnt_.get('H2', 0) >= 3)].index.intersection(h1_eligible); hh_ok = hh.loc[ok_]
R_HALF = hh_ok.corr().iloc[0, 1]; pick_ = hh.loc[h1_eligible, 'H1'].nsmallest(8).index
PERSIST = (np.expm1(hh.loc[pick_, 'H1'].mean()) * 100, np.expm1(hh.loc[pick_, 'H2'].mean()) * 100,
           len(h1_eligible), int(hh.loc[pick_, 'H2'].notna().sum()))
vf = vf_all.loc[pick_].copy()
vf['H1'] = np.expm1(hh.loc[pick_, 'H1']) * 100; vf['H2'] = np.expm1(hh.loc[pick_, 'H2']) * 100
drop_ = (~df.project.isin(cook_proj.index[:10])).values     # influence check: drop the 10 most influential projects, refit, redo the shortlist
b_d = np.linalg.lstsq(Xa_v[drop_], yl_[drop_], rcond=None)[0]; res_d = pd.Series(yl_ - Xa_v @ b_d, index=df.index)
vf2 = df[VF_MASK & drop_].assign(r2_=res_d).groupby('project').agg(deals=('price', 'size'), median_price=('price', 'median'), vs=('r2_', lambda r_: np.expm1(r_.mean()) * 100))
vf2 = vf2[(vf2.deals >= 6) & (vf2.median_price <= BUDGET)].sort_values('vs').head(8); SAME = len(set(vf2.index) & set(vf_whole.index))
# Keep each project's descriptive details beside its name, with one shared comparison panel.
# Explicit positions reserve separate space for the title, reading guide and column headings.
fig=plt.figure(figsize=(11.8,6.4))
lab_ax=fig.add_axes([.022,.145,.405,.625])
ax=fig.add_axes([.445,.145,.535,.625],sharey=lab_ax)
yy=np.arange(len(vf))[::-1];lab_ax.axis('off');lab_ax.set_xlim(0,1)
for i,(y,(pj,rw)) in enumerate(zip(yy,vf.iterrows())):
    if i%2==0:
        for panel in [lab_ax,ax]: panel.axhspan(y-.46,y+.46,color='#F3F6FA',zorder=0)
    h1,h2=-rw.H1,-rw.H2
    ax.plot([h1,h2],[y,y],color='#B6C3D2',lw=2.8,zorder=2)
    ax.scatter(h1,y,s=62,color=C['blue'],zorder=3)
    ax.scatter(h2,y,s=68,facecolor='white',edgecolor=C['red'],lw=2.2,zorder=4)
    # Above/below labels stay legible when the two estimates are close.
    ax.annotate(f'{h1:.0f}%',(h1,y),xytext=(0,10),textcoords='offset points',
                ha='center',fontsize=9,color=C['blue'],fontweight='bold')
    ax.annotate(f'{h2:.0f}%',(h2,y),xytext=(0,-14),textcoords='offset points',
                ha='center',fontsize=9,color=C['red'],fontweight='bold')
    lab_ax.text(.035,y+.12,pj.title(),fontsize=10.4,fontweight='bold',va='center')
    walk=pc.walk.get(pj,np.nan) if HAS_MRT else np.nan
    lab_ax.text(.035,y-.17,f'{sgm(rw.median_price)} · {rw.lease_left:.0f} years left · {walk:.0f} min MRT',
                fontsize=8.3,color=C['ink2'],va='center')
ax.set_xlim(0,max(-vf[['H1','H2']].min().min()+3,30));ax.set_ylim(-.65,len(vf)-.35);ax.set_yticks([])
ax.set_xlabel('Price below the model estimate (%)',labelpad=8);ax.grid(False,axis='y')
ax.set_axisbelow(True)
fig.text(.036,.795,'PROJECT · 2025 MEDIAN PRICE / LEASE / WALK',fontsize=8.5,color=C['ink2'],va='center')
ax.scatter([],[],s=62,color=C['blue'],label='Jan–Jun: selection')
ax.scatter([],[],s=68,facecolor='white',edgecolor=C['red'],lw=2.2,label='Jul–Dec: check')
ax.legend(loc='center',bbox_to_anchor=(.715,.795),bbox_transform=fig.transFigure,ncol=2,fontsize=9,borderaxespad=0)
import textwrap
fig.text(.012,.98,'All eight shortlisted projects remained below the model estimate in July–December',
         fontsize=13.5,fontweight='bold',color=C['ink'],va='top')
fig.text(.012,.923,textwrap.fill('Blue dots show the January–June selection; red rings show the July–December check. Further right means a larger gap below the model. The check uses the later market average, so it is retrospective, not a live forecast. Unrecorded drawbacks may explain lower prices.',155),
         fontsize=10,color=C['ink2'],va='top',linespacing=1.3)
fig.text(.012,.018,textwrap.fill('Source: properties2025.csv; OneMap project locations; LTA/OpenStreetMap MRT exits via sgraildata. Sales counts and district names are in the table below.',190),
         fontsize=7.8,color=C['grey'],va='bottom')
plt.show()
display(vf.round(2))
```

## Step 6 · When?
**An illustration, not a prediction.** What matters to an upgrader is the top-up: the target home's price minus the flat's.
- **How it's built:** we scaled route B's 2025 median and the flat's estimated value by the URA and HDB indices.
- **What it shows:** the top-up was about S\$0.83M at end-2025 and about S\$0.85M by mid-2026. The trend rule puts it about the same or slightly higher by end-2027, with wide ranges.
- **The takeaway:** the observed index gap moved against this illustration through mid-2026, but the forecast range is too wide to call the best buying month. Plan around cash, a suitable unit, and the sale sequence.

```python
# The upgrader's real question: will the gap between their target home and their flat grow? Forecast both with the trend rule.
tr_h, err_h = trend_fan('hdb', 'hdb'); tr_p, err_p = trend_fan('ppi', 'ppi')
tgt0 = opt.loc['B', 'real median']; flat0 = PERSON['hdb_value']; hq = np.arange(0, 9)
def path(v0, tr_, err_, k=None):
    return np.array([v0 * np.exp((tr_ * h + (0 if (k is None or h == 0) else err_[h][k])) / 100) for h in hq])
tp, tpl, tph = path(tgt0, tr_p, err_p), path(tgt0, tr_p, err_p, 0), path(tgt0, tr_p, err_p, 1)
fp, fpl, fph = path(flat0, tr_h, err_h), path(flat0, tr_h, err_h, 0), path(flat0, tr_h, err_h, 1)
fig, ax = plt.subplots(figsize=(10, 4.3)); lab_q = ['2025-Q4'] + horizon
ax.fill_between(hq, tpl / 1e6, tph / 1e6, color=C['blue'], alpha=0.12, lw=0); ax.plot(hq, tp / 1e6, color=C['blue'], lw=2.4, label=f'Target home: route B (85-88 yrs lease left, Tampines), 2025 median {sgm(tgt0)}')
ax.fill_between(hq, fpl / 1e6, fph / 1e6, color=C['violet'], alpha=0.12, lw=0); ax.plot(hq, fp / 1e6, color=C['violet'], lw=2.4, label=f'Their 5-room HDB flat, 4Q 2025 {sgm(flat0)}')
for h_ in (0, 8):
    ax.annotate('', xy=(h_, tp[h_] / 1e6), xytext=(h_, fp[h_] / 1e6), arrowprops=dict(arrowstyle='<->', color=C['ink2'], lw=1.2))
    ax.text(h_ + (0.15 if h_ == 0 else -0.15), (tp[h_] + fp[h_]) / 2e6, f'top-up needed\n{sgm(tp[h_] - fp[h_])}', ha='left' if h_ == 0 else 'right', fontsize=9.5, color=C['ink'], fontweight='bold')
act_q = ['2026-Q1', '2026-Q2']; tgt_act = np.array([tgt0 * q.ppi[k] / q.ppi['2025-Q4'] for k in act_q]); flat_act = np.array([flat0 * q.hdb[k] / q.hdb['2025-Q4'] for k in act_q])
ax.scatter([1, 2], tgt_act / 1e6, color=C['blue'], s=45, zorder=5, edgecolor=C['bg'], linewidth=1.5); ax.scatter([1, 2], flat_act / 1e6, color=C['violet'], s=45, zorder=5, edgecolor=C['bg'], linewidth=1.5)
ax.scatter([], [], color=C['ink2'], s=40, label='Actual 2026 (from the URA and HDB indices)')
TOPUP_NOW = tgt_act[-1] - flat_act[-1]
ax.set_xticks(hq); ax.set_xticklabels([l.replace('-', '\n') for l in lab_q], fontsize=8.5); ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f'S${v:.1f}M'))
ax.set_ylim(0.5, tph.max() / 1e6 * 1.12); ax.legend(loc='upper left', fontsize=8.6)
GAP_GROWTH = (tp[8] - fp[8]) - (tp[0] - fp[0])
finish(fig, f'Illustration: on the URA and HDB indices, the top-up rose to about {sgm(TOPUP_NOW)} by mid-2026',
       f'Trend-rule forecasts from end-2025 (private +{tr_p * 4:.1f}%, HDB {tr_h * 4:+.1f}% a year) with 80% ranges from each index\'s past errors; dots = the indices\' actual 2026 moves applied to this flat and home. An illustration, not an observed price for this family.',
       source='URA private residential price index and HDB resale price index via data.gov.sg; flat value from Huttons 4Q 2025; route B from 2025 sales')
```

**External sources for this section:** [URA: non-landed price index](https://data.gov.sg/datasets/d_97f8a2e995022d311c6c68cfda6d034c/view) · [HDB: resale price index](https://data.gov.sg/datasets/d_14f63e595975691e7c24a27ae4c07c79/view) · [URA: provisional 2026 Q3 release](https://www.ura.gov.sg/news/media/pr26-69/)

**The relative index gap narrowed in 2025.** Since 2009 (both indices = 100), private prices had risen only 1.9% more than HDB prices by 2025 Q2, the smallest lead since 2017. That lead has started widening again.

```python
# The upgrader's window: private prices relative to HDB resale prices (both indices = 100 in 2009 Q1).
q['upgrade_gap'] = (q.ppi / q.hdb - 1) * 100
gg = q.loc['2009-Q1':]; xg = np.arange(len(gg))
fig, ax = plt.subplots(figsize=(10, 4.0))
ax.fill_between(xg, gg.upgrade_gap, 0, where=gg.upgrade_gap >= 0, color=C['violet'], alpha=0.10, lw=0)
ax.plot(xg, gg.upgrade_gap, color=C['violet'], lw=2.2); ax.axhline(0, color=C['ink2'], lw=0.8)
for t, off in [('2019-Q3', (0, 10)), ('2025-Q2', (-55, 10)), ('2026-Q2', (-4, 12))]:
    i = list(gg.index).index(t); ax.scatter(i, gg.upgrade_gap[t], color=C['violet'], s=50, zorder=3, edgecolor=C['bg'], linewidth=1.5)
    ax.annotate(f'{t}: {gg.upgrade_gap[t]:.1f}%', (i, gg.upgrade_gap[t]), xytext=off, textcoords='offset points', ha='center', fontsize=9, color=C['ink'], fontweight='bold')
tk = [list(gg.index).index(f'{y}-Q1') for y in range(2009, 2027, 2)]; ax.set_xticks(tk); ax.set_xticklabels([gg.index[i][:4] for i in tk])
ax.set_ylabel('Private vs HDB index, % difference\n(both 2009 Q1 = 100)')
lo_t = gg.loc['2018-Q1':, 'upgrade_gap'].idxmin()
finish(fig, f'The private/HDB price-index gap reached an eight-year low in 2025',
       f'Private vs HDB resale price index, both rebased to 2009 Q1. The gap hit {gg.upgrade_gap[lo_t]:.1f}% in {lo_t} and is widening again ({gg.upgrade_gap.iloc[-1]:.1f}% in {gg.index[-1]}). This is not the price gap for a matched flat and condo.',
       source='URA private residential price index (non-landed) and HDB resale price index, via data.gov.sg')
```

## Step 7 · Five years of evidence (2020–2025)
Steps 1–6 use 2025 sales. We also have the same kind of file for **August 2020 to December 2024**, so we can check whether the 2025 patterns hold over time and follow the same projects as they age. The same five-year table also drives Step 8.
- **Cleaning:** in the 2020–23 file, March–July 2021 rows appeared an even number of times (2×, 4× or more), so the cleaner retained half of each exact-row group. August–October 2021 are missing and stay as gaps. Sub sales are only recorded from 2024, so they are left out of every comparison across years. 2024 prices were reconstructed from whole-dollar price per sqft times area; their last price digits are unreliable.
- **What's left:** 106,449 sales with a floor band and, for 99-year homes, a lease start. All 544 projects found only in 2020–24 have supplied coordinates, so 99.3% of these sales have a walking time. A new station counts from the month after it opened. Because August–October 2021 are missing, 2021 Q3 (July only, 145 sales) and 2021 Q4 (November–December only) are partial quarters.
**How to reproduce preprocessing.** The GitHub repository now includes `clean_2020_2025.py`, its input CSVs and `cleaning_log.txt`. The 2024 input already contains reconstructed prices; the cleaner flags those rows rather than recovering original negotiated prices. Joining the eleven historical-file parts reproduces the supplied cleaned table. Running the cleaner checks the earlier transformations from the supplied inputs. File hashes and the recorded software versions are listed in `CLEANING_PROVENANCE.md`.

**Repository files:** [cleaner](https://github.com/BIGBACKDIU/DAO2702_project/blob/main/clean_2020_2025.py), [cleaning log](https://github.com/BIGBACKDIU/DAO2702_project/blob/main/cleaning_log.txt) and [input hashes and provenance](https://github.com/BIGBACKDIU/DAO2702_project/blob/main/CLEANING_PROVENANCE.md). The supplied cleaned file is retained for consistent analysis; the verified rebuild has the same rows and flags with negligible floating-point round-trip differences, but its bytes need not have the same hash.


```python
# Five years of sales (Aug 2020 - Dec 2025) for Steps 7 and 8. mk_dummies drops the named base level, or the first level if that
# one is absent from a subset (e.g. a segment without District 9).
def mk_dummies(s, prefix, base):
    D = pd.get_dummies(s, prefix=prefix).astype(float); base_col = f'{prefix}_{base}'; return D.drop(columns=[base_col if base_col in D else D.columns[0]])
MK_CSV_SHA256 = "edede72a9ab8c628d30ce9605dc6a6c7ddb8864f22a734c5a92c78117aacf8b6"
assert hashlib.sha256(open('properties_2020_2025_clean.csv', 'rb').read()).hexdigest() == MK_CSV_SHA256, 'Clean CSV changed; rerun and review outputs.'
mk_raw = pd.read_csv('properties_2020_2025_clean.csv')
mk_raw['date_m'] = pd.to_datetime(mk_raw.date, format='%b-%y')
mk_raw['lpsf'] = np.log(mk_raw.psf); mk_raw['larea'] = np.log(mk_raw.area)
# Subsales are absent from the 2020–23 source, so remove them from all longitudinal comparisons.
mk_df = mk_raw.loc[~mk_raw.sub_sale].copy()
mk_df['lease_used0'] = mk_df.lease_used.fillna(0.0)
mk_model = mk_df.loc[(mk_df.psf > 0) & (mk_df.area > 0) & mk_df.floor.notna() &
               ((mk_df.tenure != '99-yr') | mk_df.lease_used.notna())].copy()
mk_model['lpsf'] = np.log(mk_model.psf); mk_model['larea'] = np.log(mk_model.area)
MK_SOURCE = 'properties_2020_2025_clean.csv; 2024 price reconstructed from rounded psf × area'
print(f"Loaded {len(mk_raw):,} sales; {len(mk_df):,} after removing {mk_raw.sub_sale.sum():,} sub sales; {len(mk_model):,} model-ready rows.")
mk_q = q.copy(); mk_q.index = mk_q.index.str.replace('-', '', regex=False)   # quarter labels as in the clean CSV
```

**The top-up from real sales.** Step 6 scaled the top-up with broad indices. Here the condo side comes from our own sales: route B and F prices are moved back quarter by quarter using the same eastern family-size projects (8,257 sales in 513 projects), reducing distortions from changes in which projects sold. The flat side still uses HDB's resale index.
- **Their flat rose faster in % terms:** about 8.3% a year, against 7.2% for eastern condos.
- **But the top-up still widened:** route B minus their flat is estimated at about **S\$0.62M** in August–September 2020 and **S\$0.85M** in late 2025, and route F from S\$0.82M to S\$1.14M. A faster rise on a smaller base still leaves a bigger dollar gap.
- **Two versions of the same number:** Step 6's index-scaled illustration puts the end-2025 top-up at about S\$0.83M; built from our own sales it is S\$0.85M. The difference comes from the condo index used.
- **What it means:** over these five years, the illustrated upgrade gap increased in nominal dollars. This does not measure affordability relative to income: the history does not track this family’s earnings, savings, debts or outstanding flat loan. It is a historical price-gap illustration, not an estimate of the financial effect of waiting.

```python
# A same-project price index for this buyer's kind of home, built from our own sales: log psf on project fixed effects and quarter
# fixed effects (plus floor and size). Each quarter's effect is the price change of the same projects, reducing shifts in observed project mix; within-project unit selection
# can still change over time. (Ageing of those projects is part of the change, as it would be for a home they own.)
def project_index(mask):
    s_ = dm[mask & dm.lpsf.notna()]
    big_ = s_.project.value_counts(); s_ = s_[s_.project.isin(big_[big_ >= 2].index)]
    q0_ = s_.qtr.min()
    Xi = pd.concat([dummies(s_.qtr.astype(str), 'Q', str(q0_)), s_[['floor', 'larea']],
                    pd.get_dummies(s_.project, prefix='P', drop_first=True).astype(float)], axis=1)
    b_ = np.linalg.lstsq(np.column_stack([np.ones(len(Xi)), Xi.values]), s_.lpsf.values, rcond=None)[0]
    qs_ = [str(q0_)] + [c[2:] for c in Xi.columns if c.startswith('Q_')]
    eff_ = pd.Series([0.0] + list(b_[1:1 + len(qs_) - 1]), index=qs_).sort_index()
    idx_ = np.exp(eff_ - eff_[[k for k in eff_.index if k.startswith('2025')]].mean())
    return idx_, len(s_), s_.project.nunique()
```

```python
# Their target over five years, from real sales: route B and F prices moved back by a same-project index for eastern family-size
# resales, and their flat moved back by HDB's resale price index. The top-up is what the upgrade costs on top of the flat.
idx_e, n_e, p_e = project_index(MY_SEG(dm) & dm.district.between(14, 19))
qs_ = [k for k in idx_e.index if k >= '2020Q3']
flat_ = pd.Series({k: PERSON['hdb_value'] * q.hdb[f'{k[:4]}-{k[4:]}'] / q.hdb['2025-Q4'] for k in qs_})
TGT = pd.DataFrame({'route B': opt.loc['B', 'real median'] * idx_e[qs_], 'route F': opt.loc['F', 'real median'] * idx_e[qs_], 'flat': flat_})
TGT['top-up B'] = TGT['route B'] - TGT.flat; TGT['top-up F'] = TGT['route F'] - TGT.flat
x_ = np.arange(len(TGT)); gap_i_ = qs_.index('2021Q3')  # July only; Aug–Oct missing
lab_ = [k[:4] if k.endswith('Q1') or k == qs_[0] else '' for k in TGT.index]
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(11.5, 4.9), gridspec_kw=dict(width_ratios=[1.25, 1]))
segs_ = [slice(0, gap_i_ + 1), slice(gap_i_ + 1, len(x_))]
for sl_ in segs_:                                                          # the top-up to route B, shaded between their flat and the condo
    ax.fill_between(x_[sl_], TGT.flat.iloc[sl_] / 1e6, TGT['route B'].iloc[sl_] / 1e6, color=C['blue_l'], alpha=0.55, lw=0)
for col_, nm_, lab2 in [(C['orange'], 'route F', 'East resale: 89–93 years left'), (C['blue'], 'route B', 'East resale: 85–88 years left'),
                        (C['aqua'], 'flat', 'Their 5-room HDB flat')]:
    s_ = TGT[nm_] / 1e6
    for k_, sl_ in enumerate(segs_ if nm_ != 'flat' else [slice(0, len(x_))]):
        ax.plot(x_[sl_], s_.iloc[sl_], color=col_, lw=2.6, label=lab2 if k_ == 0 else None, solid_capstyle='round')
    ax.annotate(sgm(s_.iloc[0] * 1e6), (0, s_.iloc[0]), xytext=(-4, 0), textcoords='offset points', ha='right', va='center', fontsize=8.3, color=col_)
    ax.annotate(sgm(s_.iloc[-1] * 1e6), (x_[-1], s_.iloc[-1]), xytext=(4, 0), textcoords='offset points', va='center', fontsize=8.3, color=col_, fontweight='bold')
mid_i = len(x_) - 6
ax.text(mid_i, (TGT.flat.iloc[mid_i] + TGT['route B'].iloc[mid_i]) / 2e6, 'top-up', ha='center', va='center', fontsize=9.5, color=C['blue_d'], fontweight='bold')
ax.set_xticks(x_); ax.set_xticklabels(lab_); ax.set_xlim(-3, len(x_) + 2); ax.set_ylim(0, TGT['route F'].max() / 1e6 * 1.15)
ax.set_ylabel('S$ million'); ax.legend(loc='lower right', fontsize=8); ax.set_title('Price of each, 2020 Q3 - 2025 Q4', loc='left', fontsize=10, color=C['ink'])
ax.axvspan(gap_i_ + .12, gap_i_ + .88, color=C['grey_l'], alpha=.28, zorder=0)
# right: the same gap as two pairs of towers, start vs end
for gx_, i_, yl_ in [(0.0, 0, '2020 Q3'), (2.75, -1, '2025 Q4')]:
    fh_, bh_ = TGT.flat.iloc[i_] / 1e6, TGT['route B'].iloc[i_] / 1e6
    draw_tower(ax2, gx_, fh_, C['aqua'], sgm(fh_ * 1e6)); draw_tower(ax2, gx_ + 0.8, bh_, C['blue'], sgm(bh_ * 1e6))
    bx_ = gx_ + 1.28
    ax2.annotate('', (bx_, bh_), (bx_, fh_), arrowprops=dict(arrowstyle='<->', color=C['ink'], lw=1.4))
    ax2.plot([gx_ + 0.31, bx_ + 0.05], [fh_, fh_], color=C['ink2'], lw=0.8, ls=':')
    ax2.text(bx_ + 0.08, (fh_ + bh_) / 2, f'top-up\n{sgm((bh_ - fh_) * 1e6)}', ha='left', va='center', fontsize=9.5, fontweight='bold', color=C['blue_d'])
    ax2.text(gx_ + 0.4, -0.12, yl_, ha='center', va='top', fontsize=10, fontweight='bold', color=C['ink'])
    ax2.text(gx_, -0.33, 'their flat', ha='center', va='top', fontsize=8, color=C['ink2']); ax2.text(gx_ + 0.8, -0.33, 'route B', ha='center', va='top', fontsize=8, color=C['ink2'])
ax2.text(-0.5, TGT['route B'].max() / 1e6 * 1.12, f"Top-up to route F: {sgm(TGT['top-up F'].iloc[0])} → {sgm(TGT['top-up F'].iloc[-1])}", ha='left', va='top', fontsize=8.5, color=C['orange'], fontweight='bold')
ax2.set_xlim(-0.6, 4.8); ax2.set_ylim(-0.55, TGT['route B'].max() / 1e6 * 1.18); ax2.axis('off')
ax2.set_title('Top-up = condo minus their flat', loc='left', fontsize=10, color=C['ink'])
yrs_ = (len(qs_) - 1) / 4
g_b = (idx_e[qs_[-1]] / idx_e[qs_[0]]) ** (1 / yrs_) - 1; g_h = (flat_.iloc[-1] / flat_.iloc[0]) ** (1 / yrs_) - 1
TGT_SUM = dict(g_b=g_b, g_h=g_h, top0=TGT['top-up B'].iloc[0], top1=TGT['top-up B'].iloc[-1], n=n_e, p=p_e,
               topF0=TGT['top-up F'].iloc[0], topF1=TGT['top-up F'].iloc[-1])
finish(fig, f'Their flat rose faster in % ({g_h:.1%} a year vs {g_b:.1%}), but the top-up still widened from {sgm(TGT_SUM["top0"])} to {sgm(TGT_SUM["top1"])}',
       f'Route prices = their 2025 medians moved back by a same-project index of east and north-east family-size resales ({n_e:,} sales, {p_e} projects). '
       f'Flat = S$795k in 2025 Q4 moved back by HDB\'s resale index. A faster % rise on a smaller base still leaves a bigger S$ gap. Grey break = missing Aug-Oct 2021; adjacent quarters are partial.',
       source='properties23.csv, properties2024.csv, properties2025.csv; HDB resale price index (data.gov.sg)')
print({k: (round(v, 4) if isinstance(v, float) else v) for k, v in TGT_SUM.items()})
```

**Does a shorter lease lose ground over time?** Step 4 compared different projects in one year. Here we follow the **same projects** from one year to the next (1,499 consecutive project-year pairs) and compare their yearly price change across lease bands, within the same years.
- **Shorter-lease projects did not fall behind.** Projects with 85–88 years left gained about **1.4 points a year more** than longer-lease ones with 89+ years left (95% CI +0.6 to +2.3); every band under 89 years did better than 89+.
- **Why the bands differ is uncertain:** the groups can differ in location, project quality and which units resell. We do not observe completion dates, so we cannot establish that a fading new-building premium explains the result.
- **Market-wide check:** across all 99-year resales, the oldest-lease projects (under 75 years left) also gained a little faster than those with 85–94 years left.
- **What can't be measured:** inside one project, lease used rises exactly one year per year, so a single yearly depreciation rate can't be separated from the market. What is measured is the difference between bands in the same years.
- **For the family:** route B’s 85–88-year lease band showed no weaker appreciation than the longer-lease band in 2020–25; this does not forecast its future price.
**Connect this to their eventual sale.** Step 4’s lease bands describe price differences between projects in 2025. Step 7’s panel compares appreciation between project groups in the observed years. Neither identifies the resale price of their chosen home after a ten-year stay. In particular, the roughly 13% lower price of today’s 75–79-year band cannot be applied as a future loss to an 85–88-year home. The planning check below tracks remaining lease and loan balance under explicit price scenarios instead.


```python
import re
# Do homes with less lease left gain less over time? Follow the SAME projects from year to year: the change in each project's
# median resale psf, compared across lease-left bands, with year fixed effects (so market swings cancel). Freehold is shown as
# a reference. Lease used rises one year per year inside a project, so a single linear "depreciation rate" can't be separated from
# the market; what IS identified is whether shorter-lease projects appreciate more slowly than longer-lease ones in the same years.
seg_ = MY_SEG(dm) & dm.tenure.isin(['99-yr', 'Freehold'])
py_ = dm[seg_].groupby(['project', 'year']).agg(m=('lpsf', 'median'), n=('lpsf', 'size'), used=('lease_used', 'median'),
                                                ten=('tenure', 'first'), dist=('district', 'first')).reset_index()
py_ = py_[py_.n >= 2].sort_values(['project', 'year'])
py_['prev_year'] = py_.groupby('project').year.shift(); py_['prev_m'] = py_.groupby('project').m.shift(); py_['prev_used'] = py_.groupby('project').used.shift()
ch_ = py_[py_.prev_year.notna()].copy(); ch_['gap'] = ch_.year - ch_.prev_year
long_gaps = int((ch_.gap > 1).sum())
ch_ = ch_[ch_.gap == 1].copy()  # year-to-year means consecutive calendar years
ch_['g'] = (ch_.m - ch_.prev_m) / ch_.gap * 100            # % a year (log points), averaged over the gap
ch_['left'] = 99 - ch_.prev_used
ch_['band'] = np.where(ch_.ten == 'Freehold', 'Freehold', pd.cut(ch_.left, AGE_BANDS, labels=AGE_LABELS, right=False).astype(str))
ch_ = ch_[ch_.band != 'nan']
Xa_ = pd.concat([dummies(ch_.band, 'B', AGE_BASE), dummies(ch_.year, 'Y', ch_.year.min())], axis=1)
age_tab, _ = ols(ch_.g, Xa_, list(Xa_.columns), cluster=ch_.project)
AGE = pd.DataFrame({'band': [AGE_BASE] + [c[2:] for c in Xa_.columns if c.startswith('B_')]})
AGE['diff'] = [0.0] + [age_tab.loc[f'B_{b}', 'coef'] for b in AGE.band[1:]]
AGE['lo'] = [np.nan] + [age_tab.loc[f'B_{b}', 'coef'] - 1.96 * age_tab.loc[f'B_{b}', 'se'] for b in AGE.band[1:]]
AGE['hi'] = [np.nan] + [age_tab.loc[f'B_{b}', 'coef'] + 1.96 * age_tab.loc[f'B_{b}', 'se'] for b in AGE.band[1:]]
AGE['p'] = [np.nan] + [age_tab.loc[f'B_{b}', 'p'] for b in AGE.band[1:]]
AGE['pairs'] = AGE.band.map(ch_.band.value_counts()); AGE['projects'] = AGE.band.map(ch_.groupby('band').project.nunique())
AGE['raw'] = AGE.band.map(ch_.groupby('band').g.mean())
order_ = AGE_LABELS + ['Freehold']; AGE = AGE.set_index('band').reindex([b for b in order_ if b in set(AGE.band)])
# Show annual appreciation contrasts as estimates and uncertainty, with route bands highlighted.
fig, ax = plt.subplots(figsize=(11.5, 5.1))
ys = np.arange(len(AGE))[::-1]
finite = AGE[['lo', 'hi', 'diff']].to_numpy(dtype=float)
xmin = min(-3.0, float(np.nanmin(finite)) - 1.2); xmax = max(3.2, float(np.nanmax(finite)) + 1.8)
ax.axvline(0, color=C['ink2'], lw=1.2, zorder=1)
for y, (band, row) in zip(ys, AGE.iterrows()):
    if band in AGE_HIGHLIGHT: ax.axhspan(y - 0.43, y + 0.43, color='#EFF4FA', zorder=0)
    if band == AGE_BASE:
        ax.scatter(0, y, color=C['grey'], s=85, zorder=4)
        ax.text(0.12, y, 'reference', va='center', fontsize=10, color=C['ink2'])
    else:
        col = C['grey'] if band == 'Freehold' else C['blue']
        ax.plot([row.lo, row.hi], [y, y], color=col, lw=3, solid_capstyle='round', zorder=3)
        ax.plot([row.lo, row.lo], [y - .10, y + .10], color=col, lw=2, zorder=3)
        ax.plot([row.hi, row.hi], [y - .10, y + .10], color=col, lw=2, zorder=3)
        ax.scatter(row['diff'], y, color=col, edgecolor='white', lw=1.4, s=105, zorder=4)
        ax.text(row.hi + .12, y, f'{row["diff"]:+.1f} pts/yr', va='center', fontsize=10, fontweight='bold', color=C['ink'])
ax.set_yticks(ys)
ax.set_yticklabels([f'{band}' + ('' if band == 'Freehold' else ' years left') +
                    f'  ·  {int(row.projects)} projects' + ('  ·  their routes' if band in AGE_HIGHLIGHT else '')
                    for band, row in AGE.iterrows()], fontsize=9.7)
ax.set_xlim(xmin, xmax); ax.set_ylim(-0.65, len(AGE) - 0.35)
ax.set_xlabel(f'Annual project price growth vs {AGE_BASE} years left (percentage points)')
ax.grid(axis='x', alpha=0.75); ax.grid(False, axis='y')
for spine in ['top', 'right', 'left']: ax.spines[spine].set_visible(False)
AGE_MEAN = ch_.groupby('band').g.mean()
finish(fig, AGE_TITLE.format(**{'b' + re.sub(r'\W', '_', k): v for k, v in AGE['diff'].items()}),
       f'{MY_LABEL}, 2020–2025: each project’s median resale psf change to the next year ({len(ch_):,} consecutive project-year pairs, '
       f'{ch_.project.nunique()} projects), compared within the same years. Dots = annual-growth differences; lines = 95% CIs clustered by project. '
       'These contrasts do not identify one home’s annual lease depreciation.',
       source='properties23.csv, properties2024.csv, properties2025.csv (cleaned)')
print(f'Excluded {long_gaps:,} nonconsecutive project observations; each included pair is one calendar year apart.')
print(AGE[['diff', 'lo', 'hi', 'p', 'pairs', 'projects', 'raw']].round(3).to_string())
```

**Plan ten years of ownership, not just the purchase price.** The lease bands elsewhere describe 2025 sales. For an illustrative 2026 purchase, subtract one further year before the ten-year stay: B and C would have about **74–77 years left** at exit, versus **78–82** for F. The printed table checks this calendar arithmetic; exact lease dates and the actual purchase year matter.

The table below starts with each route’s **2026-Q3 market-index price proxy**. It assumes a 75% loan over the persona’s loan term, a constant 1.6% rate, regular repayments and no refinancing. It asks what equity remains if the sale price after ten years is **20% lower, unchanged or 20% higher** than that starting price. These are chosen stress scenarios, not estimated future price ranges. Equity means sale price minus outstanding loan before sale costs and CPF refunds; it is not cash available to spend. The historical lease discount is not subtracted again.


```python
# Ownership planning arithmetic, separate from the lease regressions: no estimated depreciation is extrapolated.
LEASE_BASE_YEAR, PLANNING_PURCHASE_YEAR, OWNERSHIP_YEARS = 2025, 2026, 10
LEASE_ROUTE_BANDS={'B':(85,88),'C':(85,88),'F':(89,93)}
EXIT_PRICE_CHANGES=(-.20,0.,.20)  # Chosen stress scenarios, not confidence bounds.
def outstanding_balance(principal,annual_rate,term_years,elapsed_years):
    paid=int(elapsed_years*12); total=int(term_years*12)
    if paid>=total:return 0.
    r=annual_rate/12
    if r==0:return principal*(total-paid)/total
    m=payment(principal,annual_rate,term_years)
    return max(0.,principal*(1+r)**paid-m*((1+r)**paid-1)/r)
EXIT_ROWS=[];EXIT_LEASE_ROWS=[]
for route,(a,b) in LEASE_ROUTE_BANDS.items():
    calendar_shift=PLANNING_PURCHASE_YEAR-LEASE_BASE_YEAR
    entry=opt.loc[route,'2026-Q3 index proxy'];loan=LTV['first loan']*entry
    balance=outstanding_balance(loan,PAYMENT_RATE,TENURE,OWNERSHIP_YEARS)
    EXIT_LEASE_ROWS.append({'option':route,'2025 lease band':f'{a}–{b}',
                           'lease at 2026 purchase':f'{a-calendar_shift}–{b-calendar_shift}',
                           'lease after 10-year stay':f'{a-calendar_shift-OWNERSHIP_YEARS}–{b-calendar_shift-OWNERSHIP_YEARS}'})
    for change in EXIT_PRICE_CHANGES:
        exit_price=entry*(1+change)
        EXIT_ROWS.append({'option':route,'starting price proxy':entry,'price change assumption %':100*change,
                         'sale-price scenario':exit_price,'outstanding loan':balance,'equity before costs and CPF':exit_price-balance})
EXIT_LEASE=pd.DataFrame(EXIT_LEASE_ROWS).set_index('option')
EXIT_DETAIL=pd.DataFrame(EXIT_ROWS).set_index(['option','price change assumption %'])
EXIT_VIEW=EXIT_LEASE[['lease after 10-year stay']].copy()
EXIT_VIEW['outstanding loan']=EXIT_DETAIL.xs(0.,level=1)['outstanding loan'].map(sgm)
for change,label in [(-20.,'equity: price -20%'),(0.,'equity: price unchanged'),(20.,'equity: price +20%')]:
    EXIT_VIEW[label]=EXIT_DETAIL.xs(change,level=1)['equity before costs and CPF'].map(sgm)
EXIT_VIEW.index=['B · East, lower cost','C · North-east','F · East, more lease']
print({'lease baseline year':LEASE_BASE_YEAR,'illustrative purchase year':PLANNING_PURCHASE_YEAR,
       'years owned':OWNERSHIP_YEARS,'loan %':100*LTV['first loan'],'term years':TENURE,'constant rate %':100*PAYMENT_RATE,
       'price-change scenarios %':[100*x for x in EXIT_PRICE_CHANGES]})
display(EXIT_LEASE);display(EXIT_VIEW)
print('Financial assumptions, not estimated effects. Confidence intervals and p-values do not apply. Detailed currency arithmetic is in Appendix C.')
```

**Is the MRT gap a 2025 one-off?** We re-estimated Step 3b's comparison separately for each year, using the MRT network as it stood at each sale.
- **It appears in every year:** a ≤5-minute walk went with about +7% per sqft in 2020–23, +11% in 2024 and +17% in 2025 (every 95% CI above zero). 5–10 minutes was about +3% to +10%.
- **So 2025's figure is the high end.** Earlier-year estimates were about +7–11%; the higher 2025 estimate warns against treating one year as a fixed planning premium. These remain associations across projects, not station effects.

```python
# Is the MRT price gap a 2025 quirk? Re-estimate it for each year 2020-2025 in this buyer's segment, using the MRT network
# as it stood at each sale (a new station counts only after it opened).
rows_ = []
for y in sorted(dm.year.unique()):
    m_ = MY_SEG(dm) & (dm.year == y) & dm.walk_t.notna()
    s_ = dm[m_]
    if (s_.walk_band_t == '5 min or less').sum() < 30: continue
    Xy = pd.concat([dummies(s_.district, 'D', s_.district.mode()[0]), s_[['larea', 'floor', 'lease_used']], dummies(s_.tenure, 'ten', s_.tenure.mode()[0]),
                    dummies(s_.type, 'type', 'Condominium'), dummies(s_.month, 'm', s_.month.min()),
                    dummies(s_.walk_band_t.astype(str), 'W', 'over 15 min')], axis=1)
    Xy = Xy.loc[:, Xy.std() > 0]
    t_, _ = ols(s_.lpsf, Xy, list(Xy.columns), cluster=s_.project)
    for b in ['5 min or less', '5-10 min']:
        if f'W_{b}' in t_.index:
            rows_.append(dict(year=y, band=b, pct=t_.loc[f'W_{b}', 'pct'], lo=t_.loc[f'W_{b}', 'pct_lo'], hi=t_.loc[f'W_{b}', 'pct_hi'],
                              n_near=int((s_.walk_band_t == b).sum()), n_far=int((s_.walk_band_t == 'over 15 min').sum())))
MRT_YEARS = pd.DataFrame(rows_)
# Picture it: a heat strip, one tile per year and walking band; darker = bigger price gap vs a home over 15 minutes away.
yrs_m = sorted(MRT_YEARS.year.unique())
def cell_m(rw, y):
    rr = MRT_YEARS[(MRT_YEARS.year == y) & (MRT_YEARS.band == rw[0])]
    return None if rr.empty else (rr.pct.iloc[0], rr.lo.iloc[0], rr.hi.iloc[0])
fig, ax = plt.subplots(figsize=(10.5, 3.9))
draw_heat(ax, [('5 min or less', 'Walk 5 min\nor less'), ('5-10 min', 'Walk\n5-10 min')], yrs_m, cell_m, {2020: '(Aug-Dec)', 2021: '(no Aug-Oct)'})
near_ = MRT_YEARS[MRT_YEARS.band == '5 min or less']
finish(fig, f'A 5-minute walk went with higher psf in every year from 2020 to 2025: {near_.pct.min():+.0f}% to {near_.pct.max():+.0f}%, highest in {int(near_.loc[near_.pct.idxmax(), "year"])}',
       f'{MY_LABEL}, estimated separately for each year with district, size, floor, lease, tenure, type and month adjusted for; price per sqft vs homes over 15 minutes away; 95% CIs clustered by project. '
       'Network as it stood at each sale. Conditional associations across projects, not station effects.',
       source='properties23.csv, properties2024.csv, properties2025.csv; OneMap; LTA MRT exits via sgraildata; station opening dates (LTA)')
print(MRT_YEARS.round(2).to_string())
```

**A before/after comparison on their side of the island.** Seven Thomson-East Coast Line stations opened along the east coast on 23 June 2024 (Tanjong Rhu to Bayshore). We compare 237 projects in the Geylang, Katong and Bedok areas whose walk fell from over 10 to 10 minutes or less with 386 projects in the same districts whose walk did not change, before and after the opening, following the same projects.
- **No clear extra rise detected at opening:** +0.1% per sqft afterwards (95% CI −1.7% to +1.8%, p = 0.93), while pre-opening estimates have wide intervals that include zero; parallel trends are not established.
- **Reading:** the line was announced in 2014, so the value of a future station may have been reflected in prices well before it opened. That is compatible with a lasting MRT gap across projects without a detectable jump at opening.
- **For the family:** two more TEL stations (Bedok South and Sungei Bedok) are due in the second half of 2026. Treat an opening-day price jump as uncertain; assess actual access and the price already asked for the home.

```python
# A before/after comparison in their area: seven Thomson-East Coast Line stations opened on 23 June 2024 along the east coast
# (Tanjong Rhu to Bayshore). Projects in D14-D16 that went from over 10 minutes' walk to 10 minutes or less because of them are
# compared with projects in the same districts whose walk didn't change, before and after opening (same projects followed,
# project and half-year fixed effects). The line was announced in 2014, so some of the gain may have been priced in earlier.
TEL4 = pd.Timestamp('2024-06-23')
r_pre = np.searchsorted(REGIMES, np.datetime64(TEL4 - pd.Timedelta(days=1)), side='right') - 1
r_post = np.searchsorted(REGIMES, np.datetime64(TEL4), side='right') - 1
w_pre, w_post = walk_reg.iloc[:, r_pre], walk_reg.iloc[:, r_post]
area_ = dm[dm.district.isin([14, 15, 16])].project.unique()
TREAT = [p for p in area_ if p in w_pre.index and w_pre[p] > 10 and w_post[p] <= 10]
CTRL = [p for p in area_ if p in w_pre.index and abs(w_pre[p] - w_post[p]) < 0.5]
ev = dm[dm.project.isin(TREAT + CTRL) & (dm.sale == 'Resale')].dropna(subset=['lpsf', 'floor', 'larea']).copy()
ev['treat'] = ev.project.isin(TREAT).astype(float)
ev['h'] = ev.year * 2 + (ev.month > 6) - (2024 * 2 + 1)           # 0 = 2024 H2, the first half-year after opening
ev['k'] = ev.h.clip(-6, 2)
EV_K = [k for k in sorted(ev.k.unique()) if k != -1]
Xe = pd.concat([pd.DataFrame({f'T_{k}': ((ev.k == k) & (ev.treat == 1)).astype(float) for k in EV_K}, index=ev.index),
                ev[['floor', 'larea']], dummies(ev.h.astype(str), 'H', str(ev.h.min())),
                pd.get_dummies(ev.project, prefix='P', drop_first=True).astype(float)], axis=1)
ev_tab, _ = ols(ev.lpsf, Xe, list(Xe.columns), cluster=ev.project)
EV = pd.DataFrame({'k': EV_K, 'pct': [ev_tab.loc[f'T_{k}', 'pct'] for k in EV_K], 'lo': [ev_tab.loc[f'T_{k}', 'pct_lo'] for k in EV_K],
                   'hi': [ev_tab.loc[f'T_{k}', 'pct_hi'] for k in EV_K]})
EV = pd.concat([EV, pd.DataFrame({'k': [-1], 'pct': [0.0], 'lo': [0.0], 'hi': [0.0]})]).sort_values('k')
Xd = pd.concat([pd.DataFrame({'TxPost': ev.treat * (ev.h >= 0)}, index=ev.index), ev[['floor', 'larea']], dummies(ev.h.astype(str), 'H', str(ev.h.min())),
                pd.get_dummies(ev.project, prefix='P', drop_first=True).astype(float)], axis=1)
did_tab, _ = ols(ev.lpsf, Xd, list(Xd.columns), cluster=ev.project); DID = did_tab.loc['TxPost']
TEL4_N = dict(treat=len(TREAT), ctrl=len(CTRL), n_t=int((ev.treat == 1).sum()), n_c=int((ev.treat == 0).sum()), n_t_post=int(((ev.treat == 1) & (ev.h >= 0)).sum()))
lab_h = {k: (f'{2024 + (k + 1) // 2}-H{1 if (k + 1) % 2 == 0 else 2}') for k in range(-6, 3)}
lab_h[-6] = 'up to\n2021-H2'; lab_h[2] = '2025-H2'
fig, ax = plt.subplots(figsize=(10.5, 4.6))
draw_event(ax, EV.k, EV.pct, EV.lo, EV.hi, [lab_h[k] for k in EV.k], -0.5, -1, 'TEL4 opens 23 Jun 2024')
ax.set_ylabel('Homes that gained a station\nminus unchanged ones (% psf)')
verdict_ = 'no clear extra rise' if DID.pct_lo < 0 < DID.pct_hi else ('a rise' if DID.pct > 0 else 'a fall')
finish(fig, f'When TEL4 opened in 2024, east-coast homes that gained a station saw {verdict_} vs neighbours ({DID.pct:+.1f}% psf, 95% CI {DID.pct_lo:+.1f} to {DID.pct_hi:+.1f})',
       f'{TEL4_N["treat"]} Geylang, Katong and Bedok area projects whose walk to MRT fell from over 10 to 10 minutes or less when TEL4 opened, vs {TEL4_N["ctrl"]} same-district projects whose walk did not change. '
       f'Resales only ({TEL4_N["n_t"]:,} treated sales, {TEL4_N["n_t_post"]:,} after opening); project and half-year fixed effects, floor and size adjusted for; 95% CIs clustered by project; '
       '2024-H1 is the reference. The line was announced in 2014, so earlier gains may already be priced in.',
       source='properties23.csv, properties2024.csv, properties2025.csv; OneMap; LTA station exits via sgraildata; TEL4 opening date (LTA)')
print(f"TEL4: {TEL4_N} | after opening {DID.pct:+.2f}% (95% CI {DID.pct_lo:+.2f} to {DID.pct_hi:+.2f}, p = {DID.p:.3f}) | pre-period estimates: " + ', '.join(f'{k}: {v:+.1f}%' for k, v in zip(EV.k, EV.pct) if k < -1))
```

**What five years add to the advice:**

## Step 8 · The market around them (2020–2025)
The family's decision sits inside a market that moved a lot in five years. These charts use the same five-year table as Step 7, for all private condos and apartments.

**Our own index tracks URA's.** We built two like-for-like price indices from our own sales: one holding home features fixed (hedonic) and one following the same projects.
- **Over the five years:** from 2020 Q4 to 2025 Q4, the hedonic index rose 34%, the same-project index 29% and URA's non-landed index 36%. The same-project index may reflect ageing as well as different samples and model controls.
- **Quarter to quarter:** the moves agree only moderately with URA's (r = 0.51, 95% CI 0.04–0.80, 17 quarters), so we use URA's complete series for timing and treat our own index as a cross-check on the broad trend. The hedonic index dipped through 2025 while URA's rose; a change in that year's sales mix can't be ruled out.

```python
mk_q_months = mk_df.groupby('quarter').date_m.nunique()
mk_complete_q = list(mk_q_months[mk_q_months==3].index)
mk_all_q = list(pd.period_range('2020Q4','2025Q4',freq='Q').astype(str))
print('Incomplete quarters excluded from own indices:', [mk_x for mk_x in mk_all_q if mk_x not in mk_complete_q])
mk_idx_data = mk_model[mk_model.quarter.isin(mk_complete_q)].copy()
MK_BASE_Q = '2020Q4'
def mk_index_design(s):
    return pd.concat([mk_dummies(s.district,'D',9),mk_dummies(s.tenure,'ten','99-yr'),
                      mk_dummies(s.type,'type','Condominium'),mk_dummies(s.sale,'sale','Resale'),
                      s[['larea','floor','lease_used0']],mk_dummies(s.quarter,'q',MK_BASE_Q)],axis=1)
def mk_hedonic_index(s):
    X=mk_index_design(s); t,h=ols(s.lpsf,X,list(X.columns),cluster=s.project)
    z=pd.Series({k: (100 if k==MK_BASE_Q else 100*np.exp(t.loc['q_'+k,'coef'])) for k in mk_complete_q},name='hedonic')
    return z,t,h
MK_HIDX,MK_HTAB,MK_HINFO=mk_hedonic_index(mk_idx_data)
MK_HIDX_ND,_,_=mk_hedonic_index(mk_idx_data[~mk_idx_data.exact_dup])
def mk_project_index(s):
    X=pd.concat([mk_dummies(s.quarter,'q',MK_BASE_Q),s[['larea','floor']],
                 mk_dummies(s.sale,'sale','Resale'),mk_dummies(s.type,'type','Condominium')],axis=1)
    Xd=X-X.groupby(s.project.values).transform('mean')
    yd=s.lpsf-s.groupby('project').lpsf.transform('mean')
    t,h=ols(yd,Xd,list(X.columns),cluster=s.project,const=False)
    return pd.Series({k: (100 if k==MK_BASE_Q else 100*np.exp(t.loc['q_'+k,'coef'])) for k in mk_complete_q},name='within project'),t
MK_PIDX,MK_PTAB=mk_project_index(mk_idx_data)
MK_PIDX_ND,_=mk_project_index(mk_idx_data[~mk_idx_data.exact_dup])
mk_index_table=pd.DataFrame({'Own hedonic':MK_HIDX,'Own hedonic without exact duplicates':MK_HIDX_ND,
                          'Within project':MK_PIDX,'Within project without exact duplicates':MK_PIDX_ND}).reindex(mk_all_q)
mk_index_table['URA non-landed']=mk_q.ppi.reindex(mk_index_table.index)/mk_q.ppi[MK_BASE_Q]*100
mk_growth=mk_index_table.pct_change(fill_method=None)*100
mk_adjacent=mk_index_table['Own hedonic'].notna() & mk_index_table['Own hedonic'].shift().notna()
def mk_corr_ci(a,b):
    r,p=st.pearsonr(a,b); n=len(a); z=np.arctanh(r); se=1/np.sqrt(n-3)
    return r,np.tanh(z-1.96*se),np.tanh(z+1.96*se),p,n
MK_INDEX_CORR = pd.DataFrame([dict(pair=k,r=mk_corr_ci(mk_growth.loc[mk_adjacent,k],mk_growth.loc[mk_adjacent,'URA non-landed'])[0],
                                    low=mk_corr_ci(mk_growth.loc[mk_adjacent,k],mk_growth.loc[mk_adjacent,'URA non-landed'])[1],
                                    high=mk_corr_ci(mk_growth.loc[mk_adjacent,k],mk_growth.loc[mk_adjacent,'URA non-landed'])[2],
                                    p=mk_corr_ci(mk_growth.loc[mk_adjacent,k],mk_growth.loc[mk_adjacent,'URA non-landed'])[3],
                                    n=mk_corr_ci(mk_growth.loc[mk_adjacent,k],mk_growth.loc[mk_adjacent,'URA non-landed'])[4])
                            for k in ['Own hedonic','Own hedonic without exact duplicates','Within project','Within project without exact duplicates']]).set_index('pair')
display(mk_index_table.round(2));display(MK_INDEX_CORR.round(3));
# Three lines, labelled at their ends (no legend to decode); the gap between 2021 Q2 and 2022 Q1 is the incomplete-quarter break.
fig, ax = plt.subplots(figsize=(11, 4.9)); xx = np.arange(len(mk_index_table))
ax.axvspan(mk_all_q.index('2021Q3') - .45, mk_all_q.index('2021Q4') + .45, facecolor='#f1f2f4', edgecolor=C['grey'], hatch='////', lw=0)
ax.text(mk_all_q.index('2021Q3') + 0.5, 101, 'incomplete\nquarters', ha='center', va='bottom', fontsize=7.8, color=C['ink2'])
ends_ = []
for k, col, sty, lab in [('Own hedonic', C['blue'], '-', 'Our hedonic index'), ('Within project', C['aqua'], '-', 'Same-project index'), ('URA non-landed', C['ink2'], '--', 'URA non-landed index')]:
    s_ = mk_index_table[k]; ax.plot(xx, s_, color=col, lw=2.6 if k != 'URA non-landed' else 2, ls=sty)
    last_ = s_.last_valid_index(); xi_ = mk_all_q.index(last_); ends_.append([s_[last_], col, f'{lab}  {s_[last_] - 100:+.0f}%', xi_])
ax.plot(xx, mk_index_table['Own hedonic without exact duplicates'], color=C['blue'], lw=1, alpha=0.5, ls=':')
ends_.sort(key=lambda e: e[0]); ys_ = [e[0] for e in ends_]
for j in range(1, len(ys_)): ys_[j] = max(ys_[j], ys_[j - 1] + 3.2)
for (v_, col, lab, xi_), y_ in zip(ends_, ys_): ax.text(len(xx) - 0.4, y_, lab, ha='left', va='center', fontsize=9.2, fontweight='bold', color=col)
ax.set_xticks([mk_all_q.index(f'{y}Q1') for y in range(2021, 2026)]); ax.set_xticklabels([str(y) for y in range(2021, 2026)])
ax.set_xlim(-0.5, len(xx) + 4.6); ax.set_ylabel('Index, 2020 Q4 = 100'); ax.axhline(100, color=C['grey'], lw=0.8)
finish(fig,'Like-for-like prices and the URA index rise together, with different short-run swings',
       'Quarterly log-psf OLS: district, sale, tenure, type, log area, floor, lease used and quarter; project-clustered SEs. End labels = change since 2020 Q4. '
       'Dotted = our index without exact duplicates; hatched = incomplete quarters.',
       MK_SOURCE+'; URA non-landed price index via data.gov.sg')
```

**The east was among the fastest-rising areas.** Holding home features fixed:
- **By segment:** the outer suburbs (OCR) rose about 46% from late 2020, against 34% in the city fringe (RCR) and 15% in the core central region (CCR).
- **By district:** from Jan–Jun 2021 to Jan–Jun 2025, Tampines/Pasir Ris rose about **47%** (95% CI 44–51%), level with Upper Thomson at the top, and north-east towns about 41%, against about 8% in Raffles Place (confidence interval includes zero). The median district rose 34%.
- **Different growth comparisons:** Step 7’s 7.2% annual figure follows the same eastern family-size resale projects from 2020 Q3 to 2025 Q4. This district figure compares the wider sale sample in Jan–Jun 2021 and Jan–Jun 2025 using recorded-feature controls. Different homes, windows and models mean these figures need not match.
- **For the family:** the east’s growth is market context. Step 7’s widening top-up follows the larger condo price base: a smaller percentage rise on a more expensive home can exceed a larger percentage rise on a cheaper flat. The district ranking does not explain that top-up by itself.

```python
MK_SEG_INDEX={};SEG_INDEX_ND={};SEG_TABS={}
for seg in ['CCR','RCR','OCR']:
    s=mk_idx_data[mk_idx_data.segment==seg]
    MK_SEG_INDEX[seg],SEG_TABS[seg],_=mk_hedonic_index(s)
    SEG_INDEX_ND[seg],_,_=mk_hedonic_index(s[~s.exact_dup])
MK_SEG_ENDPOINT=pd.DataFrame({'2025Q4 index':{k:v['2025Q4'] for k,v in MK_SEG_INDEX.items()},
                           'without exact duplicates':{k:v['2025Q4'] for k,v in SEG_INDEX_ND.items()}})
display(MK_SEG_ENDPOINT.round(2))
# Same calendar months: Jan–Jun 2021 versus Jan–Jun 2025; conditional district-specific contrasts.
mk_geo=mk_model[(mk_model.year.isin([2021,2025]))&(mk_model.month<=6)].copy()
def mk_district_growth(s):
    if s.year.nunique()<2 or s.project.nunique()<8:return np.nan,np.nan,np.nan,0,np.nan
    X=pd.concat([(s.year==2025).astype(float).rename('year25'),s[['larea','floor','lease_used0']],
                 pd.get_dummies(s.tenure,prefix='ten',drop_first=True,dtype=float),
                 pd.get_dummies(s.sale,prefix='sale',drop_first=True,dtype=float),
                 pd.get_dummies(s.type,prefix='type',drop_first=True,dtype=float)],axis=1)
    X=X.loc[:,X.std()>0];t,_=ols(s.lpsf,X,list(X.columns),cluster=s.project)
    r=t.loc['year25'];return r.pct,r.pct_lo,r.pct_hi,len(s),r.p
MK_DG=pd.DataFrame({d:mk_district_growth(s) for d,s in mk_geo.groupby('district')},index=['growth %','low','high','sales','p']).T.sort_index()
MK_DG_ND=pd.DataFrame({d:mk_district_growth(s[~s.exact_dup]) for d,s in mk_geo.groupby('district')},index=['growth %','low','high','sales','p']).T.sort_index()
display(MK_DG.round(1));display(MK_DG_ND[['growth %','low','high']].round(1).add_suffix(' without exact duplicates'))
print('Median district estimate, all / no exact duplicates:',round(MK_DG['growth %'].median(),1),round(MK_DG_ND['growth %'].median(),1))
fig,(ax,axm)=plt.subplots(1,2,figsize=(13.2,5.7),gridspec_kw={'width_ratios':[1,1.55]})
xq_ = np.arange(len(mk_all_q)); end_ = []
for seg,col,nm in [('CCR',C['violet'],'Core central'),('RCR',C['blue'],'City fringe'),('OCR',C['red'],'Outer suburbs')]:
    s_ = pd.Series(MK_SEG_INDEX[seg]).reindex(mk_all_q); ax.plot(xq_, s_, color=col, lw=2.6)
    end_.append([s_.dropna().iloc[-1], col, f'{nm} ({seg})  {s_.dropna().iloc[-1] - 100:+.0f}%'])
end_.sort(key=lambda e: e[0]); ys_ = [e[0] for e in end_]
for j in range(1, len(ys_)): ys_[j] = max(ys_[j], ys_[j - 1] + 4)
for (v_, col, lab), y_ in zip(end_, ys_): ax.text(len(xq_) - 0.5, y_, lab, ha='left', va='center', fontsize=8.8, fontweight='bold', color=col)
ax.axvspan(mk_all_q.index('2021Q3') - .45, mk_all_q.index('2021Q4') + .45, facecolor='#f1f2f4', edgecolor=C['grey'], hatch='////', lw=0)
ax.set_xticks([mk_all_q.index(f'{y}Q1') for y in range(2021, 2026)]); ax.set_xticklabels([str(y) for y in range(2021, 2026)])
ax.set_xlim(-0.5, len(xq_) + 7.5); ax.axhline(100, color=C['grey'], lw=0.8); ax.set_ylabel('Mix-adjusted index, 2020 Q4 = 100')
ax.set_title('By market segment', loc='left', fontsize=10, color=C['ink'])
MK_EXT=(103.60,104.05,1.195,1.475);axm.set_facecolor('#f7f8f9');axm.grid(False)
for ring in SGMAP['land']:axm.add_patch(Polygon(ring,closed=True,facecolor='#e3e6e8',edgecolor='none',zorder=0))
mk_good=MK_DG['growth %'].dropna();lon=np.array([CENT[int(k)][1] for k in mk_good.index if int(k) in CENT]);lat=np.array([CENT[int(k)][0] for k in mk_good.index if int(k) in CENT]);vals=np.array([mk_good[k] for k in mk_good.index if int(k) in CENT])
mk_gx,mk_gy=np.meshgrid(np.linspace(MK_EXT[0],MK_EXT[1],540),np.linspace(MK_EXT[2],MK_EXT[3],340))
mk_dist=((mk_gx[...,None]-lon)*111.32*np.cos(np.radians(1.35)))**2+((mk_gy[...,None]-lat)*110.57)**2
mk_weights=np.exp(-mk_dist/(2*3.2**2))+1e-15;z=(mk_weights*vals).sum(-1)/mk_weights.sum(-1)
# Diverging colours compare growth with the median district, not with zero growth.
mk_growth_mid=float(mk_good.median())
mk_cmap=LinearSegmentedColormap.from_list('growth',['#174779','#4385C2','#B0CBE5','#F6F3F2','#E4B0BA','#C45168','#8C263D'])
mk_growth_norm=TwoSlopeNorm(vmin=np.floor(vals.min()/5)*5,vcenter=mk_growth_mid,vmax=np.ceil(vals.max()/5)*5)
mk_verts=[];codes=[]
for ring in SGMAP['land']:
    a=np.asarray(ring);mk_verts+=a.tolist()+[a[0].tolist()];codes += [Path.MOVETO]+[Path.LINETO]*(len(a)-1)+[Path.CLOSEPOLY]
im=axm.imshow(z,extent=MK_EXT,origin='lower',cmap=mk_cmap,norm=mk_growth_norm,interpolation='bilinear',zorder=2)
im.set_clip_path(PathPatch(Path(mk_verts,codes),transform=axm.transData))
for k in mk_good.index:
    if int(k) in CENT:
        la,lo=CENT[int(k)];axm.scatter(lo,la,s=13,color=C['ink'],edgecolor='white',linewidth=.4,zorder=4)
        if int(k) in [18,26,19,5,1]:
            offsets={18:(2,18),26:(-14,17),19:(-58,-25),5:(-34,-22),1:(26,-16)}
            axm.annotate(f'{MAP_NAME[int(k)]}\n{mk_good[k]:+.0f}%',(lo,la),xytext=offsets[int(k)],textcoords='offset points',fontsize=8.2,fontweight='bold',color=C['ink'],bbox=dict(fc='white',ec='none',alpha=.94,pad=1.8),zorder=6,
                           arrowprops=dict(arrowstyle='-',color=C['ink2'],lw=.6))
axm.set_xlim(MK_EXT[:2]);axm.set_ylim(MK_EXT[2:]);axm.set_aspect('equal');axm.set_xticks([]);axm.set_yticks([])
for s_ in axm.spines.values(): s_.set_visible(False)
cb=fig.colorbar(im,ax=axm,orientation='horizontal',fraction=.055,pad=.035);cb.set_label(f'Price growth (%) · white = district median ({mk_growth_mid:.0f}%)\nBlue = slower growth; red = faster growth',fontsize=8.5);cb.outline.set_visible(False)
axm.set_title('By district: prices rose in both blue and red areas', loc='left', fontsize=10, color=C['ink'])
finish(fig,'Price growth varied by segment and district, even after adjusting for recorded unit mix',
       'Left: segment-specific quarterly hedonic indices, labelled with growth since 2020 Q4. Right: district regressions comparing Jan–Jun 2021 with Jan–Jun 2025, adjusted for sale, tenure, type, size, floor and lease used; smoothed between district centres.',
       MK_SOURCE+'; URA planning-area map')
```

**New-sale geography shifted around cooling measures.** In the 12 months after the April 2023 ABSD increase, the core central region's share of new sales fell from 31.8% to 16.1%. This is descriptive: rates, supply and sentiment changed at the same time. For this citizen family, selling first avoids second-property ABSD. Wider market responses to the policy can still affect the choices and prices they face.

```python
mk_policy_dates={'Dec 2021':'2021-12-01','Sep 2022':'2022-09-01','Apr 2023 ABSD':'2023-04-01'}
mk_new=mk_df[mk_df.sale=='New Sale']
mk_before=mk_new[mk_new.date_m.between('2022-04-01','2023-03-01')]
mk_after=mk_new[mk_new.date_m.between('2023-04-01','2024-03-01')]
def mk_share_ccr(s):return (int((s.segment=='CCR').sum()),len(s),100*(s.segment=='CCR').mean())
MK_POLICY=pd.DataFrame([mk_share_ccr(mk_before),mk_share_ccr(mk_after),mk_share_ccr(mk_before[~mk_before.exact_dup]),mk_share_ccr(mk_after[~mk_after.exact_dup])],
                    index=['12 months before','12 months after','before, no exact duplicates','after, no exact duplicates'],columns=['CCR new sales','all new sales','CCR share %'])
display(MK_POLICY.round(2));print('After minus before (percentage points):',round(MK_POLICY.loc['12 months after','CCR share %']-MK_POLICY.loc['12 months before','CCR share %'],2))
# Left: monthly sales as a soft area, with each cooling measure pinned as a labelled flag. Right: two donuts, before vs after.
from matplotlib.patches import Wedge
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(12, 4.6), gridspec_kw={'width_ratios': [2.3, 1]})
tot_p = mk_monthly[['Resale', 'New Sale']].sum(axis=1, min_count=1); xp_ = np.arange(len(mk_months))
gp_ = [mk_months.get_loc(m) for m in mk_missing_months]
for sl_ in [slice(0, gp_[0]), slice(gp_[-1] + 1, len(xp_))]:
    ax.fill_between(xp_[sl_], 0, tot_p.iloc[sl_], color=C['blue'], alpha=0.16, lw=0); ax.plot(xp_[sl_], tot_p.iloc[sl_], color=C['blue'], lw=2)
ax.axvspan(gp_[0] - 0.5, gp_[-1] + 0.5, facecolor='#f1f2f4', edgecolor=C['grey'], hatch='////', lw=0, zorder=0)
top_p = tot_p.max() * 1.35; ax.set_ylim(0, top_p)
for k_, (label, date) in enumerate(mk_policy_dates.items()):
    i = mk_months.get_loc(pd.Timestamp(date)); yf_ = top_p * (0.93 - 0.11 * k_)
    ax.plot([i, i], [0, yf_], color=C['orange'], lw=1.3)
    ax.text(i + 0.4, yf_, label.replace(' ABSD', ': ABSD raised'), ha='left', va='center', fontsize=8.6, fontweight='bold', color='white',
            bbox=dict(boxstyle='round,pad=0.3', fc=C['orange'], ec='none'))
ax.set_xticks([mk_months.get_loc(pd.Timestamp(f'{y_}-01-01')) for y_ in range(2021, 2026)]); ax.set_xticklabels([str(y_) for y_ in range(2021, 2026)])
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f'{v:,.0f}')); ax.set_ylabel('New sales + resales per month'); ax.set_xlim(-0.6, len(xp_) - 0.4)
ax.set_title('Monthly sales and cooling measures', loc='left', fontsize=10, color=C['ink'])
for j_, (lab_d, v_d, col_d) in enumerate([('12 months before\nApr 2023', MK_POLICY.iloc[0]['CCR share %'], C['grey']),
                                         ('12 months after', MK_POLICY.iloc[1]['CCR share %'], C['blue'])]):
    cx_ = j_ * 2.4
    ax2.add_patch(Wedge((cx_, 0), 1, 90 - 3.6 * v_d, 90, width=0.32, color=col_d))
    ax2.add_patch(Wedge((cx_, 0), 1, 90, 450 - 3.6 * v_d, width=0.32, color='#eceef0'))
    ax2.text(cx_, 0.05, f'{v_d:.1f}%', ha='center', va='center', fontsize=15, fontweight='bold', color=C['ink'])
    ax2.text(cx_, -0.28, 'core central', ha='center', va='center', fontsize=7.5, color=C['ink2'])
    ax2.text(cx_, -1.25, lab_d, ha='center', va='top', fontsize=8.6, color=C['ink'])
ax2.set_xlim(-1.3, 3.7); ax2.set_ylim(-1.9, 1.35); ax2.set_aspect('equal'); ax2.axis('off')
ax2.set_title('Share of new sales in the core central region', loc='left', fontsize=10, color=C['ink'])
finish(fig,'Volume changed around cooling measures; new-sale geography changed too',
       'Left: monthly new sales + resales; hatched = missing months. Right: core central region (CCR) share of new sales, 12 months before vs after the Apr 2023 ABSD increase. '
       'Descriptive only; no causal policy estimate.', MK_SOURCE+'; MAS/IRAS policy dates')
```

**New launches stayed much dearer.** Matched within district, quarter and size band, new launches sold about 40% above resales in 2020, 50% in 2024 and 46% in 2025. Step 4's 42% (2025) also matches on floor band. The share of new units under 600 sqft fell from about 20–22% (2020–21) to 9–15% (2022–25), but the median new unit did not shrink steadily. For an upgrader, resale stays the value route.

```python
def mk_matched_new(s):
    s=s[s.sale.isin(['New Sale','Resale'])].copy()
    s['key']=s.district.astype(str)+'|'+s.quarter+'|'+(s.area//200).astype(int).astype(str)
    matched=s.groupby('key').sale.transform('nunique').eq(2);s=s[matched]
    if len(s)<100:return None
    mk_x=(s.sale=='New Sale').astype(float);xd=mk_x-mk_x.groupby(s.key).transform('mean');yd=s.lpsf-s.groupby('key').lpsf.transform('mean')
    t,_=ols(yd,xd.to_frame('new'),['new'],cluster=s.project,const=False)
    z=t.loc['new'];return dict(year=int(s.year.iloc[0]),premium=z.pct,low=z.pct_lo,high=z.pct_hi,p=z.p,
                               new_matched=int((s.sale=='New Sale').sum()),resale_matched=int((s.sale=='Resale').sum()))
MK_PREM=pd.DataFrame([mk_matched_new(s) for y,s in mk_model.groupby('year')]).set_index('year')
MK_PREM_ND=pd.DataFrame([mk_matched_new(s[~s.exact_dup]) for y,s in mk_model.groupby('year')]).set_index('year')
mk_sizes=mk_new.groupby('year').area.agg(['median','size']);mk_sizes['under600 %']=mk_new.groupby('year').area.apply(lambda mk_x:100*(mk_x<600).mean())
display(MK_PREM.round(2));display(MK_PREM_ND[['premium','low','high']].round(2).rename(columns={'premium':'premium without exact duplicates'}));display(mk_sizes.round(2))
# Left: the matched new-launch premium each year as labelled bars with 95% CIs. Right: share of new units under 600 sqft, with the
# median new-unit size written under each bar (one axis per panel, no dual axes).
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(11.8, 4.6))
mk_years = MK_PREM.index.to_numpy(); xl_ = np.arange(len(mk_years))
cols_l = [SEQ_BLUE(0.35 + 0.6 * (v - MK_PREM.premium.min()) / max(MK_PREM.premium.max() - MK_PREM.premium.min(), 1)) for v in MK_PREM.premium]
ax.bar(xl_, MK_PREM.premium, color=[C['orange']] * len(xl_), width=0.62, alpha=0.9)
ax.errorbar(xl_, MK_PREM.premium, yerr=[MK_PREM.premium - MK_PREM.low, MK_PREM.high - MK_PREM.premium], fmt='none', ecolor=C['ink'], capsize=4, lw=1.3)
for x_, v_, h_ in zip(xl_, MK_PREM.premium, MK_PREM.high): ax.text(x_, h_ + 1.2, f'+{v_:.0f}%', ha='center', va='bottom', fontsize=11, fontweight='bold', color=C['ink'])
ax.scatter(xl_, MK_PREM_ND.premium, marker='_', s=260, color=C['ink'], lw=2, zorder=4, label='Without exact duplicates')
ax.set_xticks(xl_); ax.set_xticklabels([f'{y}' + ('*' if y == 2021 else '') for y in mk_years]); ax.set_ylim(0, MK_PREM.high.max() * 1.2)
ax.set_ylabel('New launch vs matched resale, psf (%)'); ax.legend(loc='upper left', fontsize=8.3); ax.grid(False, axis='x')
ax.set_title('New-launch premium after matching', loc='left', fontsize=10, color=C['ink'])
xs2_ = np.arange(len(mk_sizes))
ax2.bar(xs2_, mk_sizes['under600 %'], color=C['violet'], width=0.62, alpha=0.85)
for x_, v_, m_ in zip(xs2_, mk_sizes['under600 %'], mk_sizes['median']):
    ax2.text(x_, v_ + 0.6, f'{v_:.0f}%', ha='center', va='bottom', fontsize=11, fontweight='bold', color=C['ink'])
    ax2.text(x_, -1.2, f'median\n{m_:,.0f} sqft', ha='center', va='top', fontsize=7.8, color=C['ink2'])
ax2.set_xticks(xs2_); ax2.set_xticklabels([str(y) for y in mk_sizes.index]); ax2.tick_params(axis='x', pad=30)
ax2.set_ylim(0, mk_sizes['under600 %'].max() * 1.25); ax2.set_ylabel('New units under 600 sqft (%)'); ax2.grid(False, axis='x')
ax2.set_title('Small new units became rarer after 2021', loc='left', fontsize=10, color=C['ink'])
finish(fig,'New-sale gaps persist after coarse matching; launch unit size also changed',
       'Left: within district × 200-sqft area band × quarter matched OLS, project-clustered 95% CIs; dash = without exact duplicates. Right: share of new units under 600 sqft '
       '(median new-unit size below). *2021 covers nine recorded months, with only 145 total sales in July.', MK_SOURCE)
```

**The MRT pattern holds market-wide.** Across all homes, a ≤5-minute walk went with +5% to +11% higher psf in every year (+8.8% in 2025 here, vs +8.7% in Step 3b, which keeps sub sales). The family's segment shows a larger gap (Step 7).
- **New stations, island-wide:** pooling the TEL2, TEL3, TEL4 and Hume openings, homes that gained a station within 10 minutes showed no clear extra jump in the first six months (+0.1%, 95% CI −1.8% to +2.1%) compared with same-district homes that didn't.
- **After that:** later half-years drift to about −1% to −2%, with one significant bin. With several bins tested, that bin could be chance. This agrees with Step 7's east-coast result: no clear jump was detected at opening.

```python
MK_POSTAL_D = {**{s: 1 for s in range(1, 7)}, 7: 2, 8: 2, 14: 3, 15: 3, 16: 3, 9: 4, 10: 4, 11: 5, 12: 5, 13: 5, 17: 6, 18: 7, 19: 7, 20: 8, 21: 8,
            22: 9, 23: 9, **{s: 10 for s in range(24, 28)}, 28: 11, 29: 11, 30: 11, 31: 12, 32: 12, 33: 12, **{s: 13 for s in range(34, 38)},
            **{s: 14 for s in range(38, 42)}, **{s: 15 for s in range(42, 46)}, 46: 16, 47: 16, 48: 16, 49: 17, 50: 17, 81: 17, 51: 18, 52: 18,
            53: 19, 54: 19, 55: 19, 82: 19, 56: 20, 57: 20, 58: 21, 59: 21, **{s: 22 for s in range(60, 65)}, **{s: 23 for s in range(65, 69)},
            69: 24, 70: 24, 71: 24, 72: 25, 73: 25, 77: 26, 78: 26, 75: 27, 76: 27, 79: 28, 80: 28}
mk_coords=pd.read_csv('project_coords.csv');mk_coords['postal_d']=mk_coords.postal.apply(lambda p: MK_POSTAL_D.get(int(str(int(p)).zfill(6)[:2])) if pd.notna(p) else np.nan)
mk_coords=mk_coords[(mk_coords.method.eq('name+street')|(mk_coords.postal_d==mk_coords.district))&mk_coords.lat.notna()&mk_coords.lon.notna()].copy()
mk_stations=pd.DataFrame(json.load(open('sg_mrt.json'))['exits'],columns=['station','network','lat','lon'])
mk_stations=mk_stations[mk_stations.network.str.contains('singapore-mrt')].copy().reset_index(drop=True)
mk_openings=pd.read_csv('mrt_station_openings.csv',parse_dates=['opened']);mk_stations=mk_stations.merge(mk_openings,on='station',how='left')
assert mk_stations[mk_stations.stage.notna()].station.nunique()==mk_openings.station.nunique(), 'MRT opening names not all matched'
mk_stations['active_from']=(mk_stations.opened+pd.offsets.MonthBegin(1)).fillna(pd.Timestamp('2000-01-01'))
# Metres: local latitude/longitude approximation; 1.3 route factor, 80 m/min.
mk_pl=mk_coords.lat.to_numpy()[:,None];pn=mk_coords.lon.to_numpy()[:,None]
mk_sl=mk_stations.lat.astype(float).to_numpy()[None,:];sn=mk_stations.lon.astype(float).to_numpy()[None,:]
MK_WALK_MATRIX=np.hypot((mk_pl-mk_sl)*110570,(pn-sn)*111320*np.cos(np.radians((mk_pl+mk_sl)/2)))*1.3/80
mk_project_names=mk_coords.project.to_numpy(); project_pos={name:i for i,name in enumerate(mk_project_names)}
MK_MONTH_WALK={}
for month in mk_months:
    available=(mk_stations.active_from<=month).to_numpy()
    MK_MONTH_WALK[month]=pd.Series(MK_WALK_MATRIX[:,available].min(axis=1),index=mk_project_names)
mk_mrt=mk_model[mk_model.project.isin(project_pos)].copy()
mk_mrt['walk_min']=[MK_MONTH_WALK[m][p] for m,p in zip(mk_mrt.date_m,mk_mrt.project)]
mk_mrt['walk_band']=pd.cut(mk_mrt.walk_min,[0,5,10,15,np.inf],labels=['≤5','5–10','10–15','>15'],include_lowest=True)
print(f'Accepted {len(mk_coords):,} of {pd.read_csv("project_coords.csv").shape[0]:,} geocodes; {len(mk_stations):,} MRT or MRT/LRT interchange exits, excluding LRT-only; {len(mk_mrt):,} transactions located ({len(mk_mrt)/len(mk_model):.1%} of model-ready).')
def mk_mrt_year_fit(s):
    mk_x=pd.concat([mk_dummies(s.walk_band,'walk','>15'),mk_dummies(s.district,'D',9),mk_dummies(s.tenure,'ten','99-yr'),
                 mk_dummies(s.sale,'sale','Resale'),mk_dummies(s.type,'type','Condominium'),mk_dummies(s.month,'m',1),
                 s[['larea','floor','lease_used0']]],axis=1)
    t,h=ols(s.lpsf,mk_x,list(mk_x.columns),cluster=s.project)
    return pd.DataFrame({b:t.loc['walk_'+b,['pct','pct_lo','pct_hi','p']] for b in ['≤5','5–10','10–15']}).T
MK_MRT_YEAR=pd.concat({year:mk_mrt_year_fit(s) for year,s in mk_mrt.groupby('year')},names=['year','band'])
MK_MRT_5_ND=pd.DataFrame({year:mk_mrt_year_fit(s[~s.exact_dup]).loc['≤5',['pct','pct_lo','pct_hi']] for year,s in mk_mrt.groupby('year')}).T
display(MK_MRT_YEAR.round(2));display(MK_MRT_5_ND.round(2).add_suffix(' without exact duplicates'))
# Picture it: tiles for every year and walking band (all private homes), darker = bigger price gap vs over 15 minutes.
yrs_k = sorted(mk_mrt.year.unique())
def cell_k(rw, y):
    try: t_k = MK_MRT_YEAR.loc[(y, rw[0])]
    except KeyError: return None
    return (t_k.pct, t_k.pct_lo, t_k.pct_hi)
fig, ax = plt.subplots(figsize=(10.5, 4.8))
draw_heat(ax, [('≤5', 'Walk 5 min\nor less'), ('5–10', 'Walk\n5-10 min'), ('10–15', 'Walk\n10-15 min')], yrs_k, cell_k, {2020: '(Aug-Dec)', 2021: '(no Aug-Oct)'})
finish(fig,'Short MRT walks carry a persistent price association within districts',
       'Separate annual log-psf OLS for all private homes: walk band versus >15 min, district, sale, tenure, type, size, floor, lease used and month; project-clustered 95% CIs. Station availability begins the month after opening.',
       MK_SOURCE+'; project_coords.csv (OneMap); sg_mrt.json; mrt_station_openings.csv')
```

```python
# Stage exposure: project was >10 min from the pre-opening network and becomes ≤10 min with this stage.
MK_STAGES=['TEL2','TEL3','TEL4','NEL extension','DTL infill']
MK_EVENT_FRAMES=[];MK_EXPOSURE=[]
for stage in MK_STAGES:
    op=mk_openings[mk_openings.stage==stage].opened.min();om=op.to_period('M').to_timestamp()
    pre=(mk_stations.active_from<=om).to_numpy();post=pre|mk_stations.stage.eq(stage).to_numpy()
    prewalk=MK_WALK_MATRIX[:,pre].min(axis=1);postwalk=MK_WALK_MATRIX[:,post].min(axis=1)
    treated=(prewalk>10)&(postwalk<=10);treated_ds=set(mk_coords.loc[treated,'district'])
    control=(prewalk>10)&(postwalk>10)&mk_coords.district.isin(treated_ds).to_numpy()
    eligible=treated|control
    MK_EXPOSURE.append(dict(stage=stage,opening=str(op.date()),treated_geocoded=int(treated.sum()),controls_geocoded=int(control.sum())))
    if not treated.any() or not control.any():continue
    pm=pd.DataFrame({'project':mk_project_names[eligible],'treated':treated[eligible]})
    e=mk_mrt[(mk_mrt.sale=='Resale')&mk_mrt.project.isin(pm.project)].merge(pm,on='project')
    e['relative_month']=(e.year-om.year)*12+(e.month-om.month)
    e=e[(e.relative_month>=-24)&(e.relative_month<=24)&(e.relative_month!=0)].copy()
    e['rel_half']=np.where(e.relative_month<0,np.floor((e.relative_month+24)/6)-4,np.floor((e.relative_month-1)/6))
    e['rel_half']=e.rel_half.astype(int)
    e['stage']=stage;e['event_project']=stage+'|'+e.project;e['stage_quarter']=stage+'|'+e.quarter
    # Require sales on both sides in the same project.
    valid=e.groupby('event_project').relative_month.agg(lambda a:(a.lt(0).any() and a.gt(0).any()))
    e=e[e.event_project.isin(valid[valid].index)]
    if len(e):MK_EVENT_FRAMES.append(e)
MK_EXPOSURE=pd.DataFrame(MK_EXPOSURE).set_index('stage');display(MK_EXPOSURE)
if MK_EXPOSURE.loc['NEL extension','treated_geocoded']==0:print('Punggol Coast: no geocoded project crosses the ≤10-minute MRT threshold; its opening effect is not estimable here.')
mk_event=pd.concat(MK_EVENT_FRAMES,ignore_index=True) if MK_EVENT_FRAMES else pd.DataFrame()
if len(mk_event):
    EVENT_COUNTS=mk_event.groupby(['stage','treated']).agg(sales=('lpsf','size'),projects=('project','nunique')).rename(index={False:'controls',True:'treated'})
    display(EVENT_COUNTS)
    rels=[-4,-3,-2,0,1,2,3]
    XR=pd.DataFrame({f'treat_h{h}':((mk_event.treated)&(mk_event.rel_half==h)).astype(float) for h in rels})
    XQ=pd.get_dummies(mk_event.stage_quarter,prefix='sq',dtype=float).iloc[:,1:]
    X=pd.concat([XR,XQ,mk_event[['larea','floor']].reset_index(drop=True)],axis=1)
    Xd=X-X.groupby(mk_event.event_project.values).transform('mean')
    yd=mk_event.lpsf-mk_event.groupby('event_project').lpsf.transform('mean')
    ET,EH=ols(yd,Xd,list(X.columns),cluster=mk_event.project,const=False)
    EVENT_T=ET.loc[[f'treat_h{h}' for h in rels],['pct','pct_lo','pct_hi','p']].copy();EVENT_T.index=rels
    EVENT_T.loc[-1]=[0,0,0,np.nan];EVENT_T=EVENT_T.sort_index()
    EVENT_ND=mk_event[~mk_event.exact_dup].reset_index(drop=True).copy()
    XN=pd.DataFrame({f'treat_h{h}':((EVENT_ND.treated)&(EVENT_ND.rel_half==h)).astype(float) for h in rels})
    XN=pd.concat([XN,pd.get_dummies(EVENT_ND.stage_quarter,prefix='sq',dtype=float).iloc[:,1:].reset_index(drop=True),EVENT_ND[['larea','floor']].reset_index(drop=True)],axis=1)
    XNd=XN-XN.groupby(EVENT_ND.event_project.values).transform('mean')
    ynd=EVENT_ND.lpsf-EVENT_ND.groupby('event_project').lpsf.transform('mean')
    ET_ND,_=ols(ynd,XNd,list(XN.columns),cluster=EVENT_ND.project,const=False)
    print('Half-year labels: -4=-24 to -19 months, -1=-6 to -1 baseline, 0=+1 to +6; opening month omitted.')
    display(EVENT_T.round(2));display(ET_ND.loc[[f'treat_h{h}' for h in rels],['pct','pct_lo','pct_hi']].round(2).rename_axis('relative half-year, no exact duplicates'))
    print('Several post-opening half-years are tested; an isolated significant bin is suggestive and may reflect chance, anticipation or local shifts.')
    fig, ax = plt.subplots(figsize=(10.5, 4.6))
    draw_event(ax, EVENT_T.index, EVENT_T.pct, EVENT_T.pct_lo, EVENT_T.pct_hi,
               ['2 yrs\nbefore', '18 mo\nbefore', '12 mo\nbefore', '6 mo\nbefore', 'first\n6 mo', '7-12\nmo', '13-18\nmo', '19-24\nmo'], -0.5, -1, 'new stations open')
    ax.set_ylabel('Homes that gained a station minus\nsame-district homes (% psf)'); ax.set_xlabel('Half-years relative to the station opening')
    finish(fig,'Station opening contrasts remain observational and may move before trains arrive',
           'TEL2, TEL3, TEL4 and Hume openings pooled. Stacked project FE + event × calendar-quarter FE, size and floor controls; resale only, project-clustered 95% CIs. '
           'Treated projects cross from >10 to ≤10 estimated minutes; same-district controls stay >10. Opening month omitted; the 6 months before opening are the reference.',
           MK_SOURCE+'; project_coords.csv; sg_mrt.json; mrt_station_openings.csv')
else: print('No comparable resale project panels; event study not estimable.')
```

## Our recommendation

**Sell the flat first, then choose among three resale routes.** The graphic below puts price, upfront funds, payment and MRT walk on one page. Route B keeps them in the east at lower cost; C widens the search to the north-east towns; F buys more years of lease at a higher price.

**Plan the eventual sale too.** B and C cost less now but have less lease remaining at the end of a ten-year stay. F buys extra lease at a higher price. The Step 7 planning check shows how the loan balance and equity vary under illustrative price scenarios; historical lease-band performance does not settle the exit-value question.

**Inspect before choosing.** The 2025 comparable sales for B span Tampines and Pasir Ris; many are roughly 20 minutes from MRT. Route C spans four north-east towns. Check the actual school trip, station route, unit condition and available listings.

**Other routes:** the Katong/Marine Parade freehold resale (E) and Tampines/Pasir Ris new launch (A) are above this buyer's assumed budget. The older Bedok route (D) has too few comparable sales for a dependable price. Buying before selling would require 20% ABSD upfront, refundable only if all IRAS conditions are met. The 2026-Q3 figures below are market-index proxies, not observed project prices.

**Lucky numbers and the Ghost Month, 2020–2025.**
- **Price endings:** prices ending in 8 rose from 23% of resales (2020) to 29% (2025), against about 5% ending in 4. The 2024 tags are faded because those prices were rebuilt.
- **September vs other months (2022–2025 only):** resale prices were +2.1% (2022, 0 festival days in September) and +1.6% (2024, 2 festival days in September) against other months in years when the Hungry Ghost Month barely touched September, but only +0.3% (2023, 14 festival days in September) and +0.1% (2025, 21 festival days in September) when it did. This pattern is suggestive, but four years of month-only dates cannot identify a festival effect. The calendar month is only a rough stand-in for the festival.
The price-ending panel spans 2020–2025, but the September price comparison requires all twelve months. It excludes 2020 because the file starts in August and 2021 because August–October are missing. September is a calendar-month contrast, not a direct festival treatment; the festival can fall mostly or entirely in August.


```python
mk_res=mk_df[mk_df.sale=='Resale'].copy();mk_res['sig']=mk_res.price.round().astype(int).astype(str).str.rstrip('0')
mk_res['last_digit']=mk_res.sig.str[-1].astype(int)
MK_DIG=mk_res.groupby('year').apply(lambda s:pd.Series({'n':len(s),'ending 8 %':100*(s.last_digit==8).mean(),
                                                    'ending 4 %':100*(s.last_digit==4).mean()}))
MK_DIGND=mk_res[~mk_res.exact_dup].groupby('year').apply(lambda s:pd.Series({'ending 8 %':100*(s.last_digit==8).mean(),
                                                                    'ending 4 %':100*(s.last_digit==4).mean()}))
MK_GHOST_DATES=pd.DataFrame({'start':['2020-08-19','2021-08-08','2022-07-29','2023-08-16','2024-08-04','2025-08-23'],
                          'end':['2020-09-16','2021-09-06','2022-08-26','2023-09-14','2024-09-02','2025-09-21']},index=range(2020,2026))
MK_GHOST_DATES['September Ghost days']=[sum(d.month==9 for d in pd.date_range(a,b)) for a,b in zip(MK_GHOST_DATES.start,MK_GHOST_DATES.end)]
def mk_sep_fit(s):
    if set(range(1,13))-set(s.month.unique()):return None
    X=pd.concat([pd.get_dummies(s.month,prefix='m',drop_first=True,dtype=float),s[['larea','floor','lease_used0']],
                 pd.get_dummies(s.district,prefix='D',drop_first=True,dtype=float),
                 pd.get_dummies(s.tenure,prefix='ten',drop_first=True,dtype=float),
                 pd.get_dummies(s.type,prefix='type',drop_first=True,dtype=float)],axis=1)
    t,h=ols(s.lpsf,X,list(X.columns),cluster=s.project)
    w=np.zeros(len(h['names']))
    for m in range(2,13):w[h['names'].index('m_'+str(m))]=1.0 if m==9 else -1/11  # September minus mean of the other 11 months (January = 0)
    b=w@h['b'];se=np.sqrt(w@h['V']@w);tc=st.t.ppf(.975,h['G']-1);p=2*st.t.sf(abs(b/se),h['G']-1)
    return {'September %':100*np.expm1(b),'low':100*np.expm1(b-tc*se),'high':100*np.expm1(b+tc*se),'p':p,'n':len(s)}
MK_SEP=pd.DataFrame({y:mk_sep_fit(s) for y,s in mk_model[mk_model.sale=='Resale'].groupby('year') if mk_sep_fit(s) is not None}).T
MK_SEP_ND=pd.DataFrame({y:mk_sep_fit(s[~s.exact_dup]) for y,s in mk_model[mk_model.sale=='Resale'].groupby('year') if mk_sep_fit(s[~s.exact_dup]) is not None}).T
display(MK_DIG.round(2));display(MK_DIGND.round(2).add_suffix(' without exact duplicates'));display(MK_GHOST_DATES)
print('September is only a calendar-month proxy: in some years most or all Hungry Ghost days fall in August.')
display(MK_SEP.round(3));display(MK_SEP_ND[['September %']].round(3).rename(columns={'September %':'without exact duplicates %'}))
# Picture it. Left: price tags, one per year, as tall as the share of resale prices ending in 8 (orange) or 4 (purple).
# Right: a Ghost Month calendar. Purple bar = the festival's dates each year; shaded = September; the label is September's adjusted price gap.
from matplotlib.patches import Polygon as Poly_, Ellipse as Ell_, FancyBboxPatch as FBP_
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(12, 4.9), gridspec_kw=dict(width_ratios=[1, 1.25]))
def tag_(a_, x_, h_, w_, col_, alpha_=1):
    t_ = min(2.4, h_ * 0.45)
    a_.add_patch(Poly_([(x_ - w_ / 2, 0), (x_ + w_ / 2, 0), (x_ + w_ / 2, h_ - t_), (x_, h_), (x_ - w_ / 2, h_ - t_)], closed=True, color=col_, alpha=alpha_, lw=0))
    a_.add_patch(Ell_((x_, h_ - t_ * 0.72), w_ * 0.26, min(1.0, t_ * 0.4), color='white', lw=0))
for i_, (y_f, r_f) in enumerate(MK_DIG.iterrows()):
    al_ = 0.4 if y_f == 2024 else 1
    tag_(ax, i_ - 0.19, r_f['ending 8 %'], 0.34, C['orange'], al_); tag_(ax, i_ + 0.19, r_f['ending 4 %'], 0.34, C['violet'], al_)
    ax.text(i_ - 0.19, r_f['ending 8 %'] + 0.6, f"{r_f['ending 8 %']:.0f}%", ha='center', va='bottom', fontsize=9.5, fontweight='bold', color=C['ink'])
    ax.text(i_ + 0.19, r_f['ending 4 %'] + 0.6, f"{r_f['ending 4 %']:.0f}%", ha='center', va='bottom', fontsize=8.5, color=C['ink2'])
ax.text(list(MK_DIG.index).index(2024), MK_DIG.loc[2024, ['ending 8 %', 'ending 4 %']].max() + 4.5, 'rebuilt\nprices', ha='center', va='center', fontsize=7.5, color=C['grey'])
ax.set_xticks(range(len(MK_DIG))); ax.set_xticklabels(MK_DIG.index, fontsize=9.5); ax.set_ylim(0, MK_DIG['ending 8 %'].max() * 1.3); ax.set_xlim(-0.6, len(MK_DIG) - 0.4)
ax.set_ylabel('Share of resale prices (%)'); ax.grid(False, axis='x')
ax.add_patch(Rectangle((-0.5, MK_DIG['ending 8 %'].max() * 1.22), 0.18, 1.2, color=C['orange'], lw=0)); ax.text(-0.27, MK_DIG['ending 8 %'].max() * 1.24, 'price ends in 8', va='center', fontsize=8.3)
ax.add_patch(Rectangle((1.4, MK_DIG['ending 8 %'].max() * 1.22), 0.18, 1.2, color=C['violet'], lw=0)); ax.text(1.63, MK_DIG['ending 8 %'].max() * 1.24, 'price ends in 4', va='center', fontsize=8.3)
ax.set_title('Lucky price tags: last significant digit', loc='left', fontsize=10, color=C['ink'])
doy_ = lambda d: (pd.Timestamp(d).replace(year=2021) - pd.Timestamp('2021-07-01')).days
x0_, x1_ = doy_('2021-07-20'), doy_('2021-10-01'); s0_, s1_ = doy_('2021-09-01'), doy_('2021-10-01')
yrs_g = list(MK_GHOST_DATES.index)
ax2.axvspan(s0_, s1_, color=C['blue_l'], alpha=0.45, lw=0); ax2.text((s0_ + s1_) / 2, len(yrs_g) - 0.35, 'September', ha='center', va='bottom', fontsize=9, color=C['blue_d'], fontweight='bold')
ax2.text((doy_('2021-08-01') + s0_) / 2, len(yrs_g) - 0.35, 'August', ha='center', va='bottom', fontsize=9, color=C['ink2'])
for r_i, y_g in enumerate(yrs_g):
    yb_ = len(yrs_g) - 1 - r_i; g_ = MK_GHOST_DATES.loc[y_g]
    a_, b_ = doy_(g_.start), doy_(g_.end)
    ax2.add_patch(FBP_((a_, yb_ - 0.2), b_ - a_, 0.4, boxstyle='round,pad=0,rounding_size=0.18', color=C['violet'], alpha=0.85, lw=0, mutation_aspect=0.05))
    ax2.text(x0_ - 1, yb_, str(y_g), ha='right', va='center', fontsize=10, fontweight='bold', color=C['ink'])
    ax2.text(b_ + 1.2, yb_, f"{int(g_['September Ghost days'])} days in Sep", ha='left', va='center', fontsize=8, color=C['violet'])
    if y_g in MK_SEP.index:
        r_s = MK_SEP.loc[y_g]
        ax2.text(x1_ + 9, yb_ + 0.12, f"Sep {r_s['September %']:+.1f}%", ha='left', va='center', fontsize=10.5, fontweight='bold', color=C['ink'])
        ax2.text(x1_ + 9, yb_ - 0.24, f"CI {sg_(r_s.low, 1)} to {sg_(r_s.high, 1)}", ha='left', va='center', fontsize=7.5, color=C['grey'])
    else:
        ax2.text(x1_ + 9, yb_, 'partial year:\nnot tested', ha='left', va='center', fontsize=7.8, color=C['grey'])
ax2.set_xlim(x0_ - 9, x1_ + 28); ax2.set_ylim(-0.6, len(yrs_g) + 0.1); ax2.axis('off')
ax2.set_title('Ghost Month dates vs the September price gap', loc='left', fontsize=10, color=C['ink'])
finish(fig,'Lucky-price endings persist; month-only dates cannot test the Ghost Month itself',
       'Left: last significant digit of resale prices; 2024 faded because those prices were rebuilt. Right: grey = Hungry Ghost Festival dates; label = September vs the mean of the other 11 months '
       '(resale OLS with district, tenure, type, size, floor and lease controls; 95% project-clustered CIs).',
       MK_SOURCE+'; Ghost Month dates: Jade Almanac lunar-calendar table, Singapore Taoist Federation (2020)')
```

## Limits and evidence grades
| Grade | Finding |
|---|---|
| **Robust associations** | District, sale type, lease and floor coefficients are stable in the stated models: the regression explains 82% in-sample and 77% on unseen projects, VIFs are under 3.2, and removing duplicates changes the key 2025 effects by under 0.3 percentage points; removing the 10 most influential projects changes the adjusted new-sale gap from 26.0% to 24.4%. Also robust: the payment vs stress-test arithmetic, and the sell-first vs buy-first cash rules |
| **Conditional association** | New launches sell about 42% higher than matched resales; matching can't rule out project quality. Family-size resales within an estimated 5-minute MRT walk sell 17.1% higher per sqft than those over 15 minutes away after controls (95% CI 12.7–21.6%, p < 0.001); projects may differ in unobserved ways |
| **Suggestive** | The lease-curve step; the inspection list (partly persistent out of sample); the 2027 top-up illustration and the 2026-Q3 market-index price proxy; route D (2 sales) |
| **Five-year checks** | The MRT gap appears in every year 2020–25 (about +7% to +17%); shorter-lease projects showed no clear evidence of losing ground to longer-lease ones; station openings added no detectable jump (east coast +0.1%, 95% CI −1.7% to +1.8%); the model's pricing rules are fairly stable and its next-year transfer test reaches R² 0.72–0.81 only after the realised URA market shift is supplied |
| **Not identifiable** | A single yearly lease-depreciation rate (age, year and lease start move together); whether newness itself *causes* the new-launch gap; building age (not in the data); whether rates or cooling measures *cause* price changes |

**Assumptions to flag:**
- **The profile is illustrative:** income, flat value, CPF refund, costs and ages are assumptions.
- **Income measure:** gross employment income excluding employer CPF, the income banks count.
- **Prices and rates:** observed comparables are 2025 sales. The 2026-Q3 figures are market-index proxies, not current project transactions. Payments at 1.6% are a separate fixed planning assumption; December 2025 SORA plus 0.30 points is about 1.49%. The 2025 rate illustration assumes SORA + 0.30%.
- **Not in the CSV:** bedroom counts and building age. Project locations were added from OneMap (99% matched); walking times use straight-line distance × 1.3, not routed paths. MRT/LRT interchanges count as MRT, LRT-only stops do not, and 2026 CCL6 stations are excluded from the 2025 analysis.
- **Duplicates:** the data includes 1,442 exact duplicate rows; results barely change without them.
- **Five-year data:** August–October 2021 are missing; an even-multiplicity March–July 2021 block was halved; sub sales exist only from 2024 and are excluded from cross-year comparisons; 2024 prices are rebuilt from whole-dollar psf. The top-up history moves 2025 route medians back with an index, so it shows the trend for homes like these, not observed past prices of these exact units. Cooling-measure and station-opening comparisons are observational.
- **Map precision:** the district heat maps smooth between broad reference locations. Planning-area outlines are not postal-district boundaries. The MRT price map uses actual project points but still relies on geocoding and a distance-to-walk approximation.
- **Buying costs and reserves:** the budget assumes no other debts and full use of the stated eligible CPF. Purchase legal fees, renovation, ongoing maintenance, property tax and temporary housing are not priced. Confirm loan approval, CPF availability and actual walking routes before choosing a unit.

- **Ten-year ownership scenarios:** the price changes and constant mortgage rate are assumptions, not forecasts. Equity is sale price minus outstanding loan before selling costs and CPF refunds; it is not spendable cash or an investment return. The 2026 starting price is a market-index proxy. Remaining lease is calendar arithmetic from the stated 2025 bands.


## Reading summary
The guide answers five connected questions:

1. **Order:** selling first releases the family’s funds and avoids paying additional buyer’s stamp duty upfront on a second property.
2. **Budget:** the flat-sale proceeds constrain this profile more than its income does. Keep a reserve below the ceiling.
3. **Location:** compare neighbourhoods and actual station routes, not just a broad district median.
4. **Home:** compare observed resale options by upfront cost, lease left now and after ten years, space and walk to MRT. Treat model discounts as an inspection list.
5. **Timing:** use the historical evidence to understand risk. Neither the trend forecast nor the station-opening comparisons promise a buying window.

The appendices retain model diagnostics, sensitivity results and numerical tables. They support the guide rather than form a separate buyer story.


## Appendix A · How the price model was built and checked
- **Simple vs multiple regression:** one variable at a time overstates the floor effect threefold.
- **Building the model step by step:** district alone explains 43% (adjusted R²); the full model explains 82%. Curves and segment interactions add 1.4 points.
- **Held-out projects:** with whole projects held out (5 folds), the model explains 77%. The 80% band covers 82% of held-out resales but only 65% of new sales.
- **Diagnostics:**
  - VIFs are all under 3.2.
  - The largest Cook's distance is 0.02. Many sales exceed the common 4/n screen (0.0002), so we refitted without the 10 most influential projects; the largest shift was the adjusted new-sale gap, 26.0% to 24.4%.
  - Residuals have heavy tails, and resale prices are about twice as noisy as developer prices.
- **Project bootstrap and duplicates:** 1,000 resamples of whole projects give intervals close to the clustered standard errors. Dropping the 1,442 exact duplicate rows changes the key effects by under 0.3 points.

```python
# Simple vs multiple regression: what goes wrong if we look at one variable at a time?
fb_ = df.groupby('floor').agg(lpsf=('lpsf', 'mean'), n=('lpsf', 'size')); fb_ = fb_[fb_.n >= 30]
slr_floor = np.polyfit(df.floor, df.lpsf, 1); slr_r2 = np.corrcoef(df.floor, df.lpsf)[0, 1] ** 2
ab_ = pd.cut(df.area, [0, 500, 650, 800, 950, 1100, 1300, 1600, 2000, 3000]); sb_ = df.groupby(ab_).agg(la=('larea', 'mean'), lpsf=('lpsf', 'mean'), n=('lpsf', 'size'))
slr_size = np.polyfit(df.larea, df.lpsf, 1)
fig, axs = plt.subplots(1, 2, figsize=(11, 4.2))
for ax, xv, yv_, sl, mlr, xlab, ttl in [
        (axs[0], fb_.index.values, fb_.lpsf.values, slr_floor[0], hed.loc['floor', 'coef'], 'Floor (lower bound of the 5-floor band)', 'Floor'),
        (axs[1], np.exp(sb_.la.values), sb_.lpsf.values, slr_size[0], hed.loc['larea', 'coef'], 'Unit size (sqft, log scale)', 'Size')]:
    xx = np.log(xv) if ttl == 'Size' else xv; x0 = np.average(xx, weights=None); y0 = np.mean(yv_)
    ax.scatter(xv, np.exp(yv_), s=40, color=C['grey'], alpha=0.8, zorder=3, label='Average price per sqft (all 2025 sales)')
    grid_ = np.linspace(xx.min(), xx.max(), 50); gx_ = np.exp(grid_) if ttl == 'Size' else grid_
    ax.plot(gx_, np.exp(y0 + sl * (grid_ - x0)), color=C['orange'], lw=2.4, label='Simple regression (one variable)')
    ax.plot(gx_, np.exp(y0 + mlr * (grid_ - x0)), color=C['blue'], lw=2.4, label='Multiple regression (location etc. adjusted for)')
    if ttl == 'Size': ax.set_xscale('log'); ax.set_xticks([400, 600, 1000, 1500, 2500]); ax.set_xticklabels(['400', '600', '1,000', '1,500', '2,500']); ax.xaxis.set_minor_formatter(plt.NullFormatter())
    ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f'S${v:,.0f}')); ax.set_xlabel(xlab); ax.set_ylabel('Price per sqft')
axs[0].set_title(f'Floor: simple {np.expm1(slr_floor[0]) * 100:+.2f}% per floor vs multiple {hed.loc["floor", "pct"]:+.2f}%', loc='left', fontsize=10, color=C['ink'])
axs[1].set_title(f'Size: simple {slr_size[0] * np.log(1.1) * 100:+.1f}% psf per 10% bigger vs multiple {hed.loc["larea", "coef"] * np.log(1.1) * 100:+.1f}%', loc='left', fontsize=10, color=C['ink'])
axs[0].legend(loc='upper left', fontsize=8.3)
finish(fig, f'Looking at one variable at a time triples the floor premium',
       f'Tall towers and small units cluster in prime, new projects, so simple regressions (R² {slr_r2:.2f} for floor) mix their effect with location. '
       'The multiple regression separates them. Lines pass through the average point; compare the slopes.')
```

```python
# Building the model up step by step (adjusted R² and a partial F-test), plus curves and interactions.
def fit_r2(Xm):
    A = np.column_stack([np.ones(len(Xm)), np.asarray(Xm, float)]); b_ = np.linalg.lstsq(A, df.lpsf.values, rcond=None)[0]
    rss = ((df.lpsf.values - A @ b_) ** 2).sum(); return rss, np.linalg.matrix_rank(A)
tss = ((df.lpsf - df.lpsf.mean()) ** 2).sum(); n_ = len(df)
dist_ = dummies(df.district, 'D', 9); sale_ = pd.concat([dummies(df.sale, 'sale', 'Resale'), dummies(df.tenure, 'ten', '99-yr'), dummies(df.type, 'type', 'Condominium')], axis=1)
num_ = df[['larea', 'floor', 'lease_used']]; mon_ = dummies(df.month, 'm', 1)
band_ = pd.cut(99 - df.lease_used, [0, 70, 75, 80, 85, 89, 93, 100], right=False).astype(str).where(df.tenure == '99-yr', 'not 99-yr')
segd_ = dummies(df.segment, 'seg', 'OCR')
inter_ = pd.concat([segd_.mul(df.floor, axis=0).add_prefix('floorX'), segd_.mul((df.sale == 'New Sale').astype(float), axis=0).add_prefix('newX'),
                    pd.get_dummies(band_, prefix='band', dtype=float).drop(columns='band_not 99-yr')], axis=1)
ladder = {'Floor only (simple regression)': df[['floor']], 'District': dist_, '+ sale type, tenure, property type': pd.concat([dist_, sale_], axis=1),
          '+ size, floor, lease, month (our model)': pd.concat([dist_, sale_, num_, mon_], axis=1),
          '+ lease bands and segment interactions': pd.concat([dist_, sale_, num_, mon_, inter_], axis=1)}
res_ = {k: fit_r2(v) for k, v in ladder.items()}
adj = pd.Series({k: 1 - (rss / (n_ - p_)) / (tss / (n_ - 1)) for k, (rss, p_) in res_.items()})
(r0, p0), (r1, p1) = res_['+ size, floor, lease, month (our model)'], res_['+ lease bands and segment interactions']
F_ = ((r0 - r1) / (p1 - p0)) / (r1 / (n_ - p1)); pF = st.f.sf(F_, p1 - p0, n_ - p1)
fig, ax = plt.subplots(figsize=(10, 3.9)); yy = np.arange(len(adj))[::-1]
ax.barh(yy, adj.values, color=[C['grey_l'], C['blue_l'], '#B2C9E4', C['blue'], C['blue_d']], height=0.58)
for y_, v in zip(yy, adj.values): ax.text(v + 0.008, y_, f'{v:.3f}', va='center', fontsize=9.5, color=C['ink'], fontweight='bold' if y_ == 1 else 'normal')
ax.set_yticks(yy); ax.set_yticklabels(adj.index, fontsize=9.5, color=C['ink']); ax.grid(False, axis='y'); ax.set_xlim(0, 1)
ax.set_xlabel('Adjusted R² (share of variation in log price per sqft explained, penalised for extra terms)')
finish(fig, f'Location explains most of the price; adding curves and interactions adds only {adj.iloc[-1] - adj.iloc[-2]:.3f}',
       f"From 'District' down, each row adds terms to the one above (the floor-only row is for comparison). The last step passes a classical partial F-test "
       f"(F = {F_:.0f} on {p1 - p0} and {n_ - p1:,} df, p < 0.001; weaker once sales are clustered by project) but adds little, so the simpler model is kept.")
```

```python
# Model diagnostics: residuals vs fitted, normality, spread by sale type, and influence (leverage, studentised residuals, Cook's distance).
fig, axs = plt.subplots(2, 2, figsize=(11, 7.4))
ax = axs[0, 0]; ax.hexbin(np.exp(fitted), e_, gridsize=60, cmap=LinearSegmentedColormap.from_list('h', ['#ffffff', '#B2C9E4', C['blue_d']]), mincnt=1, bins='log', xscale='log')
qb = pd.qcut(fitted, 12); sd_ = pd.Series(e_).groupby(qb).std(); mx = pd.Series(np.exp(fitted)).groupby(qb).median()
ax.plot(mx, 2 * sd_.values, color=C['orange'], lw=2); ax.plot(mx, -2 * sd_.values, color=C['orange'], lw=2, label='±2 SD of residuals')
ax.set_xticks([800, 1200, 2000, 3000, 5000]); ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f'{v:,.0f}')); ax.xaxis.set_minor_formatter(plt.NullFormatter())
ax.axhline(0, color=C['ink2'], lw=0.8); ax.set_xlabel('Fitted price per sqft (S$, log scale)'); ax.set_ylabel('Residual (log points)'); ax.legend(loc='lower right', fontsize=8.3)
ax.set_title('Residuals vs fitted: wider for cheaper homes', loc='left', fontsize=10, color=C['ink'])
ax = axs[0, 1]; smp = np.sort(np.random.default_rng(0).choice(rstud, 3000, replace=False)); th = st.norm.ppf((np.arange(1, 3001) - 0.5) / 3000)
ax.scatter(th, smp, s=8, color=C['blue'], alpha=0.5, lw=0); ax.plot([-4, 4], [-4, 4], color=C['ink2'], lw=1, ls=(0, (4, 3)))
ax.set_xlabel('Normal quantiles'); ax.set_ylabel('Studentised residuals'); ax.set_xlim(-4, 4); ax.set_ylim(-11, 9)
ax.set_title(f'Normal Q-Q: heavier tails than normal (kurtosis {pd.Series(e_).kurt() + 3:.1f})', loc='left', fontsize=10, color=C['ink'])
ax = axs[1, 0]; grp_ = ['New Sale', 'Sub Sale', 'Resale']
bp = ax.boxplot([e_[df.sale.values == g_] for g_ in grp_], vert=False, showfliers=False, widths=0.5, patch_artist=True,
                medianprops=dict(color=C['ink'], lw=1.6), whiskerprops=dict(color=C['ink2']), capprops=dict(color=C['ink2']))
for patch, col in zip(bp['boxes'], [C['blue_l'], '#B2C9E4', C['orange_l']]): patch.set_facecolor(col); patch.set_edgecolor(C['ink2'])
ax.set_yticks([1, 2, 3]); ax.set_yticklabels(grp_)
for i, g_ in enumerate(grp_): ax.text(0.47, i + 1, f'SD {SD_BY_SALE[g_]:.3f}', va='center', fontsize=9, color=C['ink'])
ax.axvline(0, color=C['ink2'], lw=0.8); ax.set_xlim(-0.5, 0.62); ax.set_xlabel('Residual (log points)'); ax.grid(False, axis='y')
ax.set_title('Resale prices are about twice as hard to predict', loc='left', fontsize=10, color=C['ink'])
ax = axs[1, 1]; big = cook > 0.01
ax.scatter(lev[~big], rstud[~big], s=6, color=C['grey'], alpha=0.35, lw=0); ax.scatter(lev[big], rstud[big], s=9, color=C['orange'], alpha=0.6, lw=0, label="Cook's distance > 0.01")
ax.axhline(3, color=C['ink2'], lw=0.8, ls=(0, (4, 3))); ax.axhline(-3, color=C['ink2'], lw=0.8, ls=(0, (4, 3))); ax.axvline(3 * Xa.shape[1] / Xa.shape[0], color=C['ink2'], lw=0.8, ls=(0, (4, 3)))
ax.set_xlabel('Leverage'); ax.set_ylabel('Studentised residual'); ax.legend(loc='lower right', fontsize=8.3)
ax.set_title(f"Influence screen: largest Cook's distance {cook.max():.3f}", loc='left', fontsize=10, color=C['ink'])
finish(fig, 'Model checks show uneven residual spread and heavy tails',
       f"VIF: size {VIF['larea']:.1f}, floor {VIF['floor']:.1f}, lease {VIF['lease_used']:.1f}; highest of all {len(VIF_ALL)} predictors {VIF_MAX:.1f} (below the common screening threshold of 5; "
       "this does not mean there is no variance inflation). Heavy tails and uneven spread are why we use clustered and bootstrap intervals.")
```

```python
# Project bootstrap: resample whole projects (sales in one building are related) 1,000 times, refit, and take percentile intervals.
codes_, uniq_ = pd.factorize(df.project); G_ = len(uniq_); rng_b = np.random.default_rng(0); yl = df.lpsf.values
keys = ['lease_used', 'floor', 'sale_New Sale']; ki = [H['names'].index(k) for k in keys]; boots = []
for _ in range(1000):
    wcount = np.bincount(rng_b.integers(0, G_, G_), minlength=G_)[codes_].astype(float); Xw = Xa * wcount[:, None]
    boots.append(np.linalg.lstsq(Xw.T @ Xa, Xw.T @ yl, rcond=None)[0][ki])
boots = np.array(boots); scale = np.array([1, 10, 1])
BOOT = pd.DataFrame({'estimate': np.expm1(H['b'][ki] * scale) * 100, 'bootstrap low': np.percentile(np.expm1(boots * scale) * 100, 2.5, axis=0),
                     'bootstrap high': np.percentile(np.expm1(boots * scale) * 100, 97.5, axis=0),
                     'clustered-SE low': [np.expm1((hed.loc[k, 'coef'] - 1.96 * hed.loc[k, 'se']) * s_) * 100 for k, s_ in zip(keys, scale)],
                     'clustered-SE high': [np.expm1((hed.loc[k, 'coef'] + 1.96 * hed.loc[k, 'se']) * s_) * 100 for k, s_ in zip(keys, scale)]},
                    index=['Each year of lease used', 'Every 10 floors higher', 'New launch vs resale'])
fig, axs = plt.subplots(1, 3, figsize=(11, 3.6))
for ax, (lab_, rw), col in zip(axs, BOOT.iterrows(), range(3)):
    vals = np.expm1(boots[:, col] * scale[col]) * 100
    ax.hist(vals, bins=40, color=C['blue'], alpha=0.75, edgecolor=C['bg'], linewidth=0.4)
    ax.axvspan(rw['bootstrap low'], rw['bootstrap high'], color=C['blue'], alpha=0.08, lw=0)
    for v in (rw['bootstrap low'], rw['bootstrap high']): ax.axvline(v, color=C['blue_d'], lw=1.5)
    for v in (rw['clustered-SE low'], rw['clustered-SE high']): ax.axvline(v, color=C['orange'], lw=1.5, ls=(0, (4, 2)))
    ax.set_title(f"{lab_}: {rw.estimate:+.2f}%", loc='left', fontsize=10, color=C['ink']); ax.set_yticks([]); ax.grid(False, axis='y')
    ax.set_xlabel(f"95%: {rw['bootstrap low']:+.2f} to {rw['bootstrap high']:+.2f}%", fontsize=8.8)
finish(fig, 'Resampling whole projects 1,000 times gives the same answers',
       f'Bootstrap distributions of three key effects ({G_:,} projects resampled with replacement). Solid blue lines = bootstrap 95% interval; dashed red = clustered-SE 95% CI. They are close, so the intervals don\'t rest on assuming normal errors.')
```

```python
# Robustness: do the headline effects survive dropping exact duplicate rows (mostly new-sale bulk entries) or the 10 most
# influential projects? And how well does the model do on projects it has never seen?
dup_ = df.duplicated(subset=['project', 'street', 'type', 'district', 'segment', 'tenure', 'commence', 'sale', 'price', 'area', 'level', 'date'], keep='first').values
b_nd = np.linalg.lstsq(Xa[~dup_], yl_all[~dup_], rcond=None)[0]
keys_ = [('lease_used', 'Each year of lease used', 1), ('floor', 'Every 10 floors higher', 10), ('sale_New Sale', 'New launch vs resale', 1)]
ROBUST = pd.DataFrame({lab: {'All sales': np.expm1(H['b'][H['names'].index(k)] * sc) * 100,
                             f'Without {dup_.sum():,} exact duplicates': np.expm1(b_nd[H['names'].index(k)] * sc) * 100,
                             'Without 10 most influential projects': np.expm1(b_k[H['names'].index(k)] * sc) * 100} for k, lab, sc in keys_}).T
fig, axs = plt.subplots(1, 3, figsize=(11, 3.2))
for ax, (lab, rw) in zip(axs, ROBUST.iterrows()):
    yy = np.arange(len(rw))[::-1]; ax.scatter(rw.values, yy, s=70, color=[C['blue'], C['orange'], C['violet']], zorder=3, edgecolor=C['bg'], linewidth=1.5)
    for y_, v in zip(yy, rw.values): ax.text(v, y_ + 0.25, f'{v:+.2f}%', ha='center', fontsize=8.8, color=C['ink'])
    ax.set_title(lab, loc='left', fontsize=10, color=C['ink']); ax.set_yticks(yy); ax.set_yticklabels(rw.index if ax is axs[0] else [''] * len(rw), fontsize=8.8)
    ax.grid(False, axis='y'); ax.set_ylim(-0.6, len(rw) - 0.2); pad = max(abs(rw.values).max() * 0.15, 0.3); ax.set_xlim(rw.min() - pad, rw.max() + pad)
finish(fig, 'Results are similar after removing duplicates or the 10 most influential projects',
       f'On projects it has never seen (5-fold, whole projects held out) the model explains {R2_CV:.0%} of price variation (82% in-sample). '
       f'The 80% band covers {COVER["Resale"]:.0%} of held-out resales and {COVER["New Sale"]:.0%} of held-out new sales.')
```

**Five years of pricing rules and a forward test.** The checks above use 2025 only. With five years we can ask whether the model's rules stay put, and whether its price rules transfer to the next year after supplying the realised URA market shift.
- **The rules barely move:** ten floors add 4–6% every year, each extra year of lease used is associated with about 1.1–1.2% lower price per sqft, and the new-sale gap stays at 26–32%.
- **Forward transfer test:** fitted on one year and then given the realised URA index change for the next, the model explains 72–81% of next-year price variation, with a typical single-unit error of 8–10%. This market shift was unknown at forecast time. Its 80% band held 73–81% of next-year sales, so single-unit bands are a little too narrow in some years.
- **Scope:** these are three year-to-next-year checks (2022–23, 2023–24 and 2024–25); 2021 is excluded because of missing months.

```python
def mk_year_design(s,columns=None):
    mk_x=pd.concat([pd.get_dummies(s.district,prefix='D',dtype=float),pd.get_dummies(s.tenure,prefix='ten',dtype=float),
                 pd.get_dummies(s.type,prefix='type',dtype=float),pd.get_dummies(s.sale,prefix='sale',dtype=float),
                 s[['larea','floor','lease_used0']]],axis=1)
    mk_x=mk_x.drop(columns=['D_9','ten_99-yr','type_Condominium','sale_Resale'],errors='ignore')
    return mk_x if columns is None else mk_x.reindex(columns=columns,fill_value=0.0)
MK_STAB=[];YEAR_MODELS={}
for year,s in mk_model.groupby('year'):
    X=mk_year_design(s);t,h=ols(s.lpsf,X,list(X.columns),cluster=s.project);YEAR_MODELS[year]=(t,h,list(X.columns))
    for key,scale,name in [('floor',10,'Ten floors'),('lease_used0',1,'Year lease used'),('larea',np.log(1.1),'10% larger area'),('sale_New Sale',1,'New sale')]:
        if key in t.index:
            b,se=t.loc[key,['coef','se']]*scale;tc=st.t.ppf(.975,h['G']-1)
            MK_STAB.append(dict(year=year,effect=name,pct=100*np.expm1(b),low=100*np.expm1(b-tc*se),high=100*np.expm1(b+tc*se),p=t.loc[key,'p']))
MK_STAB=pd.DataFrame(MK_STAB)
def mk_backtest_year(a,b):
    train=mk_model[mk_model.year==a];test=mk_model[mk_model.year==b];t,h,cols=YEAR_MODELS[a]
    annual_q=mk_q.ppi.groupby(mk_q.index.str[:4]).mean()
    drift=np.log(annual_q[str(b)]/annual_q[str(a)])
    pred=np.c_[np.ones(len(test)),mk_year_design(test,cols).values]@h['b']+drift
    resid=test.lpsf.values-pred;train_resid=h['e'];ql,qh=np.quantile(train_resid,[.1,.9])
    return dict(train=a,test=b,n=len(test),R2=1-np.sum(resid**2)/np.sum((test.lpsf-test.lpsf.mean())**2),
                MdAPE=100*np.median(np.abs(np.exp(pred-test.lpsf.values)-1)),coverage=100*np.mean((resid>=ql)&(resid<=qh)))
MK_OOTT=pd.DataFrame([mk_backtest_year(a,a+1) for a in [2022,2023,2024]]).set_index('test')
display(MK_STAB.round(3));display(MK_OOTT.round(3));print('Annual coefficient duplicate sensitivity:')
def mk_year_fit(s):
    X=mk_year_design(s);t,_=ols(s.lpsf,X,list(X.columns),cluster=s.project);return t
MK_STAB_ND_ROWS=[]
for y,s in mk_model[~mk_model.exact_dup].groupby('year'):
    t=mk_year_fit(s)
    for key,scale,name in [('floor',10,'Ten floors'),('lease_used0',1,'Year lease used'),('larea',np.log(1.1),'10% larger area'),('sale_New Sale',1,'New sale')]:
        MK_STAB_ND_ROWS.append({'year':y,'effect':name,'without exact duplicates %':100*np.expm1(t.loc[key,'coef']*scale)})
MK_STAB_ND=pd.DataFrame(MK_STAB_ND_ROWS)
display(MK_STAB[['year','effect','pct']].merge(MK_STAB_ND,on=['year','effect']).round(2))
def mk_backtest_no_duplicates(a,b):
    train=mk_model[(mk_model.year==a)&~mk_model.exact_dup];test=mk_model[(mk_model.year==b)&~mk_model.exact_dup]
    X=mk_year_design(train);t,h=ols(train.lpsf,X,list(X.columns),cluster=train.project)
    annual_q=mk_q.ppi.groupby(mk_q.index.str[:4]).mean();drift=np.log(annual_q[str(b)]/annual_q[str(a)])
    pred=np.c_[np.ones(len(test)),mk_year_design(test,list(X.columns)).values]@h['b']+drift
    resid=test.lpsf.values-pred;ql,qh=np.quantile(h['e'],[.1,.9])
    return {'test':b,'R2 without exact duplicates':1-np.sum(resid**2)/np.sum((test.lpsf-test.lpsf.mean())**2),
            'MdAPE without exact duplicates':100*np.median(np.abs(np.exp(pred-test.lpsf.values)-1)),
            'coverage without exact duplicates':100*np.mean((resid>=ql)&(resid<=qh))}
MK_OOTT_ND=pd.DataFrame([mk_backtest_no_duplicates(a,a+1) for a in [2022,2023,2024]]).set_index('test')
display(MK_OOTT_ND.round(3))
# One small panel per pricing rule (same years on every panel), each labelled with its range; then the transfer test as three cards.
from matplotlib.patches import FancyBboxPatch as FBP_
fig, AX_ = plt.subplot_mosaic([['a', 'b', 'c', 'd'], ['e', 'f', 'g', 'h']], figsize=(13.4, 5.8), height_ratios=[1.25, 0.75])
yrs_s = sorted(mk_model.year.unique())
for j_, (k, col, unit) in enumerate([('Ten floors', C['blue'], '%'), ('Year lease used', C['orange'], '%'), ('10% larger area', C['violet'], '%'), ('New sale', C['aqua'], '%')]):
    a_ = AX_['abcd'[j_]]; d_ = MK_STAB[MK_STAB.effect == k].set_index('year').reindex(yrs_s)
    a_.fill_between(yrs_s, d_.low, d_.high, color=col, alpha=0.15, lw=0); a_.plot(yrs_s, d_.pct, color=col, lw=2.2, marker='o', ms=5)
    a_.axhline(0, color=C['ink2'], lw=0.8); a_.set_xticks(yrs_s); a_.set_xticklabels([f"'{str(y)[2:]}" for y in yrs_s], fontsize=8.5)
    lo_, hi_ = d_.low.min(), d_.high.max(); pad_ = (hi_ - lo_) * 0.25 + 0.3; a_.set_ylim(min(lo_, 0) - pad_, max(hi_, 0) + pad_)
    a_.set_title(f'{k}\n{d_.pct.min():+.1f}% to {d_.pct.max():+.1f}% across years', loc='left', fontsize=9.5, color=C['ink'], fontweight='bold')
for j_, (yr_t, r_t) in enumerate(MK_OOTT.iterrows()):
    a_ = AX_['efg'[j_]]; a_.axis('off'); a_.set_xlim(0, 1); a_.set_ylim(0, 1)
    a_.add_patch(FBP_((0.02, 0.05), 0.96, 0.9, boxstyle='round,pad=0,rounding_size=0.06', facecolor='#f4f6f8', edgecolor='none'))
    a_.text(0.08, 0.78, f'{int(r_t.train)} model → {yr_t} prices', fontsize=9.3, fontweight='bold', color=C['ink'], va='center')
    a_.text(0.08, 0.48, f'R² {r_t.R2:.2f}', fontsize=13, fontweight='bold', color=C['blue_d'], va='center')
    a_.text(0.52, 0.48, f'error {r_t.MdAPE:.0f}%', fontsize=11, color=C['ink'], va='center')
    a_.text(0.08, 0.2, f'80% band held {r_t.coverage:.0f}% of sales', fontsize=8.3, color=C['ink2'], va='center')
a_ = AX_['h']; a_.axis('off')
a_.text(0.02, 0.5, 'Transfer test: fit on one year,\ngiven the actual URA market\nchange, test on the next.\nError = typical single-unit miss.', fontsize=8.3, color=C['ink2'], va='center')
finish(fig,'Key associations vary by year; next-year errors remain material',
       'Top: separate annual hedonic OLS; shaded = project-clustered 95% CIs. Bottom: train in year t, test in t+1 after the observed URA annual-index drift; '
       '2021 is excluded from the consecutive-year test because three months are missing.',
       MK_SOURCE+'; URA non-landed index')
```

## Appendix B · Why this buyer is a useful case study
The HDB upgrader connects the strongest parts of this dataset: selling an existing flat, financing the next home, resale versus new sale, lease length, geography and MRT access. The table below records the original judgement-based scoring of candidate buyers. It explains the choice; it is not a statistical result or advice for those other buyers.

This notebook follows only the upgrader family. A score of 1 means central to the decision, 0.5 means relevant and 0 means not used. The scoring is subjective and is not evidence that one buyer is more important than another.


```python
# Appendix B: how the buyer was chosen. Score each candidate on which findings and datasets bear on their decision
# (1 = central to their decision, 0.5 = relevant, 0 = not relevant).
evidence = {
    'Location premium (CCR/RCR/OCR, km to CBD)':        [1, 1, 1, 1, 0.5],
    'New launch vs resale gap':                          [1, 1, 1, 0.5, 0.5],
    'Lease used (cross-project association)':                [1, 1, 1, 0.5, 1],
    'Floor premium':                                     [0.5, 0.5, 1, 0.5, 0.5],
    'Unit size and price per sqft':                      [0.5, 1, 1, 0.5, 1],
    '2025 timing: launch calendar, mix illusion':       [1, 1, 0.5, 0.5, 0.5],
    'Interest rates (SORA) and monthly payments':        [1, 1, 1, 0.5, 0],
    'Loan rules (TDSR, stress test, LTV)':               [1, 1, 1, 1, 0.5],
    "Buyer's stamp duty (BSD)":                          [1, 1, 1, 1, 1],
    'ABSD (20% second home / 60% foreigner / refund)':   [0, 1, 1, 1, 0.5],
    "Seller's stamp duty (SSD, 4-year hold)":           [0, 0.5, 1, 0.5, 0],
    'HDB resale prices (selling the flat)':              [0, 1, 0, 0, 1],
    'Rents and rental yield':                            [0, 0, 1, 0.5, 0],
    'Price history and 2026 forecast':                   [1, 1, 1, 1, 0.5],
    'Model residuals (projects to inspect)':        [1, 1, 1, 0.5, 0.5],
    'Feng shui (lucky numbers, Ghost Month)':            [1, 1, 0.5, 0.5, 1],
}
BUYERS = ['First-time couple', 'HDB upgrader family', 'Investor (2nd property)', 'Foreign buyer', 'Downsizing retiree']
score = pd.DataFrame(evidence, index=BUYERS).T
total = score.sum().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(10.5, 6.6)); ax.grid(False)
order = list(total.index); mat = score[order].values
ax.imshow(mat, cmap=LinearSegmentedColormap.from_list('s', ['#f7f8f9', '#B2C9E4', C['blue']]), vmin=0, vmax=1, aspect='auto')
for i in range(mat.shape[0]):
    for j in range(mat.shape[1]):
        ax.text(j, i, {1: 'core', 0.5: 'some', 0: ''}[mat[i, j]], ha='center', va='center', fontsize=8, color=C['bg'] if mat[i, j] == 1 else C['ink'])
ax.set_xticks(range(len(order))); ax.set_xticklabels([f'{b}\nscore {total[b]:.1f}' for b in order], fontsize=8.8, color=C['ink'])
ax.set_yticks(range(len(score))); ax.set_yticklabels(score.index, fontsize=8.6, color=C['ink']); ax.xaxis.tick_top()
for s_ in ax.spines.values(): s_.set_visible(False)
finish(fig, 'Why choose an HDB upgrader? The profile connects many parts of the evidence', 'Judgement-based relevance scores, not statistical evidence. The family connects a flat sale with the next private-home purchase.', source=None)
```

```python
# Appendix C: the numbers behind the charts, for checking.
display(pd.Series({'Budget ceiling': sgm(BUDGET), 'Binding limit': BINDING, 'Net proceeds after costs': sgm(NET_PROCEEDS), 'Price cap from funds alone': sgm(CASH_CAP), 'Price cap from the loan alone (TDSR at 4%)': sgm(LOAN_CAP),
                   'Loan tenure (years)': TENURE, 'ABSD if they buy before selling the flat (20%, refundable)': sgd(0.2 * BUDGET)}, name='budget'))
show = opt.copy()
for c_ in ['model price', 'real median', 'cash + CPF needed', 'monthly @1.6%', 'PI low', 'PI high', '2026-Q3 index proxy', 'cash + CPF at proxy']: show[c_] = show[c_].map(lambda v: sgd(v) if pd.notna(v) else '-')
display(show)
display(pd.DataFrame({'2025 median': opt['real median'], '2026-Q3 market-index proxy': opt['2026-Q3 index proxy'],
                      'upfront at proxy': opt['cash + CPF at proxy'], 'within budget at proxy': opt['fits at proxy']}).round(0))
def fmt_p(v):
    return '<0.0001' if pd.notna(v) and v < 0.0001 else (f'{v:.4f}' if pd.notna(v) else '—')
if HAS_MRT:
    mrt_view = MRT_TAB.round(3).rename(columns={'pct': 'premium %', 'lo': '95% low', 'hi': '95% high', 'p': 'p-value', 'n': 'sales in band'})
    mrt_view['p-value'] = MRT_TAB['p'].map(fmt_p)
    display(mrt_view)
    display(MRT_LRT_SENS.round(2))
    route_areas = pd.DataFrame([{'route': code, 'district': opt.loc[code, 'district'],
                                 'planning-area mix of comparable sales': '; '.join(f'{town} {n}' for town, n in co.town.value_counts().items())}
                                for code, co in comps.items()]).set_index('route')
    display(route_areas)
    route_town_prices = pd.concat([co.groupby('town').agg(sales=('price', 'size'), median_price=('price', 'median'),
                                          median_MRT_walk=('walk', 'median')).assign(route=code)
                                   for code, co in comps.items() if len(co)]).reset_index().set_index(['route', 'town']).sort_index()
    display(route_town_prices.round(1))
    display(vf[['deals', 'median_price', 'vs_model', 'H1', 'H2']].round(2).rename(columns={'vs_model': 'whole-year gap %', 'H1': 'Jan-Jun gap %', 'H2': 'Jul-Dec gap %'}))
display(causal.round(1))
display(pd.DataFrame({k: {m: f'{v[0]:+.2f}% ({v[1]:+.2f} to {v[2]:+.2f})' for m, v in dd.items()} for k, dd in seg_res.items()}))
display(SEG_P.rename_axis('effect').apply(lambda col: col.map(fmt_p)).rename(columns=lambda col: f'{col} p-value'))
hed_view = hed.loc[[v for v in ['lease_used', 'floor', 'sale_New Sale', 'larea'] if v in hed.index],
                   ['pct', 'pct_lo', 'pct_hi', 'p']].copy()
hed_view['p'] = hed_view['p'].map(fmt_p)
display(hed_view.round(4).rename(columns={'pct': 'effect %', 'pct_lo': '95% low', 'pct_hi': '95% high', 'p': 'p-value'}))
display(curve.round(1)); display(rmse.round(2).rename('typical forecast error, % points per quarter'))
display(adj.round(3).rename('adjusted R2')); display(BOOT.round(2)); display(pd.Series({**{f'VIF {k}': round(v, 2) for k, v in VIF.items()}, 'max Cook distance': round(cook.max(), 4)}, name='diagnostics'))
fc.round(1)
# Put all new-sale estimates in one labelled reference table.
PREMIUM_REFERENCE=causal[['gap %','low','high']].rename(columns={'gap %':'gap %','low':'95% low','high':'95% high'}).copy()
PREMIUM_REFERENCE['comparison']='2025, 99-year units; see Step 4'
PREMIUM_REFERENCE.loc['Broad hedonic model (all tenures)',:]=[hed.loc['sale_New Sale','pct'],hed.loc['sale_New Sale','pct_lo'],hed.loc['sale_New Sale','pct_hi'],'2025 market-wide coefficient, not the matched gap']
display(PREMIUM_REFERENCE.round(2))
for year,r in MK_PREM.iterrows():
    PREMIUM_REFERENCE.loc[f'{int(year)} annual matching, no floor bands',:]=[r.premium,r.low,r.high,'All tenures; district × quarter × size-band matched OLS']
for _,r in MK_STAB[MK_STAB.effect=='New sale'].iterrows():
    PREMIUM_REFERENCE.loc[f'{int(r.year)} annual hedonic coefficient',:]=[r.pct,r.low,r.high,'All tenures; separate annual hedonic model']
print('Central reference: the first five rows and annual rows use different samples, controls and weighting. They are not equivalent estimates.')
display(PREMIUM_REFERENCE.round(2))
display(SCREEN_CHECK.round(2));display(EXIT_LEASE);display(EXIT_DETAIL.round(0))
```
